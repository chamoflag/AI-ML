```python
"""
====================================================================
DDF-v2 (ported): Impact-Concentration Detector for Two-Leg Trading
====================================================================
Context: SEBI's order against Jane Street describes a smaller, illiquid
leg (cash/futures, "Leg1") traded at an outsized share of its own
market's volume, while a much larger position in a far deeper market
("Leg2", options) monetizes the resulting price move. Structurally this
is a "manufactured concentration" problem -- nearly identical in shape
to the spatial hotspot detection problem the original DDF was built
for. Swap "geography" for "instrument x time" and most of the toolkit
carries over directly.

Original DDF pieces being ported here:
  - Moran's I         -> global autocorrelation (smooth, broad pattern)
  - Geary's C         -> local-sensitive autocorrelation (this carried
                          the real signal in the original work; tuned
                          for SHARP local discontinuities, which is
                          exactly the shape of the JS pattern)
  - HH_frac (HHI)      -> concentration index. Was a zero-variance
                          suppressor in the spatial context -- re-tested
                          fresh here since concentration IS the headline
                          regulatory signal in this domain, not a
                          side variable.
  - Cauchy combination  -> combine correlated per-unit p-values into one
                          global statistic without needing to know the
                          dependence structure (Liu & Xie, 2020, JASA).
  - Permutation null    -> empirical null via resampling.
  - Block structure     -> spatial blocks -> temporal/instrument blocks,
                          to avoid leaking correlated-neighbor info.

Design stance: this is a runnable v0, not a finished instrument. Where
a real-data swap-in point exists, it's flagged with TODO. Favoring
"works end-to-end today on synthetic data" over "theoretically perfect
but nothing to run."
====================================================================
"""

import numpy as np
import pandas as pd
import warnings
warnings.filterwarnings("ignore")

# ====================================================================
# SECTION 0: CONFIG
# ====================================================================

RNG_SEED = 42
N_PERMUTATIONS = 1000   # null-distribution resample count. 1000 gives
                         # p-value resolution down to ~0.001, runs in a
                         # few seconds on the demo-sized grid below.
                         # Bump up for production, not for a first pass.


# ====================================================================
# SECTION 1: SYNTHETIC DATA GENERATOR
# (stand-in until real NSE bhavcopy + options chain data is wired in)
# ====================================================================

def generate_synthetic_market_data(
    n_stocks=12,                  # BANKNIFTY has 12 constituents
    n_buckets=75,                 # 375-min trading day / 5-min buckets
    baseline_participation=0.04,  # "our" share of market volume, normal
    spike_participation=0.30,     # "our" share of market volume, spike
    spike_stocks_frac=0.85,       # SEBI: spike hit 11/12 stocks
    spike_start_frac=0.80,        # spike begins ~80% through the day
    impact_coef=0.00018,          # price impact per unit signed participation
    noise_vol=0.0009,             # baseline per-bucket return noise
    inject_pattern=True,          # False -> "clean" day, for false-positive check
    seed=RNG_SEED,
):
    """
    Long-format DataFrame, one row per (stock, time_bucket):
        stock, bucket, mkt_volume, our_volume, participation, ret

    If inject_pattern=True: a subset of stocks gets an artificial
    participation + price-impact spike in the last fraction of the day,
    with the impact sign flipping mid-spike (buy-then-sell, mirroring
    the SEBI Patch I -> reversal structure). This is the ground-truth
    positive case to validate the detector against.

    TODO (real data): replace this whole function with a loader that
    pulls NSE bhavcopy (cash) + futures + options chain volumes per
    5-min bucket, and computes participation as (entity volume / total
    market volume) per stock per bucket. Everything downstream is
    agnostic to where this DataFrame comes from.
    """
    rng = np.random.default_rng(seed)
    stocks = [f"STOCK_{i:02d}" for i in range(n_stocks)]
    spike_stock_set = set(rng.choice(n_stocks, size=int(n_stocks * spike_stocks_frac), replace=False))
    spike_start = int(n_buckets * spike_start_frac)

    rows = []
    for si, stock in enumerate(stocks):
        # crude U-shaped intraday volume profile (high at open/close) --
        # this is a real, innocent pattern every detector needs to
        # survive without false-flagging.
        t = np.linspace(0, 1, n_buckets)
        mkt_vol_profile = 1.0 + 1.5 * ((t - 0.5) ** 2) * 4
        mkt_vol_profile *= rng.uniform(0.8, 1.2, size=n_buckets)
        mkt_vol = (mkt_vol_profile * rng.uniform(8e4, 2e5)).astype(int)

        participation = baseline_participation * rng.uniform(0.6, 1.4, size=n_buckets)
        ret = rng.normal(0, noise_vol, size=n_buckets)

        if inject_pattern and si in spike_stock_set:
            participation[spike_start:] = spike_participation * rng.uniform(0.7, 1.3, size=n_buckets - spike_start)
            mid = spike_start + (n_buckets - spike_start) // 2
            signed_part = participation.copy()
            signed_part[spike_start:mid] *= 1     # buying -> price up
            signed_part[mid:] *= -1               # selling -> price down
            ret[spike_start:] += (
                impact_coef * signed_part[spike_start:] * rng.uniform(0.8, 1.2, size=n_buckets - spike_start)
            )

        our_vol = (participation * mkt_vol).astype(int)
        for b in range(n_buckets):
            rows.append((stock, b, int(mkt_vol[b]), int(our_vol[b]), float(ret[b])))

    df = pd.DataFrame(rows, columns=["stock", "bucket", "mkt_volume", "our_volume", "ret"])
    df["participation"] = df["our_volume"] / df["mkt_volume"]
    return df


# ====================================================================
# SECTION 2: SPATIAL WEIGHTS -> INSTRUMENT x TIME WEIGHTS
# ====================================================================

def build_weight_matrix(df, stock_col="stock", bucket_col="bucket",
                         temporal_window=1, cross_stock_corr=None):
    """
    Original DDF: W was spatial contiguity/distance. Here, "neighbors" are:
      - same stock, adjacent time buckets (temporal contiguity)
      - different stocks, same/adjacent bucket, weighted by return
        correlation (cross-sectional contiguity)

    This matters: BANKNIFTY constituents move together, so treating them
    as independent units understates how unusual simultaneous
    concentration across 11/12 stocks really is -- same reason adjacent
    geographic units can't be treated as independent draws.

    Row-standardized (rows sum to 1) per the standard Moran's I/Geary's C
    convention.

    NOTE: this is an O(n^2) python double loop -- fine for the demo grid
    (12 stocks x 75 buckets = 900 rows -> ~5-10 sec, run once per
    dataset). Vectorize before scaling to 1-min buckets x a larger
    instrument universe.
    """
    units = df[[stock_col, bucket_col]].reset_index(drop=True)
    n = len(units)
    stock_idx = {s: i for i, s in enumerate(units[stock_col].unique())}
    bucket_arr = units[bucket_col].values
    stock_arr = units[stock_col].map(stock_idx).values

    if cross_stock_corr is None:
        # Fallback prior: BANKNIFTY constituents are genuinely correlated.
        # Swap in realized correlation from real data when available.
        n_stocks_ = len(stock_idx)
        cross_stock_corr = np.full((n_stocks_, n_stocks_), 0.45)
        np.fill_diagonal(cross_stock_corr, 1.0)

    W = np.zeros((n, n))
    for i in range(n):
        for j in range(n):
            if i == j:
                continue
            same_stock = stock_arr[i] == stock_arr[j]
            dt = abs(int(bucket_arr[i]) - int(bucket_arr[j]))
            if same_stock and 0 < dt <= temporal_window:
                W[i, j] = 1.0 / dt
            elif (not same_stock) and dt <= temporal_window:
                corr = cross_stock_corr[stock_arr[i], stock_arr[j]]
                if corr > 0:
                    W[i, j] = corr / (dt + 1)

    row_sums = W.sum(axis=1, keepdims=True)
    row_sums[row_sums == 0] = 1.0
    return W / row_sums


# ====================================================================
# SECTION 3: MORAN'S I / GEARY'S C ON THE PARTICIPATION GRID
# ====================================================================

def morans_i(x, W):
    """
    I = (n/S0) * (z'Wz) / (z'z),  z = x - mean(x), S0 = sum of all weights.
    High I = participation spikes are clustered (a few stocks + a few
    buckets carry it all) rather than spread diffusely.
    """
    x = np.asarray(x, dtype=float)
    n = len(x)
    z = x - x.mean()
    S0 = W.sum()
    den = z @ z
    if den == 0 or S0 == 0:
        return 0.0
    return (n / S0) * (z @ W @ z) / den


def gearys_c(x, W):
    """
    C = [(n-1)/(2*S0)] * sum_ij w_ij(x_i-x_j)^2 / sum_i z_i^2
    Ranges ~0 (sharp local clustering) to ~2 (negative autocorrelation),
    1 = no autocorrelation.

    This carried the real signal in the original DDF work -- more
    sensitive to sharp LOCAL discontinuities than Moran's I (which is
    better tuned to smooth global trends). The JS pattern is exactly the
    sharp-local case: most of the day is quiet, a narrow window in
    specific stocks is not.
    """
    x = np.asarray(x, dtype=float)
    n = len(x)
    z = x - x.mean()
    S0 = W.sum()
    if S0 == 0 or (z @ z) == 0:
        return 1.0
    diff_sq = (x[:, None] - x[None, :]) ** 2
    num = (W * diff_sq).sum()
    return ((n - 1) / (2 * S0)) * (num / (z @ z))


# ====================================================================
# SECTION 4: HH_frac (HHI) -- CONCENTRATION INDEX
# ====================================================================

def hhi_concentration(volumes):
    """
    HHI = sum(share_i^2), share_i = v_i / sum(v). Ranges 1/n (even
    spread) to 1 (one unit has everything).

    NOTE: in the original DDF, HH_frac was a zero-variance suppressor
    once combined with Geary's C -- didn't carry independent signal
    there. Don't assume that holds here: SEBI's own order leans on raw
    participation % as primary evidence, so HHI may carry real,
    non-redundant signal in THIS domain. Test it fresh -- see
    validate_hhi_redundancy() below, don't just inherit the old verdict.
    """
    v = np.asarray(volumes, dtype=float)
    total = v.sum()
    if total == 0:
        return 0.0
    shares = v / total
    return float((shares ** 2).sum())


def hhi_by_bucket(df):
    """HHI of our volume across stocks, per time bucket. High = at that
    moment, our participation is dominated by one or two stocks rather
    than spread across the index."""
    return df.groupby("bucket")["our_volume"].apply(hhi_concentration)


def validate_hhi_redundancy(df):
    """
    Crude but honest redundancy check: correlate bucket-level HHI against
    a bucket-level "local clustering" proxy (cross-stock variance of
    participation within the bucket). High correlation (~>0.8) ->
    redundant with the autocorrelation stats, drop/downweight. Low
    correlation -> independent signal, keep it.
    """
    hhi_vals = hhi_by_bucket(df).values
    local_var = df.groupby("bucket")["participation"].var().values
    corr = np.corrcoef(hhi_vals, local_var)[0, 1]
    verdict = "REDUNDANT (drop/downweight)" if abs(corr) > 0.8 else "INDEPENDENT SIGNAL (keep)"
    print(f"HHI vs local-concentration-proxy correlation: {corr:.3f} -> {verdict}")
    return corr


# ====================================================================
# SECTION 5: PERMUTATION NULL DISTRIBUTION
# ====================================================================

def permutation_test(df, stat_fn, W, value_col="participation",
                      n_perm=N_PERMUTATIONS, block_col="bucket",
                      tail="greater", seed=RNG_SEED):
    """
    BLOCK PERMUTATION, NOT A FULL SHUFFLE -- this is the important
    design choice. A naive full shuffle across the whole grid would
    destroy the (totally innocent) U-shaped intraday volume profile,
    making the null too easy to beat -> false positives.

    Instead, shuffle WITHIN time buckets only: permute which stock gets
    which participation value, independently per bucket. This preserves
    the legitimate intraday shape while destroying any STOCK-SPECIFIC
    concentration -- exactly the null we want ("is this stock unusually
    singled out, given the time-of-day baseline").

    Swap block_col="stock" to test the complementary null: "is this time
    window unusual, given each stock's normal level."

    tail="greater" for stats where HIGH = clustered (Moran's I, HHI).
    tail="less" for stats where LOW = clustered (Geary's C).
    """
    rng = np.random.default_rng(seed)
    observed = stat_fn(df[value_col].values, W)

    df_work = df.copy().reset_index(drop=True)
    null_stats = np.empty(n_perm)
    for p in range(n_perm):
        df_perm = df_work.copy()
        df_perm[value_col] = (
            df_perm.groupby(block_col)[value_col].transform(lambda s: rng.permutation(s.values))
        )
        null_stats[p] = stat_fn(df_perm[value_col].values, W)

    if tail == "greater":
        p_value = (np.sum(null_stats >= observed) + 1) / (n_perm + 1)
    else:
        p_value = (np.sum(null_stats <= observed) + 1) / (n_perm + 1)
    return observed, null_stats, p_value


# ====================================================================
# SECTION 6: CAUCHY COMBINATION TEST (Liu & Xie, 2020, JASA)
# ====================================================================

def cauchy_combination_test(p_values, weights=None):
    """
    T = sum_i w_i * tan((0.5 - p_i) * pi),  sum(w_i) = 1
    combined p-value = 0.5 - arctan(T) / pi

    Why this over Fisher's or Bonferroni: BANKNIFTY constituents are
    correlated, so per-stock p-values below are NOT independent.
    Fisher's assumes independence and is miscalibrated here; Bonferroni
    is valid under dependence but needlessly conservative, and will bury
    a real signal spread thinly across several correlated stocks. CCT
    stays valid under arbitrary, unknown dependence (Liu & Xie 2020) --
    exactly our situation, and you don't have to estimate the
    correlation structure to use it.
    """
    p = np.clip(np.asarray(p_values, dtype=float), 1e-10, 1 - 1e-10)
    n = len(p)
    if weights is None:
        weights = np.full(n, 1.0 / n)
    else:
        weights = np.asarray(weights, dtype=float)
        weights = weights / weights.sum()

    T = np.sum(weights * np.tan((0.5 - p) * np.pi))
    combined_p = 0.5 - np.arctan(T) / np.pi
    return T, combined_p


def per_stock_local_pvalues(df, n_perm=500, seed=RNG_SEED):
    """
    Per stock: is THIS stock's own participation-vs-time profile more
    temporally concentrated (higher HHI across buckets) than expected if
    its participation were randomly reshuffled across the day? One
    p-value per stock, feeding into the Cauchy combination above.
    """
    rng = np.random.default_rng(seed)
    results = {}
    for stock, g in df.groupby("stock"):
        vals = g.sort_values("bucket")["participation"].values
        observed_hhi = hhi_concentration(vals)
        null_hhi = np.array([hhi_concentration(rng.permutation(vals)) for _ in range(n_perm)])
        results[stock] = (np.sum(null_hhi >= observed_hhi) + 1) / (n_perm + 1)
    return pd.Series(results, name="local_p_value")


# ====================================================================
# SECTION 7: TWO-LEG PnL ASYMMETRY + SCALE-INVARIANCE TEST
# ====================================================================
# Directly operationalizes the forum comment's heuristic: simulate a
# strategy trading Leg1 (illiquid) against Leg2 (liquid, ~10x notional,
# ~100x deeper market per SEBI's numbers), then ask whether risk-
# adjusted PnL gets BETTER or WORSE as size shrinks toward zero.
# Genuine arb -> better (impact drag vanishes, real edge remains).
# Impact-harvesting -> worse / collapses (no footprint, no profit).

def simulate_two_leg_strategy(
    size_multiplier=1.0,
    leg1_mkt_depth=1.0,
    leg1_base_notional=1.0,
    impact_coef=0.02,
    true_edge=0.0,     # >0 simulates a strategy with SOME genuine
                        # informational edge, to see how the curve's
                        # shape changes vs. the pure-impact case
    n_days=300,
    seed=RNG_SEED,
):
    """
    Deliberately simple toy model -- the point is the SHAPE of the
    scale curve, not calibrated PnL magnitudes.

    Leg1 impact cost is convex in participation (standard Almgren-Chriss
    intuition: cost grows faster than linearly as you trade a bigger
    fraction of available liquidity) -> impact_coef * participation^1.5.
    Leg2 monetizes the SAME price move Leg1 generated, scaled by Leg2's
    much larger notional, in a market 100x deeper so Leg2 barely moves
    its own price.
    """
    rng = np.random.default_rng(seed)
    leg1_notional = leg1_base_notional * size_multiplier
    leg2_notional = leg1_notional * 10  # SEBI ratio: ~7-10x

    participation1 = leg1_notional / leg1_mkt_depth
    participation2 = leg2_notional / (leg1_mkt_depth * 100)

    leg1_impact_cost = impact_coef * participation1 ** 1.5
    price_move_generated = impact_coef * participation1

    leg1_mean = -leg1_impact_cost + true_edge * leg1_notional
    leg1_pnl = rng.normal(leg1_mean, abs(leg1_mean) * 0.3 + 1e-6, n_days)

    leg2_mean = price_move_generated * leg2_notional * (1 - participation2)
    leg2_pnl = rng.normal(leg2_mean, abs(leg2_mean) * 0.15 + 1e-6, n_days)

    total_pnl = leg1_pnl + leg2_pnl
    sharpe = total_pnl.mean() / (total_pnl.std() + 1e-9)
    return {
        "size_multiplier": size_multiplier,
        "leg1_mean_pnl": leg1_pnl.mean(),
        "leg2_mean_pnl": leg2_pnl.mean(),
        "total_mean_pnl": total_pnl.mean(),
        "sharpe": sharpe,
    }


def scale_invariance_curve(size_grid=None, **kwargs):
    """Sharpe-vs-size curve across a grid of size multipliers."""
    if size_grid is None:
        size_grid = np.array([0.01, 0.05, 0.1, 0.25, 0.5, 1.0, 2.0, 5.0, 10.0])
    rows = [simulate_two_leg_strategy(size_multiplier=s, **kwargs) for s in size_grid]
    return pd.DataFrame(rows)


# ====================================================================
# SECTION 8: DEMO RUN
# ====================================================================

if __name__ == "__main__":
    print("=" * 70)
    print("CASE A: clean day (no injected pattern) -- false-positive check")
    print("=" * 70)
    df_clean = generate_synthetic_market_data(inject_pattern=False)
    W_clean = build_weight_matrix(df_clean)
    print(f"Moran's I  = {morans_i(df_clean['participation'].values, W_clean):.4f}  (expect ~0)")
    print(f"Geary's C  = {gearys_c(df_clean['participation'].values, W_clean):.4f}  (expect ~1)")
    _, _, p_clean = permutation_test(df_clean, gearys_c, W_clean, tail="less")
    print(f"Geary's C permutation p-value = {p_clean:.4f}  (expect NOT significant)")

    print("\n" + "=" * 70)
    print("CASE B: injected JS-style concentration pattern")
    print("=" * 70)
    df_spike = generate_synthetic_market_data(inject_pattern=True)
    W_spike = build_weight_matrix(df_spike)
    print(f"Moran's I  = {morans_i(df_spike['participation'].values, W_spike):.4f}")
    print(f"Geary's C  = {gearys_c(df_spike['participation'].values, W_spike):.4f}  (expect well below 1)")
    obs_g, null_g, p_spike = permutation_test(df_spike, gearys_c, W_spike, tail="less")
    print(f"Geary's C permutation p-value = {p_spike:.4f}  (observed {obs_g:.4f} vs null mean {null_g.mean():.4f})")

    print("\n--- Per-stock local concentration test + Cauchy combination ---")
    local_p = per_stock_local_pvalues(df_spike)
    print(local_p.sort_values())
    T_cct, combined_p = cauchy_combination_test(local_p.values)
    print(f"\nCauchy combination statistic = {T_cct:.4f}")
    print(f"Combined global p-value       = {combined_p:.6f}")

    print("\n--- HHI redundancy check ---")
    validate_hhi_redundancy(df_spike)

    print("\n" + "=" * 70)
    print("Scale-invariance test (the forum '/100' heuristic, formalized)")
    print("=" * 70)
    print("\n[A] Pure impact-harvesting (true_edge=0):")
    curve_impact = scale_invariance_curve(true_edge=0.0)
    print(curve_impact.to_string(index=False, float_format=lambda x: f"{x:,.4f}"))
    print("\n[B] Some genuine edge present (true_edge=0.005):")
    curve_edge = scale_invariance_curve(true_edge=0.005)
    print(curve_edge.to_string(index=False, float_format=lambda x: f"{x:,.4f}"))

    print("\nReading it: in [A], Sharpe should COLLAPSE as size shrinks toward 0 --")
    print("no footprint, no profit. In [B], Sharpe should hold up or improve at small")
    print("size, because the edge term doesn't depend on your own impact. That contrast")
    print("IS the diagnostic the forum comment described, made falsifiable.")
```

```python
"""
====================================================================
DDF-v2 (ported): Impact-Concentration Detector for Two-Leg Trading
====================================================================
Context: SEBI's order against Jane Street describes a smaller, illiquid
leg (cash/futures, "Leg1") traded at an outsized share of its own
market's volume, while a much larger position in a far deeper market
("Leg2", options) monetizes the resulting price move. Structurally this
is a "manufactured concentration" problem -- nearly identical in shape
to the spatial hotspot detection problem the original DDF was built
for. Swap "geography" for "instrument x time" and most of the toolkit
carries over directly.

Original DDF pieces being ported here:
  - Moran's I         -> global autocorrelation (smooth, broad pattern)
  - Geary's C         -> local-sensitive autocorrelation (this carried
                          the real signal in the original work; tuned
                          for SHARP local discontinuities, which is
                          exactly the shape of the JS pattern)
  - HH_frac (HHI)      -> concentration index. Was a zero-variance
                          suppressor in the spatial context -- re-tested
                          fresh here since concentration IS the headline
                          regulatory signal in this domain, not a
                          side variable.
  - Cauchy combination  -> combine correlated per-unit p-values into one
                          global statistic without needing to know the
                          dependence structure (Liu & Xie, 2020, JASA).
  - Permutation null    -> empirical null via resampling.
  - Block structure     -> spatial blocks -> temporal/instrument blocks,
                          to avoid leaking correlated-neighbor info.

Design stance: this is a runnable v0, not a finished instrument. Where
a real-data swap-in point exists, it's flagged with TODO. Favoring
"works end-to-end today on synthetic data" over "theoretically perfect
but nothing to run."

----------------------------------------------------------------------
CHANGE LOG (post code-review fixes)
----------------------------------------------------------------------
1. Section 6 / per_stock_local_pvalues: HHI is invariant to permutation
   order (it only depends on sum(x) and sum(x^2), neither of which
   changes when you reorder x). Using it as a permutation-test statistic
   on a single shuffled vector measures floating-point summation noise,
   not signal. Replaced with max_consecutive_elevated_run(), a statistic
   that genuinely depends on the SEQUENCE of values, so shuffling
   actually destroys it and the resulting null is mathematically valid.

2. Section 5 / permutation_test: default block_col flipped from
   "bucket" to "stock". Blocking by bucket only shuffles which stock
   owns which value WITHIN a bucket -- it can't remove a pattern where
   most stocks are simultaneously elevated in that bucket, because the
   set of values in the bucket is unchanged either way. That's exactly
   the JS-style pattern (a systemic, simultaneous, multi-asset spike),
   so the old default null inherited the anomaly it was meant to test
   against. Blocking by stock instead preserves each stock's own daily
   behavior but shatters cross-sectional alignment in time, which is
   the right null for "did these stocks spike together by chance."
   NOTE: this assumes participation itself has no real intraday
   seasonality (true for the synthetic generator below, since the
   U-shape lives in raw volume, not the participation ratio). On real
   data, check for genuine time-of-day structure in participation
   before relying on this block; detrend first if it's there.
====================================================================
"""

import numpy as np
import pandas as pd
import warnings
warnings.filterwarnings("ignore")

# ====================================================================
# SECTION 0: CONFIG
# ====================================================================

RNG_SEED = 42
N_PERMUTATIONS = 1000   # null-distribution resample count. 1000 gives
                         # p-value resolution down to ~0.001, runs in a
                         # few seconds on the demo-sized grid below.
                         # Bump up for production, not for a first pass.


# ====================================================================
# SECTION 1: SYNTHETIC DATA GENERATOR
# (stand-in until real NSE bhavcopy + options chain data is wired in)
# ====================================================================

def generate_synthetic_market_data(
    n_stocks=12,                  # BANKNIFTY has 12 constituents
    n_buckets=75,                 # 375-min trading day / 5-min buckets
    baseline_participation=0.04,  # "our" share of market volume, normal
    spike_participation=0.30,     # "our" share of market volume, spike
    spike_stocks_frac=0.85,       # SEBI: spike hit 11/12 stocks
    spike_start_frac=0.80,        # spike begins ~80% through the day
    impact_coef=0.00018,          # price impact per unit signed participation
    noise_vol=0.0009,             # baseline per-bucket return noise
    inject_pattern=True,          # False -> "clean" day, for false-positive check
    seed=RNG_SEED,
):
    """
    Long-format DataFrame, one row per (stock, time_bucket):
        stock, bucket, mkt_volume, our_volume, participation, ret

    If inject_pattern=True: a subset of stocks gets an artificial
    participation + price-impact spike in the last fraction of the day,
    with the impact sign flipping mid-spike (buy-then-sell, mirroring
    the SEBI Patch I -> reversal structure). This is the ground-truth
    positive case to validate the detector against.

    TODO (real data): replace this whole function with a loader that
    pulls NSE bhavcopy (cash) + futures + options chain volumes per
    5-min bucket, and computes participation as (entity volume / total
    market volume) per stock per bucket. Everything downstream is
    agnostic to where this DataFrame comes from.
    """
    rng = np.random.default_rng(seed)
    stocks = [f"STOCK_{i:02d}" for i in range(n_stocks)]
    spike_stock_set = set(rng.choice(n_stocks, size=int(n_stocks * spike_stocks_frac), replace=False))
    spike_start = int(n_buckets * spike_start_frac)

    rows = []
    for si, stock in enumerate(stocks):
        # crude U-shaped intraday volume profile (high at open/close) --
        # this is a real, innocent pattern every detector needs to
        # survive without false-flagging.
        t = np.linspace(0, 1, n_buckets)
        mkt_vol_profile = 1.0 + 1.5 * ((t - 0.5) ** 2) * 4
        mkt_vol_profile *= rng.uniform(0.8, 1.2, size=n_buckets)
        mkt_vol = (mkt_vol_profile * rng.uniform(8e4, 2e5)).astype(int)

        participation = baseline_participation * rng.uniform(0.6, 1.4, size=n_buckets)
        ret = rng.normal(0, noise_vol, size=n_buckets)

        if inject_pattern and si in spike_stock_set:
            participation[spike_start:] = spike_participation * rng.uniform(0.7, 1.3, size=n_buckets - spike_start)
            mid = spike_start + (n_buckets - spike_start) // 2
            signed_part = participation.copy()
            signed_part[spike_start:mid] *= 1     # buying -> price up
            signed_part[mid:] *= -1               # selling -> price down
            ret[spike_start:] += (
                impact_coef * signed_part[spike_start:] * rng.uniform(0.8, 1.2, size=n_buckets - spike_start)
            )

        our_vol = (participation * mkt_vol).astype(int)
        for b in range(n_buckets):
            rows.append((stock, b, int(mkt_vol[b]), int(our_vol[b]), float(ret[b])))

    df = pd.DataFrame(rows, columns=["stock", "bucket", "mkt_volume", "our_volume", "ret"])
    df["participation"] = df["our_volume"] / df["mkt_volume"]
    return df


# ====================================================================
# SECTION 2: SPATIAL WEIGHTS -> INSTRUMENT x TIME WEIGHTS
# ====================================================================

def build_weight_matrix(df, stock_col="stock", bucket_col="bucket",
                         temporal_window=1, cross_stock_corr=None):
    """
    Original DDF: W was spatial contiguity/distance. Here, "neighbors" are:
      - same stock, adjacent time buckets (temporal contiguity)
      - different stocks, same/adjacent bucket, weighted by return
        correlation (cross-sectional contiguity)

    This matters: BANKNIFTY constituents move together, so treating them
    as independent units understates how unusual simultaneous
    concentration across 11/12 stocks really is -- same reason adjacent
    geographic units can't be treated as independent draws.

    Row-standardized (rows sum to 1) per the standard Moran's I/Geary's C
    convention.

    NOTE: this is an O(n^2) python double loop -- fine for the demo grid
    (12 stocks x 75 buckets = 900 rows -> ~5-10 sec, run once per
    dataset). Vectorize before scaling to 1-min buckets x a larger
    instrument universe.
    """
    units = df[[stock_col, bucket_col]].reset_index(drop=True)
    n = len(units)
    stock_idx = {s: i for i, s in enumerate(units[stock_col].unique())}
    bucket_arr = units[bucket_col].values
    stock_arr = units[stock_col].map(stock_idx).values

    if cross_stock_corr is None:
        # Fallback prior: BANKNIFTY constituents are genuinely correlated.
        # Swap in realized correlation from real data when available.
        n_stocks_ = len(stock_idx)
        cross_stock_corr = np.full((n_stocks_, n_stocks_), 0.45)
        np.fill_diagonal(cross_stock_corr, 1.0)

    W = np.zeros((n, n))
    for i in range(n):
        for j in range(n):
            if i == j:
                continue
            same_stock = stock_arr[i] == stock_arr[j]
            dt = abs(int(bucket_arr[i]) - int(bucket_arr[j]))
            if same_stock and 0 < dt <= temporal_window:
                W[i, j] = 1.0 / dt
            elif (not same_stock) and dt <= temporal_window:
                corr = cross_stock_corr[stock_arr[i], stock_arr[j]]
                if corr > 0:
                    W[i, j] = corr / (dt + 1)

    row_sums = W.sum(axis=1, keepdims=True)
    row_sums[row_sums == 0] = 1.0
    return W / row_sums


# ====================================================================
# SECTION 3: MORAN'S I / GEARY'S C ON THE PARTICIPATION GRID
# ====================================================================

def morans_i(x, W):
    """
    I = (n/S0) * (z'Wz) / (z'z),  z = x - mean(x), S0 = sum of all weights.
    High I = participation spikes are clustered (a few stocks + a few
    buckets carry it all) rather than spread diffusely.
    """
    x = np.asarray(x, dtype=float)
    n = len(x)
    z = x - x.mean()
    S0 = W.sum()
    den = z @ z
    if den == 0 or S0 == 0:
        return 0.0
    return (n / S0) * (z @ W @ z) / den


def gearys_c(x, W):
    """
    C = [(n-1)/(2*S0)] * sum_ij w_ij(x_i-x_j)^2 / sum_i z_i^2
    Ranges ~0 (sharp local clustering) to ~2 (negative autocorrelation),
    1 = no autocorrelation.

    This carried the real signal in the original DDF work -- more
    sensitive to sharp LOCAL discontinuities than Moran's I (which is
    better tuned to smooth global trends). The JS pattern is exactly the
    sharp-local case: most of the day is quiet, a narrow window in
    specific stocks is not.
    """
    x = np.asarray(x, dtype=float)
    n = len(x)
    z = x - x.mean()
    S0 = W.sum()
    if S0 == 0 or (z @ z) == 0:
        return 1.0
    diff_sq = (x[:, None] - x[None, :]) ** 2
    num = (W * diff_sq).sum()
    return ((n - 1) / (2 * S0)) * (num / (z @ z))


# ====================================================================
# SECTION 4: HH_frac (HHI) -- CONCENTRATION INDEX
# ====================================================================

def hhi_concentration(volumes):
    """
    HHI = sum(share_i^2), share_i = v_i / sum(v). Ranges 1/n (even
    spread) to 1 (one unit has everything).

    NOTE: in the original DDF, HH_frac was a zero-variance suppressor
    once combined with Geary's C -- didn't carry independent signal
    there. Don't assume that holds here: SEBI's own order leans on raw
    participation % as primary evidence, so HHI may carry real,
    non-redundant signal in THIS domain. Test it fresh -- see
    validate_hhi_redundancy() below, don't just inherit the old verdict.

    IMPORTANT: HHI only depends on sum(v) and sum(v^2) -- both are
    permutation-invariant. Never use this as a permutation-test
    statistic on a single reordered vector; the "null distribution"
    you'd get is floating-point summation noise, not signal. It's fine
    as a plain descriptive snapshot (see hhi_by_bucket / validate_hhi_
    redundancy below), just not as a shuffle-and-recompute test stat.
    """
    v = np.asarray(volumes, dtype=float)
    total = v.sum()
    if total == 0:
        return 0.0
    shares = v / total
    return float((shares ** 2).sum())


def hhi_by_bucket(df):
    """HHI of our volume across stocks, per time bucket. High = at that
    moment, our participation is dominated by one or two stocks rather
    than spread across the index."""
    return df.groupby("bucket")["our_volume"].apply(hhi_concentration)


def validate_hhi_redundancy(df):
    """
    Crude but honest redundancy check: correlate bucket-level HHI against
    a bucket-level "local clustering" proxy (cross-stock variance of
    participation within the bucket). High correlation (~>0.8) ->
    redundant with the autocorrelation stats, drop/downweight. Low
    correlation -> independent signal, keep it.
    """
    hhi_vals = hhi_by_bucket(df).values
    local_var = df.groupby("bucket")["participation"].var().values
    corr = np.corrcoef(hhi_vals, local_var)[0, 1]
    verdict = "REDUNDANT (drop/downweight)" if abs(corr) > 0.8 else "INDEPENDENT SIGNAL (keep)"
    print(f"HHI vs local-concentration-proxy correlation: {corr:.3f} -> {verdict}")
    return corr


# ====================================================================
# SECTION 5: PERMUTATION NULL DISTRIBUTION
# ====================================================================

def permutation_test(df, stat_fn, W, value_col="participation",
                      n_perm=N_PERMUTATIONS, block_col="stock",
                      tail="greater", seed=RNG_SEED):
    """
    BLOCK PERMUTATION, NOT A FULL SHUFFLE -- this is the important
    design choice. A naive full shuffle across the whole grid would
    destroy any legitimate structure that's specific to a stock or a
    time-of-day, making the null too easy to beat -> false positives.

    DEFAULT IS block_col="stock": for each stock independently, shuffle
    its own values across time buckets. This preserves each stock's
    overall daily level/behavior but destroys cross-sectional time
    alignment -- exactly the right null for "did several stocks spike
    at the SAME moment by chance," which is the systemic, multi-asset
    pattern this module targets.

    Why not block_col="bucket" (shuffle stock identity within a fixed
    bucket) as the default: that only reassigns which stock owns which
    value WITHIN a bucket. The set of values in the bucket is unchanged
    either way. If 11/12 stocks are already elevated in a given bucket,
    that fact survives the shuffle untouched -- the null distribution
    inherits the exact simultaneous-spike pattern you're testing
    against, and Geary's C/Moran's I can barely tell observed from null
    (this is what showed up in Case B below before the fix: observed
    Geary's C and the permutation null mean came out nearly identical).
    Use block_col="bucket" only if you specifically want the
    complementary question: "is this one stock unusually singled out,
    given the cross-sectional level at each point in time" -- a stock-
    specific outlier test, not a systemic-event test.

    CAVEAT for real data: this default assumes participation itself has
    no genuine intraday seasonality (true in the synthetic generator
    below -- the U-shape lives in raw volume, not in the participation
    ratio). On real data, check for real time-of-day structure in
    participation before trusting this null as-is; detrend first if
    it's there, or you'll inflate false positives instead.

    tail="greater" for stats where HIGH = clustered (Moran's I, HHI).
    tail="less" for stats where LOW = clustered (Geary's C).
    """
    rng = np.random.default_rng(seed)
    observed = stat_fn(df[value_col].values, W)

    df_work = df.copy().reset_index(drop=True)
    null_stats = np.empty(n_perm)
    for p in range(n_perm):
        df_perm = df_work.copy()
        df_perm[value_col] = (
            df_perm.groupby(block_col)[value_col].transform(lambda s: rng.permutation(s.values))
        )
        null_stats[p] = stat_fn(df_perm[value_col].values, W)

    if tail == "greater":
        p_value = (np.sum(null_stats >= observed) + 1) / (n_perm + 1)
    else:
        p_value = (np.sum(null_stats <= observed) + 1) / (n_perm + 1)
    return observed, null_stats, p_value


# ====================================================================
# SECTION 6: CAUCHY COMBINATION TEST (Liu & Xie, 2020, JASA)
# ====================================================================

def cauchy_combination_test(p_values, weights=None):
    """
    T = sum_i w_i * tan((0.5 - p_i) * pi),  sum(w_i) = 1
    combined p-value = 0.5 - arctan(T) / pi

    Why this over Fisher's or Bonferroni: BANKNIFTY constituents are
    correlated, so per-stock p-values below are NOT independent.
    Fisher's assumes independence and is miscalibrated here; Bonferroni
    is valid under dependence but needlessly conservative, and will bury
    a real signal spread thinly across several correlated stocks. CCT
    stays valid under arbitrary, unknown dependence (Liu & Xie 2020) --
    exactly our situation, and you don't have to estimate the
    correlation structure to use it.
    """
    p = np.clip(np.asarray(p_values, dtype=float), 1e-10, 1 - 1e-10)
    n = len(p)
    if weights is None:
        weights = np.full(n, 1.0 / n)
    else:
        weights = np.asarray(weights, dtype=float)
        weights = weights / weights.sum()

    T = np.sum(weights * np.tan((0.5 - p) * np.pi))
    combined_p = 0.5 - np.arctan(T) / np.pi
    return T, combined_p


def max_consecutive_elevated_run(x, k=1.0):
    """
    Order-sensitive local-concentration statistic, replacing HHI as the
    per-stock permutation-test stat.

    Why HHI had to go: HHI = sum((x_i/sum(x))^2) depends only on
    sum(x) and sum(x^2). Both are permutation-invariant -- reordering
    x can never change either one. So shuffling x and recomputing HHI
    cannot produce a real null distribution; whatever spread shows up
    is floating-point summation round-off (different addition order ->
    different rounding, at roughly machine-epsilon scale), not signal.
    A statistic used in a permutation test MUST depend on sequence/
    order, or the test is measuring noise.

    This one does: it's the length of the longest run of consecutive
    buckets where x exceeds (mean + k*std) of its own daily values.
    Shuffling x scatters those elevated buckets around, genuinely
    shrinking the run length on average -- giving a valid null.
    It also matches the structural signature actually being hunted for
    (a SUSTAINED elevated window), not just an elevated total.
    """
    x = np.asarray(x, dtype=float)
    if len(x) == 0:
        return 0
    thresh = x.mean() + k * x.std()
    elevated = x > thresh
    best = run = 0
    for v in elevated:
        run = run + 1 if v else 0
        best = max(best, run)
    return best


def per_stock_local_pvalues(df, n_perm=500, seed=RNG_SEED, k=1.0):
    """
    Per stock: is THIS stock's own participation-vs-time profile more
    temporally concentrated (a longer run of consecutive elevated
    buckets) than expected if its participation were randomly reshuffled
    across the day? One p-value per stock, feeding into the Cauchy
    combination above.

    `k` controls the elevation threshold (mean + k*std) used by
    max_consecutive_elevated_run -- raise it to demand a sharper spike
    before a bucket counts as "elevated."
    """
    rng = np.random.default_rng(seed)
    results = {}
    for stock, g in df.groupby("stock"):
        vals = g.sort_values("bucket")["participation"].values
        observed_stat = max_consecutive_elevated_run(vals, k=k)
        null_stat = np.array([
            max_consecutive_elevated_run(rng.permutation(vals), k=k)
            for _ in range(n_perm)
        ])
        results[stock] = (np.sum(null_stat >= observed_stat) + 1) / (n_perm + 1)
    return pd.Series(results, name="local_p_value")


# ====================================================================
# SECTION 7: TWO-LEG PnL ASYMMETRY + SCALE-INVARIANCE TEST
# ====================================================================
# Directly operationalizes the forum comment's heuristic: simulate a
# strategy trading Leg1 (illiquid) against Leg2 (liquid, ~10x notional,
# ~100x deeper market per SEBI's numbers), then ask whether risk-
# adjusted PnL gets BETTER or WORSE as size shrinks toward zero.
# Genuine arb -> better (impact drag vanishes, real edge remains).
# Impact-harvesting -> worse / collapses (no footprint, no profit).

def simulate_two_leg_strategy(
    size_multiplier=1.0,
    leg1_mkt_depth=1.0,
    leg1_base_notional=1.0,
    impact_coef=0.02,
    true_edge=0.0,     # >0 simulates a strategy with SOME genuine
                        # informational edge, to see how the curve's
                        # shape changes vs. the pure-impact case
    n_days=300,
    seed=RNG_SEED,
):
    """
    Deliberately simple toy model -- the point is the SHAPE of the
    scale curve, not calibrated PnL magnitudes.

    Leg1 impact cost is convex in participation (standard Almgren-Chriss
    intuition: cost grows faster than linearly as you trade a bigger
    fraction of available liquidity) -> impact_coef * participation^1.5.
    Leg2 monetizes the SAME price move Leg1 generated, scaled by Leg2's
    much larger notional, in a market 100x deeper so Leg2 barely moves
    its own price.
    """
    rng = np.random.default_rng(seed)
    leg1_notional = leg1_base_notional * size_multiplier
    leg2_notional = leg1_notional * 10  # SEBI ratio: ~7-10x

    participation1 = leg1_notional / leg1_mkt_depth
    participation2 = leg2_notional / (leg1_mkt_depth * 100)

    leg1_impact_cost = impact_coef * participation1 ** 1.5
    price_move_generated = impact_coef * participation1

    leg1_mean = -leg1_impact_cost + true_edge * leg1_notional
    leg1_pnl = rng.normal(leg1_mean, abs(leg1_mean) * 0.3 + 1e-6, n_days)

    leg2_mean = price_move_generated * leg2_notional * (1 - participation2)
    leg2_pnl = rng.normal(leg2_mean, abs(leg2_mean) * 0.15 + 1e-6, n_days)

    total_pnl = leg1_pnl + leg2_pnl
    sharpe = total_pnl.mean() / (total_pnl.std() + 1e-9)
    return {
        "size_multiplier": size_multiplier,
        "leg1_mean_pnl": leg1_pnl.mean(),
        "leg2_mean_pnl": leg2_pnl.mean(),
        "total_mean_pnl": total_pnl.mean(),
        "sharpe": sharpe,
    }


def scale_invariance_curve(size_grid=None, **kwargs):
    """Sharpe-vs-size curve across a grid of size multipliers."""
    if size_grid is None:
        size_grid = np.array([0.01, 0.05, 0.1, 0.25, 0.5, 1.0, 2.0, 5.0, 10.0])
    rows = [simulate_two_leg_strategy(size_multiplier=s, **kwargs) for s in size_grid]
    return pd.DataFrame(rows)


# ====================================================================
# SECTION 8: DEMO RUN
# ====================================================================

if __name__ == "__main__":
    print("=" * 70)
    print("CASE A: clean day (no injected pattern) -- false-positive check")
    print("=" * 70)
    df_clean = generate_synthetic_market_data(inject_pattern=False)
    W_clean = build_weight_matrix(df_clean)
    print(f"Moran's I  = {morans_i(df_clean['participation'].values, W_clean):.4f}  (expect ~0)")
    print(f"Geary's C  = {gearys_c(df_clean['participation'].values, W_clean):.4f}  (expect ~1)")
    _, _, p_clean = permutation_test(df_clean, gearys_c, W_clean, tail="less")
    print(f"Geary's C permutation p-value (block_col=stock) = {p_clean:.4f}  (expect NOT significant)")

    print("\n" + "=" * 70)
    print("CASE B: injected JS-style concentration pattern")
    print("=" * 70)
    df_spike = generate_synthetic_market_data(inject_pattern=True)
    W_spike = build_weight_matrix(df_spike)
    print(f"Moran's I  = {morans_i(df_spike['participation'].values, W_spike):.4f}")
    print(f"Geary's C  = {gearys_c(df_spike['participation'].values, W_spike):.4f}  (expect well below 1)")
    obs_g, null_g, p_spike = permutation_test(df_spike, gearys_c, W_spike, tail="less")
    print(f"Geary's C permutation p-value (block_col=stock) = {p_spike:.4f}  (observed {obs_g:.4f} vs null mean {null_g.mean():.4f})")

    print("\n--- Per-stock local concentration test (run-length stat) + Cauchy combination ---")
    local_p = per_stock_local_pvalues(df_spike)
    print(local_p.sort_values())
    T_cct, combined_p = cauchy_combination_test(local_p.values)
    print(f"\nCauchy combination statistic = {T_cct:.4f}")
    print(f"Combined global p-value       = {combined_p:.6f}")

    print("\n--- HHI redundancy check ---")
    validate_hhi_redundancy(df_spike)

    print("\n" + "=" * 70)
    print("Scale-invariance test (the forum '/100' heuristic, formalized)")
    print("=" * 70)
    print("\n[A] Pure impact-harvesting (true_edge=0):")
    curve_impact = scale_invariance_curve(true_edge=0.0)
    print(curve_impact.to_string(index=False, float_format=lambda x: f"{x:,.4f}"))
    print("\n[B] Some genuine edge present (true_edge=0.005):")
    curve_edge = scale_invariance_curve(true_edge=0.005)
    print(curve_edge.to_string(index=False, float_format=lambda x: f"{x:,.4f}"))

    print("\nReading it: in [A], Sharpe should COLLAPSE as size shrinks toward 0 --")
    print("no footprint, no profit. In [B], Sharpe should hold up or improve at small")
    print("size, because the edge term doesn't depend on your own impact. That contrast")
    print("IS the diagnostic the forum comment described, made falsifiable.")
```

```python
"""
Real-data false-positive calibration run for the DDF detector.

WHAT THIS DOES
----------------
Pulls real 5-min intraday volume bars for Bank Nifty constituent stocks
from Yahoo Finance, for every real trading day in the last ~60 days
(Yahoo's free intraday history limit -- a platform limit, not a code
limit).

For each real day, builds a public-data participation proxy:

    share[stock, bucket] = that stock's volume in the bucket
                            / total basket volume in the bucket

This is NOT SEBI's actual evidence (one named entity's % of a stock's
OWN volume) -- that needs confidential trade-level data nobody public
has. What this DOES give you: a real check of whether the fixed
detector throws false alarms on ordinary real trading days. That's a
necessary step before arXiv, not the full case replication.

VERIFY BEFORE RUNNING: ticker list below against
https://www.niftyindices.com/Factsheet/ind_nifty_bank.pdf
-- Bank Nifty composition changes periodically.
"""

import numpy as np
import pandas as pd
import yfinance as yf
import warnings
warnings.filterwarnings("ignore")

RNG_SEED = 42
N_PERMUTATIONS = 500
LOCAL_N_PERM = 300
TEMPORAL_WINDOW = 1

# CHECK THIS LIST against NSE's factsheet before running -- composition
# changes. Best estimate as of mid-2026 search results, may be stale.
BANKNIFTY_TICKERS = [
    "HDFCBANK.NS", "SBIN.NS", "ICICIBANK.NS", "AXISBANK.NS",
    "KOTAKBANK.NS", "BANKBARODA.NS", "UNIONBANK.NS", "PNB.NS",
    "CANBK.NS", "FEDERALBNK.NS", "AUBANK.NS", "YESBANK.NS",
    "INDUSINDBK.NS",
]


def fetch_intraday_volumes(tickers, period="60d", interval="5m"):
    """
    yfinance caps 5m bars at ~60 days back. For real SEBI-case dates
    you need a paid intraday vendor or a broker API (e.g. Zerodha Kite
    Connect) with historical entitlements -- free public sources don't
    go back that far at this granularity.
    """
    frames = {}
    for tk in tickers:
        df = yf.download(tk, period=period, interval=interval, progress=False)
        if df.empty:
            print(f"WARNING: no data for {tk}, skipping")
            continue
        vol = df["Volume"]
        if isinstance(vol, pd.DataFrame):   # yfinance sometimes returns multiindex
            vol = vol.iloc[:, 0]
        if vol.index.tz is None:
            vol = vol.tz_localize("UTC")
        vol = vol.tz_convert("Asia/Kolkata")
        frames[tk] = vol
    return frames


def build_daily_grid(frames, session_start="09:15", session_end="15:30"):
    """
    Reshape {ticker: Volume series} -> one long-format grid per real
    trading day: columns [stock, bucket, mkt_volume, participation].
    participation here = public basket-share proxy (see module docstring).
    """
    vol_df = pd.DataFrame(frames)
    vol_df = vol_df.between_time(session_start, session_end)
    vol_df = vol_df.dropna(how="all")

    daily = {}
    for date, day_slice in vol_df.groupby(vol_df.index.date):
        day_slice = day_slice.sort_index().fillna(0.0)
        if len(day_slice) < 50:     # holiday-truncated/partial session, skip
            continue
        rows = []
        for b, (ts, row) in enumerate(day_slice.iterrows()):
            for stock in day_slice.columns:
                rows.append((stock, b, float(row[stock])))
        long_df = pd.DataFrame(rows, columns=["stock", "bucket", "mkt_volume"])
        bucket_totals = long_df.groupby("bucket")["mkt_volume"].transform("sum")
        long_df["participation"] = (long_df["mkt_volume"] / bucket_totals.replace(0, np.nan)).fillna(0.0)
        daily[str(date)] = long_df
    return daily


# ---- fixed DDF machinery, unchanged from the corrected script ----

def build_weight_matrix(df, stock_col="stock", bucket_col="bucket",
                         temporal_window=1, cross_stock_corr=None):
    units = df[[stock_col, bucket_col]].reset_index(drop=True)
    n = len(units)
    stock_idx = {s: i for i, s in enumerate(units[stock_col].unique())}
    bucket_arr = units[bucket_col].values
    stock_arr = units[stock_col].map(stock_idx).values

    if cross_stock_corr is None:
        n_stocks_ = len(stock_idx)
        cross_stock_corr = np.full((n_stocks_, n_stocks_), 0.45)
        np.fill_diagonal(cross_stock_corr, 1.0)

    W = np.zeros((n, n))
    for i in range(n):
        for j in range(n):
            if i == j:
                continue
            same_stock = stock_arr[i] == stock_arr[j]
            dt = abs(int(bucket_arr[i]) - int(bucket_arr[j]))
            if same_stock and 0 < dt <= temporal_window:
                W[i, j] = 1.0 / dt
            elif (not same_stock) and dt <= temporal_window:
                corr = cross_stock_corr[stock_arr[i], stock_arr[j]]
                if corr > 0:
                    W[i, j] = corr / (dt + 1)

    row_sums = W.sum(axis=1, keepdims=True)
    row_sums[row_sums == 0] = 1.0
    return W / row_sums


def gearys_c(x, W):
    x = np.asarray(x, dtype=float)
    n = len(x)
    z = x - x.mean()
    S0 = W.sum()
    if S0 == 0 or (z @ z) == 0:
        return 1.0
    diff_sq = (x[:, None] - x[None, :]) ** 2
    num = (W * diff_sq).sum()
    return ((n - 1) / (2 * S0)) * (num / (z @ z))


def permutation_test(df, stat_fn, W, value_col="participation",
                      n_perm=N_PERMUTATIONS, block_col="stock",
                      tail="less", seed=RNG_SEED):
    rng = np.random.default_rng(seed)
    observed = stat_fn(df[value_col].values, W)
    df_work = df.copy().reset_index(drop=True)
    null_stats = np.empty(n_perm)
    for p in range(n_perm):
        df_perm = df_work.copy()
        df_perm[value_col] = (
            df_perm.groupby(block_col)[value_col].transform(lambda s: rng.permutation(s.values))
        )
        null_stats[p] = stat_fn(df_perm[value_col].values, W)
    if tail == "greater":
        p_value = (np.sum(null_stats >= observed) + 1) / (n_perm + 1)
    else:
        p_value = (np.sum(null_stats <= observed) + 1) / (n_perm + 1)
    return observed, null_stats, p_value


def max_consecutive_elevated_run(x, k=1.0):
    x = np.asarray(x, dtype=float)
    if len(x) == 0:
        return 0
    thresh = x.mean() + k * x.std()
    elevated = x > thresh
    best = run = 0
    for v in elevated:
        run = run + 1 if v else 0
        best = max(best, run)
    return best


def per_stock_local_pvalues(df, n_perm=LOCAL_N_PERM, seed=RNG_SEED, k=1.0):
    rng = np.random.default_rng(seed)
    results = {}
    for stock, g in df.groupby("stock"):
        vals = g.sort_values("bucket")["participation"].values
        observed_stat = max_consecutive_elevated_run(vals, k=k)
        null_stat = np.array([
            max_consecutive_elevated_run(rng.permutation(vals), k=k)
            for _ in range(n_perm)
        ])
        results[stock] = (np.sum(null_stat >= observed_stat) + 1) / (n_perm + 1)
    return pd.Series(results, name="local_p_value")


def cauchy_combination_test(p_values, weights=None):
    p = np.clip(np.asarray(p_values, dtype=float), 1e-10, 1 - 1e-10)
    n = len(p)
    if weights is None:
        weights = np.full(n, 1.0 / n)
    else:
        weights = np.asarray(weights, dtype=float)
        weights = weights / weights.sum()
    T = np.sum(weights * np.tan((0.5 - p) * np.pi))
    combined_p = 0.5 - np.arctan(T) / np.pi
    return T, combined_p


# ====================================================================
# MAIN: run the pipeline once per real trading day, collect a real
# false-positive rate
# ====================================================================

if __name__ == "__main__":
    print("Downloading real intraday volume data (last 60 days, 5-min bars)...")
    frames = fetch_intraday_volumes(BANKNIFTY_TICKERS)
    print(f"Got data for {len(frames)}/{len(BANKNIFTY_TICKERS)} tickers")

    daily_grids = build_daily_grid(frames)
    print(f"Usable real trading days: {len(daily_grids)}")
    print("NOTE: this loop is slow (O(n^2) weight matrix per day x permutations).")
    print("Expect this to take several minutes. Reduce N_PERMUTATIONS / period if needed.\n")

    results = []
    for date, df_day in daily_grids.items():
        W = build_weight_matrix(df_day, temporal_window=TEMPORAL_WINDOW)
        gc_obs, gc_null, gc_p = permutation_test(df_day, gearys_c, W, tail="less")
        local_p = per_stock_local_pvalues(df_day)
        _, cauchy_p = cauchy_combination_test(local_p.values)
        results.append({
            "date": date,
            "gearys_c": gc_obs,
            "gearys_c_p": gc_p,
            "cauchy_combined_p": cauchy_p,
            "n_significant_stocks_5pct": int((local_p < 0.05).sum()),
        })
        print(f"{date}: Geary's C={gc_obs:.4f} p={gc_p:.4f} | Cauchy p={cauchy_p:.4f}")

    results_df = pd.DataFrame(results)
    print("\n" + "=" * 70)
    print("FALSE-POSITIVE CALIBRATION SUMMARY (real days, public-data proxy)")
    print("=" * 70)
    fp_rate_gc = (results_df["gearys_c_p"] < 0.05).mean()
    fp_rate_cauchy = (results_df["cauchy_combined_p"] < 0.05).mean()
    print(f"Geary's C:    flagged {fp_rate_gc:.1%} of real days at p<0.05 (want ~5%)")
    print(f"Cauchy combo: flagged {fp_rate_cauchy:.1%} of real days at p<0.05 (want ~5%)")
    print("\nIf either rate is way above 5%, the detector is too trigger-happy on")
    print("real market microstructure and needs recalibration before it's")
    print("trustworthy on anything beyond synthetic data.")
```

```python
"""
Real-data calibration run, v2: cross-day empirical null instead of
within-day permutation null.

WHY THIS CHANGE
----------------
v1 results: Geary's C flagged 78.9% of real days, Cauchy combo flagged
21.1%. Both way above the ~5% you'd want for "just noise."

Root cause: a within-day permutation null assumes "clean" data is
random noise. Real market data isn't -- each stock's volume share
moves smoothly through the day, and stocks share real, benign
common-factor co-movement (everyone's volume tends to rise at
open/close together -- this is "the totally innocent U-shaped intraday
volume profile" Section 5's docstring already warned about, it just
didn't show up in the synthetic generator because that generator's
noise WAS i.i.d.). Shuffle real smooth structure and almost any real
day looks "clustered" next to its own scrambled self. That's not
manipulation, that's just what markets look like every day.

FIX: stop comparing a day to a randomized version of itself. Compare
it to OTHER real days instead (leave-one-out empirical baseline).
Flag a day only if it's unusual relative to the rest of the sample.

Bonus: this is much faster too -- no permutation loop per day, just
one pass to collect each day's statistic, then a cross-day ranking.
"""

import numpy as np
import pandas as pd
import yfinance as yf
import warnings
warnings.filterwarnings("ignore")

TEMPORAL_WINDOW = 1
RUN_LENGTH_K = 1.0   # threshold multiplier for max_consecutive_elevated_run

BANKNIFTY_TICKERS = [
    "HDFCBANK.NS", "SBIN.NS", "ICICIBANK.NS", "AXISBANK.NS",
    "KOTAKBANK.NS", "BANKBARODA.NS", "UNIONBANK.NS", "PNB.NS",
    "CANBK.NS", "FEDERALBNK.NS", "AUBANK.NS", "YESBANK.NS",
    "INDUSINDBK.NS",
]


def fetch_intraday_volumes(tickers, period="60d", interval="5m"):
    frames = {}
    for tk in tickers:
        df = yf.download(tk, period=period, interval=interval, progress=False)
        if df.empty:
            print(f"WARNING: no data for {tk}, skipping")
            continue
        vol = df["Volume"]
        if isinstance(vol, pd.DataFrame):
            vol = vol.iloc[:, 0]
        if vol.index.tz is None:
            vol = vol.tz_localize("UTC")
        vol = vol.tz_convert("Asia/Kolkata")
        frames[tk] = vol
    return frames


def build_daily_grid(frames, session_start="09:15", session_end="15:30"):
    vol_df = pd.DataFrame(frames)
    vol_df = vol_df.between_time(session_start, session_end)
    vol_df = vol_df.dropna(how="all")
    daily = {}
    for date, day_slice in vol_df.groupby(vol_df.index.date):
        day_slice = day_slice.sort_index().fillna(0.0)
        if len(day_slice) < 50:
            continue
        rows = []
        for b, (ts, row) in enumerate(day_slice.iterrows()):
            for stock in day_slice.columns:
                rows.append((stock, b, float(row[stock])))
        long_df = pd.DataFrame(rows, columns=["stock", "bucket", "mkt_volume"])
        bucket_totals = long_df.groupby("bucket")["mkt_volume"].transform("sum")
        long_df["participation"] = (long_df["mkt_volume"] / bucket_totals.replace(0, np.nan)).fillna(0.0)
        daily[str(date)] = long_df
    return daily


def build_weight_matrix(df, stock_col="stock", bucket_col="bucket",
                         temporal_window=1, cross_stock_corr=None):
    units = df[[stock_col, bucket_col]].reset_index(drop=True)
    n = len(units)
    stock_idx = {s: i for i, s in enumerate(units[stock_col].unique())}
    bucket_arr = units[bucket_col].values
    stock_arr = units[stock_col].map(stock_idx).values
    if cross_stock_corr is None:
        n_stocks_ = len(stock_idx)
        cross_stock_corr = np.full((n_stocks_, n_stocks_), 0.45)
        np.fill_diagonal(cross_stock_corr, 1.0)
    W = np.zeros((n, n))
    for i in range(n):
        for j in range(n):
            if i == j:
                continue
            same_stock = stock_arr[i] == stock_arr[j]
            dt = abs(int(bucket_arr[i]) - int(bucket_arr[j]))
            if same_stock and 0 < dt <= temporal_window:
                W[i, j] = 1.0 / dt
            elif (not same_stock) and dt <= temporal_window:
                corr = cross_stock_corr[stock_arr[i], stock_arr[j]]
                if corr > 0:
                    W[i, j] = corr / (dt + 1)
    row_sums = W.sum(axis=1, keepdims=True)
    row_sums[row_sums == 0] = 1.0
    return W / row_sums


def gearys_c(x, W):
    x = np.asarray(x, dtype=float)
    n = len(x)
    z = x - x.mean()
    S0 = W.sum()
    if S0 == 0 or (z @ z) == 0:
        return 1.0
    diff_sq = (x[:, None] - x[None, :]) ** 2
    num = (W * diff_sq).sum()
    return ((n - 1) / (2 * S0)) * (num / (z @ z))


def max_consecutive_elevated_run(x, k=1.0):
    x = np.asarray(x, dtype=float)
    if len(x) == 0:
        return 0
    thresh = x.mean() + k * x.std()
    elevated = x > thresh
    best = run = 0
    for v in elevated:
        run = run + 1 if v else 0
        best = max(best, run)
    return best


def cauchy_combination_test(p_values, weights=None):
    p = np.clip(np.asarray(p_values, dtype=float), 1e-10, 1 - 1e-10)
    n = len(p)
    if weights is None:
        weights = np.full(n, 1.0 / n)
    else:
        weights = np.asarray(weights, dtype=float)
        weights = weights / weights.sum()
    T = np.sum(weights * np.tan((0.5 - p) * np.pi))
    combined_p = 0.5 - np.arctan(T) / np.pi
    return T, combined_p


def leave_one_out_pvalue(values, idx, tail="less"):
    """
    values: array of a statistic across all days.
    idx: index of the day under test.
    tail="less": flag if today is unusually LOW vs the rest (Geary's C).
    tail="greater": flag if today is unusually HIGH vs the rest (run-length).
    """
    others = np.delete(values, idx)
    n = len(others)
    if tail == "less":
        return (np.sum(others <= values[idx]) + 1) / (n + 1)
    else:
        return (np.sum(others >= values[idx]) + 1) / (n + 1)


if __name__ == "__main__":
    print("Downloading real intraday volume data (last 60 days, 5-min bars)...")
    frames = fetch_intraday_volumes(BANKNIFTY_TICKERS)
    print(f"Got data for {len(frames)}/{len(BANKNIFTY_TICKERS)} tickers")

    daily_grids = build_daily_grid(frames)
    dates = sorted(daily_grids.keys())
    print(f"Usable real trading days: {len(dates)}\n")

    if len(dates) < 15:
        print("WARNING: fewer than 15 days. Cross-day baseline will be unstable.")

    # ---- pass 1: collect raw statistics per day, no permutation ----
    gc_values = []
    run_length_by_stock = {}
    W_cache = {}

    for date in dates:
        df_day = daily_grids[date]
        n_rows = len(df_day)
        if n_rows not in W_cache:
            W_cache[n_rows] = build_weight_matrix(df_day, temporal_window=TEMPORAL_WINDOW)
        W = W_cache[n_rows]

        gc_values.append(gearys_c(df_day["participation"].values, W))

        for stock, g in df_day.groupby("stock"):
            vals = g.sort_values("bucket")["participation"].values
            run_len = max_consecutive_elevated_run(vals, k=RUN_LENGTH_K)
            run_length_by_stock.setdefault(stock, []).append(run_len)

    gc_values = np.array(gc_values)
    stocks = sorted(run_length_by_stock.keys())
    run_matrix = np.array([run_length_by_stock[s] for s in stocks])  # (n_stocks, n_days)

    # ---- pass 2: leave-one-out cross-day p-values ----
    results = []
    for i, date in enumerate(dates):
        gc_p = leave_one_out_pvalue(gc_values, i, tail="less")
        local_p = np.array([
            leave_one_out_pvalue(run_matrix[s_idx], i, tail="greater")
            for s_idx in range(len(stocks))
        ])
        _, cauchy_p = cauchy_combination_test(local_p)
        results.append({
            "date": date,
            "gearys_c": gc_values[i],
            "gearys_c_p_crossday": gc_p,
            "cauchy_combined_p_crossday": cauchy_p,
            "n_significant_stocks_5pct": int((local_p < 0.05).sum()),
        })
        print(f"{date}: Geary's C={gc_values[i]:.4f} p={gc_p:.4f} | Cauchy p={cauchy_p:.4f}")

    results_df = pd.DataFrame(results)
    print("\n" + "=" * 70)
    print("FALSE-POSITIVE CALIBRATION SUMMARY -- cross-day empirical null")
    print("=" * 70)
    fp_rate_gc = (results_df["gearys_c_p_crossday"] < 0.05).mean()
    fp_rate_cauchy = (results_df["cauchy_combined_p_crossday"] < 0.05).mean()
    print(f"Geary's C:    flagged {fp_rate_gc:.1%} of real days at p<0.05 (want ~5%)")
    print(f"Cauchy combo: flagged {fp_rate_cauchy:.1%} of real days at p<0.05 (want ~5%)")
    print("\nCAVEAT: leave-one-out on ~57 days is a small, noisy baseline -- a sanity")
    print("check, not a final calibration. A real paper needs months of history,")
    print("not weeks, before this number means anything solid.")
```

```python
import yfinance as yf
import pandas as pd
import numpy as np

# ==========================================
# 1. YFINANCE DATA ACQUISITION
# ==========================================

def fetch_intraday_market_data(tickers, period="5d", interval="5m"):
    """
    Fetches real intraday volume data using yfinance.
    Calculates stock participation relative to the total basket volume.
    """
    print(f"Fetching {interval} data for {len(tickers)} tickers over {period}...")

    # Download data; suppress yfinance progress bar for clean logs
    raw_data = yf.download(tickers, period=period, interval=interval, progress=False)

    if 'Volume' not in raw_data:
        raise ValueError("Failed to fetch volume data. Check your tickers or connection.")

    df_vol = raw_data['Volume'].dropna()

    # Restructure data into a row-based format
    df_melt = df_vol.reset_index().melt(
        id_vars=['Datetime'],
        var_name='stock',
        value_name='our_volume'
    )

    # Extract Date and Time for grouping
    df_melt['date'] = df_melt['Datetime'].dt.date
    df_melt['bucket'] = df_melt['Datetime'].dt.time

    # Calculate Total Market Volume (sum of our basket) per time bucket
    df_melt['mkt_volume'] = df_melt.groupby('Datetime')['our_volume'].transform('sum')

    # Calculate Participation rate (avoiding division by zero)
    df_melt['participation'] = np.where(
        df_melt['mkt_volume'] > 0,
        df_melt['our_volume'] / df_melt['mkt_volume'],
        0.0
    )

    return df_melt

def create_equal_weight_matrix(n_stocks):
    """
    Creates a simple spatial weight matrix for the basket.
    """
    W = np.ones((n_stocks, n_stocks)) / (n_stocks - 1)
    np.fill_diagonal(W, 0.0)
    return W

# ==========================================
# 2. LOCAL METRIC: MAX ROLLING SUM
# ==========================================

def max_rolling_sum(volumes, window=5):
    """
    Sequence-dependent local metric. Shuffling perfectly destroys this.
    """
    v = np.asarray(volumes, dtype=float)
    if len(v) < window:
        return v.sum()
    return np.convolve(v, np.ones(window), mode='valid').max()

def per_stock_local_pvalues(df_day, n_perm=500, seed=42):
    rng = np.random.default_rng(seed)
    results = {}

    for stock, g in df_day.groupby('stock'):
        vals = g.sort_values('bucket')['participation'].values

        # Observed Metric
        observed_val = max_rolling_sum(vals, window=5)

        # Null Distribution: Shuffle the sequence
        null_vals = np.array([
            max_rolling_sum(rng.permutation(vals), window=5)
            for _ in range(n_perm)
        ])

        # Calculate P-value
        p_val = (np.sum(null_vals >= observed_val) + 1) / (n_perm + 1)
        results[stock] = p_val

    return pd.Series(results, name="local_p_value")

# ==========================================
# 3. GLOBAL METRIC: CROSS-SECTIONAL GEARY'S C
# ==========================================

def compute_gearys_c(x, W):
    N = len(x)
    var_x = np.sum((x - x.mean())**2)
    if var_x == 0:
        return 1.0

    diff_sq = (x[:, None] - x[None, :])**2
    numerator = (N - 1) * np.sum(W * diff_sq)
    denominator = 2 * W.sum() * var_x
    return numerator / denominator

def global_gearys_c_test(df_day, W, n_perm=500, seed=42):
    rng = np.random.default_rng(seed)

    def calc_mean_geary(data):
        c_vals = []
        for _, g in data.groupby('bucket'):
            x = g.sort_values('stock')['participation'].values
            c_vals.append(compute_gearys_c(x, W))
        return np.mean(c_vals)

    observed_c = calc_mean_geary(df_day)

    # The Fix: Shuffle time buckets for each stock *independently*
    null_c = []
    for _ in range(n_perm):
        shuffled_df = df_day.copy()
        shuffled_df['bucket'] = shuffled_df.groupby('stock')['bucket'].transform(
            lambda x: rng.permutation(x.values)
        )
        null_c.append(calc_mean_geary(shuffled_df))

    null_c = np.array(null_c)
    p_val = (np.sum(null_c <= observed_c) + 1) / (n_perm + 1)

    return observed_c, np.mean(null_c), p_val

# ==========================================
# 4. FINAL CAUCHY COMBINATION
# ==========================================

def cauchy_combination(p_values):
    p_values = np.asarray(p_values)
    p_values = np.clip(p_values, 1e-15, 1 - 1e-15)
    w = np.ones(len(p_values)) / len(p_values)
    t_stat = np.sum(w * np.tan(np.pi * (0.5 - p_values)))
    combined_p = 0.5 - (np.arctan(t_stat) / np.pi)
    return float(combined_p)

# ==========================================
# 5. EXECUTION PIPELINE
# ==========================================

if __name__ == "__main__":
    # Define a basket of highly correlated stocks (e.g., Tech/Semiconductors)
    tickers = ['AAPL', 'MSFT', 'NVDA', 'GOOG', 'META', 'AMD', 'TSLA', 'AMZN', 'INTC', 'NFLX']

    # 1. Fetch real market data
    full_data = fetch_intraday_market_data(tickers, period="5d", interval="5m")

    # 2. Setup standard weight matrix based on successful downloads
    unique_stocks = full_data['stock'].unique()
    unique_stocks.sort()
    W_matrix = create_equal_weight_matrix(len(unique_stocks))

    print("\nStarting daily scan on real market data...")

    # 3. Iterate through each trading day
    for date, df_day in full_data.groupby('date'):
        print(f"\n{'='*50}\n=== TRADING DAY: {date} ===\n{'='*50}")

        # Ensure we have enough data points for the day to run a valid test
        if len(df_day['bucket'].unique()) < 10:
            print(f"Skipping {date} - Insufficient intraday data (Early close or missing data).")
            continue

        # A. Local P-Values
        local_pvals = per_stock_local_pvalues(df_day)

        # B. Global P-Values
        obs_c, null_c_mean, global_p = global_gearys_c_test(df_day, W_matrix)

        # C. Cauchy Combination
        combined_local_p = cauchy_combination(local_pvals.values)

        # Print Results
        print(f"Global Geary's C: {obs_c:.4f} (Baseline: {null_c_mean:.4f}) | P-Value: {global_p:.4f}")
        print(f"Cauchy Combined Local P-Value: {combined_local_p:.4f}")

        # Final Verdict Matrix
        if global_p < 0.05 and combined_local_p < 0.05:
            print("🚨 VERDICT: ANOMALOUS CONCENTRATION DETECTED")
        else:
            print("✅ VERDICT: NORMAL TRADING BEHAVIOR")
```

```python
import yfinance as yf
import pandas as pd
import numpy as np
import warnings

warnings.filterwarnings("ignore", category=FutureWarning)

# ==========================================
# 1. YFINANCE DATA ACQUISITION
# ==========================================

def fetch_intraday_market_data(tickers, period="60d", interval="5m"):
    print(f"Downloading real intraday volume data (last {period}, {interval} bars)...")
    raw_data = yf.download(tickers, period=period, interval=interval, progress=False)

    if 'Volume' not in raw_data:
        raise ValueError("Failed to fetch volume data.")

    df_vol = raw_data['Volume'].dropna()
    valid_tickers = sorted(df_vol.columns.tolist())
    print(f"Got data for {len(valid_tickers)}/{len(tickers)} tickers")

    df_melt = df_vol.reset_index().melt(
        id_vars=['Datetime'], var_name='stock', value_name='our_volume'
    )

    df_melt['date'] = df_melt['Datetime'].dt.date
    df_melt['bucket'] = df_melt['Datetime'].dt.time

    df_melt['mkt_volume'] = df_melt.groupby('Datetime')['our_volume'].transform('sum')
    df_melt['participation'] = np.where(
        df_melt['mkt_volume'] > 0,
        df_melt['our_volume'] / df_melt['mkt_volume'],
        0.0
    )

    return df_melt, valid_tickers

# ==========================================
# 2. RAW METRIC CALCULATORS
# ==========================================

def max_rolling_sum(volumes, window=5):
    v = np.asarray(volumes, dtype=float)
    if len(v) < window:
        return v.sum()
    return np.convolve(v, np.ones(window), mode='valid').max()

def compute_gearys_c(x, W):
    N = len(x)
    var_x = np.sum((x - x.mean())**2)
    if var_x == 0:
        return 1.0
    diff_sq = (x[:, None] - x[None, :])**2
    numerator = (N - 1) * np.sum(W * diff_sq)
    denominator = 2 * W.sum() * var_x
    return numerator / denominator

def create_sector_weight_matrix(tickers):
    """
    THE FIX: Assigns 0.8 weight to stocks in the same sub-sector,
    and 0.2 to stocks across sectors. This gives Geary's C the
    mathematical spatial structure it needs to function.
    """
    semis = {'NVDA', 'AMD', 'INTC', 'QCOM', 'TXN'}
    n = len(tickers)
    W = np.ones((n, n)) * 0.2

    for i in range(n):
        for j in range(n):
            if i == j:
                W[i, j] = 0.0
            else:
                # If both are semis, or both are NOT semis (MegaTech)
                if (tickers[i] in semis) == (tickers[j] in semis):
                    W[i, j] = 0.8
    return W

def cauchy_combination(p_values):
    p_values = np.asarray(p_values)
    p_values = np.clip(p_values, 1e-15, 1 - 1e-15)
    w = np.ones(len(p_values)) / len(p_values)
    t_stat = np.sum(w * np.tan(np.pi * (0.5 - p_values)))
    combined_p = 0.5 - (np.arctan(t_stat) / np.pi)
    return float(combined_p)

# ==========================================
# 3. EXECUTION PIPELINE
# ==========================================

if __name__ == "__main__":
    tickers = ['AAPL', 'MSFT', 'NVDA', 'GOOG', 'META', 'AMD', 'TSLA',
               'AMZN', 'INTC', 'NFLX', 'CSCO', 'QCOM', 'TXN']

    full_data, valid_tickers = fetch_intraday_market_data(tickers, period="60d", interval="5m")

    daily_bucket_counts = full_data.groupby('date')['bucket'].nunique()
    valid_dates = daily_bucket_counts[daily_bucket_counts >= 70].index
    full_data = full_data[full_data['date'].isin(valid_dates)]

    dates = sorted(full_data['date'].unique())
    print(f"Usable real trading days: {len(dates)}\n")

    # Generate the corrected spatial matrix
    W_matrix = create_sector_weight_matrix(valid_tickers)

    raw_gearys = {}
    raw_locals = {stock: {} for stock in valid_tickers}

    for date in dates:
        df_day = full_data[full_data['date'] == date]

        c_vals = []
        for _, g in df_day.groupby('bucket'):
            # Data is sorted alphabetically to perfectly align with W_matrix
            x = g.sort_values('stock')['participation'].values
            c_vals.append(compute_gearys_c(x, W_matrix))
        raw_gearys[date] = np.mean(c_vals)

        for stock, g in df_day.groupby('stock'):
            vals = g.sort_values('bucket')['participation'].values
            raw_locals[stock][date] = max_rolling_sum(vals, window=5)

    geary_p_values = {}
    cauchy_p_values = {}

    geary_array = np.array(list(raw_gearys.values()))

    for date in dates:
        obs_geary = raw_gearys[date]
        # Empirical P-value: How often does the market naturally cluster tighter than this?
        g_pval = np.sum(geary_array <= obs_geary) / len(dates)
        geary_p_values[date] = g_pval

        local_pvals_today = []
        for stock in raw_locals.keys():
            obs_local = raw_locals[stock][date]
            stock_history = np.array(list(raw_locals[stock].values()))
            l_pval = np.sum(stock_history >= obs_local) / len(dates)
            local_pvals_today.append(l_pval)

        cauchy_p_values[date] = cauchy_combination(local_pvals_today)

        print(f"{date}: Geary's C={obs_geary:.4f} p={g_pval:.4f} | Cauchy p={cauchy_p_values[date]:.4f}")

    geary_flagged = sum(1 for p in geary_p_values.values() if p < 0.05)
    cauchy_flagged = sum(1 for p in cauchy_p_values.values() if p < 0.05)

    pct_geary = (geary_flagged / len(dates)) * 100
    pct_cauchy = (cauchy_flagged / len(dates)) * 100

    print("\n" + "="*70)
    print("FALSE-POSITIVE CALIBRATION SUMMARY -- cross-day empirical null")
    print("="*70)
    print(f"Geary's C:    flagged {pct_geary:.1f}% of real days at p<0.05 (want ~5%)")
    print(f"Cauchy combo: flagged {pct_cauchy:.1f}% of real days at p<0.05 (want ~5%)")
    print("\nCAVEAT: leave-one-out on ~57 days is a small, noisy baseline -- a sanity")
    print("check, not a final calibration. A real paper needs months of history,")
    print("not weeks, before this number means anything solid.")
    # ==========================================
# 4. DETECTION POWER TEST -- inject a synthetic
#    burst into a REAL day and see if it's caught
# ==========================================

def inject_synthetic_burst(df_day, target_stock, burst_buckets=12, burst_multiplier=8.0):
    """
    Takes one real day's long-format data and inflates target_stock's
    volume in the LAST `burst_buckets` buckets by burst_multiplier, then
    recomputes participation for the whole day.

    NOTE: because participation is compositional (sums to 1 per bucket
    by construction), inflating one stock's volume mechanically deflates
    everyone else's SHARE even though their raw volume is untouched.
    That's not an artifact to fix -- a real volume spike genuinely does
    cannibalize the rest of the bucket's share, so this is the correct
    behavior, just worth knowing when reading results.
    """
    df_mod = df_day.copy()
    last_buckets = sorted(df_mod['bucket'].unique())[-burst_buckets:]
    mask = (df_mod['stock'] == target_stock) & (df_mod['bucket'].isin(last_buckets))
    df_mod.loc[mask, 'our_volume'] *= burst_multiplier

    df_mod['mkt_volume'] = df_mod.groupby('bucket')['our_volume'].transform('sum')
    df_mod['participation'] = np.where(
        df_mod['mkt_volume'] > 0, df_mod['our_volume'] / df_mod['mkt_volume'], 0.0
    )
    return df_mod


def run_power_test(full_data, valid_tickers, W_matrix, raw_gearys, raw_locals,
                    target_stock='NVDA', burst_buckets=12, burst_multiplier=8.0,
                    n_trial_days=15, seed=42):
    """
    Picks n_trial_days real days at random, injects a burst into
    target_stock on each, and checks how often the ALREADY-CALIBRATED
    detector (using the untouched 60-day distribution as the reference
    null -- raw_gearys / raw_locals are NOT recomputed here) flags the
    doctored day at p<0.05.

    This is the missing complement to the false-positive run: that told
    you the noise floor, this tells you whether a signal of a given size
    actually clears it.
    """
    rng = np.random.default_rng(seed)
    dates = sorted(full_data['date'].unique())
    trial_dates = rng.choice(dates, size=min(n_trial_days, len(dates)), replace=False)

    geary_array = np.array(list(raw_gearys.values()))
    n_dates = len(dates)
    flagged_geary, flagged_cauchy = 0, 0

    for date in trial_dates:
        df_day = full_data[full_data['date'] == date]
        df_burst = inject_synthetic_burst(df_day, target_stock, burst_buckets, burst_multiplier)

        c_vals = [
            compute_gearys_c(g.sort_values('stock')['participation'].values, W_matrix)
            for _, g in df_burst.groupby('bucket')
        ]
        obs_geary_burst = np.mean(c_vals)
        g_pval = np.sum(geary_array <= obs_geary_burst) / n_dates

        local_pvals = []
        for stock in raw_locals.keys():
            vals = df_burst[df_burst['stock'] == stock].sort_values('bucket')['participation'].values
            obs_local = max_rolling_sum(vals, window=5)
            stock_history = np.array(list(raw_locals[stock].values()))
            local_pvals.append(np.sum(stock_history >= obs_local) / n_dates)
        cauchy_p = cauchy_combination(local_pvals)

        flagged_geary += g_pval < 0.05
        flagged_cauchy += cauchy_p < 0.05
        print(f"{date} [{target_stock} x{burst_multiplier:.0f}]: "
              f"Geary p={g_pval:.4f}{'  <-- FLAGGED' if g_pval < 0.05 else ''} | "
              f"Cauchy p={cauchy_p:.4f}{'  <-- FLAGGED' if cauchy_p < 0.05 else ''}")

    n = len(trial_dates)
    print(f"\nDetection power @ {burst_multiplier:.0f}x burst, {burst_buckets} buckets, target={target_stock}:")
    print(f"  Geary's C caught {flagged_geary}/{n} ({100*flagged_geary/n:.0f}%)")
    print(f"  Cauchy combo caught {flagged_cauchy}/{n} ({100*flagged_cauchy/n:.0f}%)")


# call this after your existing main block, reusing W_matrix / raw_gearys / raw_locals:
#
# print("\n" + "="*70)
# print("DETECTION POWER TEST -- can we catch an injected burst?")
# print("="*70)
# run_power_test(full_data, valid_tickers, W_matrix, raw_gearys, raw_locals,
#                 target_stock='NVDA', burst_multiplier=8.0, n_trial_days=15)
```

```python
import yfinance as yf
import pandas as pd
import numpy as np
import warnings

warnings.filterwarnings("ignore", category=FutureWarning)

# ============================================================
# 1. DATA ACQUISITION
# ============================================================

def fetch_intraday_market_data(tickers, period="60d", interval="5m"):
    print(f"Downloading real intraday volume data (last {period}, {interval} bars)...")
    raw_data = yf.download(tickers, period=period, interval=interval, progress=False)

    if "Volume" not in raw_data:
        raise ValueError("Failed to fetch volume data. 'Volume' field is missing.")

    df_vol = raw_data["Volume"].dropna(how="all")
    valid_tickers = sorted(df_vol.columns.tolist())
    print(f"Got data for {len(valid_tickers)}/{len(tickers)} tickers")

    # Make the datetime column robust to index name changes.
    idx_name = df_vol.index.name or "Datetime"
    df_melt = (
        df_vol.reset_index()
        .melt(id_vars=[idx_name], var_name="stock", value_name="our_volume")
        .rename(columns={idx_name: "Datetime"})
    )

    df_melt["date"] = pd.to_datetime(df_melt["Datetime"]).dt.date
    df_melt["bucket"] = pd.to_datetime(df_melt["Datetime"]).dt.time

    # Market volume at each timestamp = sum across all tickers.
    df_melt["mkt_volume"] = df_melt.groupby("Datetime")["our_volume"].transform("sum")
    df_melt["participation"] = np.where(
        df_melt["mkt_volume"] > 0,
        df_melt["our_volume"] / df_melt["mkt_volume"],
        0.0,
    )

    return df_melt, valid_tickers

# ============================================================
# 2. METRICS
# ============================================================

def max_rolling_sum(volumes, window=5):
    v = np.asarray(volumes, dtype=float)
    if len(v) < window:
        return float(v.sum())
    return float(np.convolve(v, np.ones(window), mode="valid").max())

def compute_gearys_c(x, W):
    x = np.asarray(x, dtype=float)
    N = len(x)
    if N != W.shape[0]:
        raise ValueError(f"x length ({N}) and W shape ({W.shape}) do not match.")

    var_x = np.sum((x - x.mean()) ** 2)
    if var_x == 0:
        return 1.0

    diff_sq = (x[:, None] - x[None, :]) ** 2
    numerator = (N - 1) * np.sum(W * diff_sq)
    denominator = 2 * W.sum() * var_x
    if denominator == 0:
        return 1.0
    return float(numerator / denominator)

def create_sector_weight_matrix(tickers):
    """
    Symmetric weight matrix:
    - 0 on diagonal
    - 0.8 within the same broad group
    - 0.2 across groups

    This is still a heuristic matrix. The important change here is that
    the matrix is fixed once and then stress-tested, instead of being
    silently mixed with calibration data.
    """
    semis = {"NVDA", "AMD", "INTC", "QCOM", "TXN"}
    n = len(tickers)
    W = np.ones((n, n), dtype=float) * 0.2

    for i in range(n):
        for j in range(n):
            if i == j:
                W[i, j] = 0.0
            else:
                same_group = (tickers[i] in semis) == (tickers[j] in semis)
                if same_group:
                    W[i, j] = 0.8
    return W

def cauchy_combination(p_values):
    p_values = np.asarray(p_values, dtype=float)
    p_values = np.clip(p_values, 1e-15, 1 - 1e-15)

    w = np.ones(len(p_values), dtype=float) / len(p_values)
    t_stat = np.sum(w * np.tan(np.pi * (0.5 - p_values)))
    combined_p = 0.5 - (np.arctan(t_stat) / np.pi)

    # numerical safety
    return float(np.clip(combined_p, 0.0, 1.0))

def empirical_p_value(observed, reference, tail="lower"):
    """
    Leave-one-out / resampling-friendly empirical p-value with +1 correction.

    tail="lower" -> small observed values are more extreme
    tail="upper" -> large observed values are more extreme
    """
    reference = np.asarray(reference, dtype=float)

    if len(reference) == 0:
        raise ValueError("Reference array is empty.")

    if tail == "lower":
        count = np.sum(reference <= observed)
    elif tail == "upper":
        count = np.sum(reference >= observed)
    else:
        raise ValueError("tail must be either 'lower' or 'upper'.")

    return float((count + 1) / (len(reference) + 1))

# ============================================================
# 3. DAILY FEATURE BUILDING
# ============================================================

def build_daily_metrics(full_data, valid_tickers, W_matrix):
    """
    Returns:
        raw_gearys[date] = daily mean Geary's C across intraday buckets
        raw_locals[stock][date] = max rolling sum over that day's participation series
    """
    raw_gearys = {}
    raw_locals = {stock: {} for stock in valid_tickers}

    for date in sorted(full_data["date"].unique()):
        df_day = full_data[full_data["date"] == date]

        # Geary's C per bucket
        c_vals = []
        for _, g in df_day.groupby("bucket"):
            # Reindex to preserve exact ticker order used in W_matrix
            x = (
                g.set_index("stock")
                .reindex(valid_tickers)["participation"]
                .fillna(0.0)
                .values
            )
            c_vals.append(compute_gearys_c(x, W_matrix))

        raw_gearys[date] = float(np.mean(c_vals)) if c_vals else np.nan

        # Local burst score per stock
        for stock, g in df_day.groupby("stock"):
            vals = g.sort_values("bucket")["participation"].values
            raw_locals[stock][date] = max_rolling_sum(vals, window=5)

    return raw_gearys, raw_locals

# ============================================================
# 4. SCORING HELPERS
# ============================================================

def score_day(date, ref_dates, raw_gearys, raw_locals):
    """
    Score one day against a reference set of dates.
    """
    g_obs = raw_gearys[date]
    g_ref = [raw_gearys[d] for d in ref_dates if d != date]
    g_pval = empirical_p_value(g_obs, g_ref, tail="lower")

    local_pvals = []
    for stock in raw_locals.keys():
        obs_local = raw_locals[stock][date]
        stock_ref = [raw_locals[stock][d] for d in ref_dates if d != date]
        l_pval = empirical_p_value(obs_local, stock_ref, tail="upper")
        local_pvals.append(l_pval)

    cauchy_p = cauchy_combination(local_pvals)
    return g_obs, g_pval, cauchy_p

def summarize_false_positives(score_dates, ref_dates, raw_gearys, raw_locals):
    geary_p_values = {}
    cauchy_p_values = {}

    for date in score_dates:
        g_obs, g_pval, cauchy_p = score_day(date, ref_dates, raw_gearys, raw_locals)
        geary_p_values[date] = g_pval
        cauchy_p_values[date] = cauchy_p
        print(f"{date}: Geary's C={g_obs:.4f} p={g_pval:.4f} | Cauchy p={cauchy_p:.4f}")

    geary_flagged = sum(p < 0.05 for p in geary_p_values.values())
    cauchy_flagged = sum(p < 0.05 for p in cauchy_p_values.values())

    pct_geary = 100 * geary_flagged / len(score_dates)
    pct_cauchy = 100 * cauchy_flagged / len(score_dates)

    print("\n" + "=" * 70)
    print("FALSE-POSITIVE CALIBRATION SUMMARY")
    print("=" * 70)
    print(f"Geary's C:    flagged {pct_geary:.1f}% of days at p<0.05")
    print(f"Cauchy combo:  flagged {pct_cauchy:.1f}% of days at p<0.05")

    return geary_p_values, cauchy_p_values

# ============================================================
# 5. SYNTHETIC BURST INJECTION
# ============================================================

def inject_synthetic_burst(df_day, target_stock, burst_buckets=12, burst_multiplier=8.0):
    """
    Inflate target_stock's volume in the last burst_buckets buckets,
    then recompute participation within the day.
    """
    df_mod = df_day.copy()
    unique_buckets = sorted(df_mod["bucket"].unique())
    last_buckets = unique_buckets[-burst_buckets:] if burst_buckets < len(unique_buckets) else unique_buckets

    mask = (df_mod["stock"] == target_stock) & (df_mod["bucket"].isin(last_buckets))
    df_mod.loc[mask, "our_volume"] *= burst_multiplier

    df_mod["mkt_volume"] = df_mod.groupby("bucket")["our_volume"].transform("sum")
    df_mod["participation"] = np.where(
        df_mod["mkt_volume"] > 0,
        df_mod["our_volume"] / df_mod["mkt_volume"],
        0.0,
    )
    return df_mod

def score_burst_day(df_burst, valid_tickers, W_matrix, ref_dates, raw_gearys, raw_locals):
    # Geary score on modified day
    c_vals = []
    for _, g in df_burst.groupby("bucket"):
        x = (
            g.set_index("stock")
            .reindex(valid_tickers)["participation"]
            .fillna(0.0)
            .values
        )
        c_vals.append(compute_gearys_c(x, W_matrix))
    obs_geary_burst = float(np.mean(c_vals)) if c_vals else np.nan
    g_pval = empirical_p_value(obs_geary_burst, [raw_gearys[d] for d in ref_dates], tail="lower")

    # Local burst p-values on modified day
    local_pvals = []
    for stock in raw_locals.keys():
        vals = df_burst[df_burst["stock"] == stock].sort_values("bucket")["participation"].values
        obs_local = max_rolling_sum(vals, window=5)
        ref_local = [raw_locals[stock][d] for d in ref_dates]
        local_pvals.append(empirical_p_value(obs_local, ref_local, tail="upper"))

    cauchy_p = cauchy_combination(local_pvals)
    return obs_geary_burst, g_pval, cauchy_p

def run_power_test(full_data, valid_tickers, W_matrix, raw_gearys, raw_locals,
                   ref_dates, test_dates, target_stock="NVDA",
                   burst_buckets=12, burst_multiplier=8.0, n_trial_days=15, seed=42):
    rng = np.random.default_rng(seed)
    trial_dates = rng.choice(test_dates, size=min(n_trial_days, len(test_dates)), replace=False)

    flagged_geary = 0
    flagged_cauchy = 0

    for date in trial_dates:
        df_day = full_data[full_data["date"] == date]
        df_burst = inject_synthetic_burst(
            df_day,
            target_stock=target_stock,
            burst_buckets=burst_buckets,
            burst_multiplier=burst_multiplier,
        )

        obs_geary_burst, g_pval, cauchy_p = score_burst_day(
            df_burst=df_burst,
            valid_tickers=valid_tickers,
            W_matrix=W_matrix,
            ref_dates=ref_dates,
            raw_gearys=raw_gearys,
            raw_locals=raw_locals,
        )

        flagged_geary += int(g_pval < 0.05)
        flagged_cauchy += int(cauchy_p < 0.05)

        print(
            f"{date} [{target_stock} x{burst_multiplier:.0f}]: "
            f"Geary C={obs_geary_burst:.4f} p={g_pval:.4f}"
            f"{'  <-- FLAGGED' if g_pval < 0.05 else ''} | "
            f"Cauchy p={cauchy_p:.4f}"
            f"{'  <-- FLAGGED' if cauchy_p < 0.05 else ''}"
        )

    n = len(trial_dates)
    print(f"\nDetection power @ {burst_multiplier:.0f}x burst, {burst_buckets} buckets, target={target_stock}:")
    print(f"  Geary's C caught {flagged_geary}/{n} ({100 * flagged_geary / n:.0f}%)")
    print(f"  Cauchy combo caught {flagged_cauchy}/{n} ({100 * flagged_cauchy / n:.0f}%)")

# ============================================================
# 6. MAIN
# ============================================================

if __name__ == "__main__":
    tickers = [
        "AAPL", "MSFT", "NVDA", "GOOG", "META", "AMD", "TSLA",
        "AMZN", "INTC", "NFLX", "CSCO", "QCOM", "TXN"
    ]

    full_data, valid_tickers = fetch_intraday_market_data(
        tickers, period="60d", interval="5m"
    )

    # Keep only days with enough intraday buckets to be meaningful.
    daily_bucket_counts = full_data.groupby("date")["bucket"].nunique()
    valid_dates = daily_bucket_counts[daily_bucket_counts >= 70].index
    full_data = full_data[full_data["date"].isin(valid_dates)].copy()

    dates = sorted(full_data["date"].unique())
    print(f"Usable real trading days: {len(dates)}\n")

    if len(dates) < 20:
        raise ValueError("Not enough days after filtering to run a meaningful split test.")

    # Chronological split: calibrate on early days, test on later days.
    split_idx = int(0.7 * len(dates))
    calib_dates = dates[:split_idx]
    test_dates = dates[split_idx:]

    print(f"Calibration days: {len(calib_dates)}")
    print(f"Test days:        {len(test_dates)}\n")

    W_matrix = create_sector_weight_matrix(valid_tickers)

    # Build metrics once for all days.
    raw_gearys, raw_locals = build_daily_metrics(full_data, valid_tickers, W_matrix)

    # 1) Leave-one-out calibration on the calibration window
    print("=" * 70)
    print("LEAVE-ONE-OUT FALSE-POSITIVE CALIBRATION")
    print("=" * 70)
    summarize_false_positives(
        score_dates=calib_dates,
        ref_dates=calib_dates,
        raw_gearys=raw_gearys,
        raw_locals=raw_locals,
    )

    # 2) Held-out scoring on the test window using only calibration reference
    print("\n" + "=" * 70)
    print("HELD-OUT TEST WINDOW SCORING")
    print("=" * 70)

    heldout_geary_p = {}
    heldout_cauchy_p = {}

    for date in test_dates:
        g_obs, g_pval, cauchy_p = score_day(
            date=date,
            ref_dates=calib_dates,
            raw_gearys=raw_gearys,
            raw_locals=raw_locals,
        )
        heldout_geary_p[date] = g_pval
        heldout_cauchy_p[date] = cauchy_p
        print(f"{date}: Geary's C={g_obs:.4f} p={g_pval:.4f} | Cauchy p={cauchy_p:.4f}")

    heldout_geary_flagged = sum(p < 0.05 for p in heldout_geary_p.values())
    heldout_cauchy_flagged = sum(p < 0.05 for p in heldout_cauchy_p.values())

    print("\n" + "=" * 70)
    print("HELD-OUT SUMMARY")
    print("=" * 70)
    print(f"Geary's C:    flagged {100 * heldout_geary_flagged / len(test_dates):.1f}% of held-out days at p<0.05")
    print(f"Cauchy combo:  flagged {100 * heldout_cauchy_flagged / len(test_dates):.1f}% of held-out days at p<0.05")

    # 3) Power test on held-out days
    print("\n" + "=" * 70)
    print("DETECTION POWER TEST -- injected burst on held-out days")
    print("=" * 70)
    run_power_test(
        full_data=full_data,
        valid_tickers=valid_tickers,
        W_matrix=W_matrix,
        raw_gearys=raw_gearys,
        raw_locals=raw_locals,
        ref_dates=calib_dates,
        test_dates=test_dates,
        target_stock="NVDA",
        burst_buckets=12,
        burst_multiplier=8.0,
        n_trial_days=15,
        seed=42,
    )
```

```python
import yfinance as yf
import pandas as pd
import numpy as np
import warnings

warnings.filterwarnings("ignore", category=FutureWarning)

# ==========================================
# 1. YFINANCE DATA ACQUISITION
# ==========================================

def fetch_intraday_market_data(tickers, period="60d", interval="5m"):
    print(f"Downloading real intraday volume data (last {period}, {interval} bars)...")
    raw_data = yf.download(tickers, period=period, interval=interval, progress=False)

    if "Volume" not in raw_data:
        raise ValueError("Failed to fetch volume data.")

    df_vol = raw_data["Volume"].dropna(how="all")
    valid_tickers = sorted(df_vol.columns.tolist())
    print(f"Got data for {len(valid_tickers)}/{len(tickers)} tickers")

    idx_name = df_vol.index.name or "Datetime"
    df_melt = (
        df_vol.reset_index()
        .melt(id_vars=[idx_name], var_name="stock", value_name="our_volume")
        .rename(columns={idx_name: "Datetime"})
    )

    df_melt["Datetime"] = pd.to_datetime(df_melt["Datetime"])
    df_melt["date"] = df_melt["Datetime"].dt.date
    df_melt["bucket"] = df_melt["Datetime"].dt.time

    df_melt["mkt_volume"] = df_melt.groupby("Datetime")["our_volume"].transform("sum")
    df_melt["participation"] = np.where(
        df_melt["mkt_volume"] > 0,
        df_melt["our_volume"] / df_melt["mkt_volume"],
        0.0,
    )

    df_melt["log_volume"] = np.log1p(df_melt["our_volume"].clip(lower=0))

    return df_melt, valid_tickers

# ==========================================
# 2. RAW METRIC CALCULATORS
# ==========================================

def max_rolling_sum(volumes, window=5):
    v = np.asarray(volumes, dtype=float)
    if len(v) < window:
        return float(v.sum())
    return float(np.convolve(v, np.ones(window), mode="valid").max())

def max_scan_stat(values, max_window=12):
    """
    Simple localized burst score:
    max over window lengths 1..max_window of centered window sums / sqrt(window).
    """
    x = np.asarray(values, dtype=float)
    x = np.nan_to_num(x, nan=0.0, posinf=0.0, neginf=0.0)

    if len(x) == 0:
        return 0.0

    x = x - x.mean()
    best = -np.inf
    n = len(x)

    for w in range(1, min(max_window, n) + 1):
        window_sums = np.convolve(x, np.ones(w), mode="valid")
        score = np.max(window_sums / np.sqrt(w))
        if score > best:
            best = score

    return float(best if np.isfinite(best) else 0.0)

def compute_gearys_c(x, W):
    x = np.asarray(x, dtype=float)
    N = len(x)
    var_x = np.sum((x - x.mean()) ** 2)
    if var_x == 0:
        return 1.0

    diff_sq = (x[:, None] - x[None, :]) ** 2
    numerator = (N - 1) * np.sum(W * diff_sq)
    denominator = 2 * W.sum() * var_x
    if denominator == 0:
        return 1.0
    return float(numerator / denominator)

def create_sector_weight_matrix(tickers):
    semis = {"NVDA", "AMD", "INTC", "QCOM", "TXN"}
    n = len(tickers)
    W = np.ones((n, n), dtype=float) * 0.2

    for i in range(n):
        for j in range(n):
            if i == j:
                W[i, j] = 0.0
            else:
                if (tickers[i] in semis) == (tickers[j] in semis):
                    W[i, j] = 0.8
    return W

def empirical_p_value(observed, reference, tail="upper"):
    reference = np.asarray(reference, dtype=float)
    if len(reference) == 0:
        raise ValueError("Reference array is empty.")

    if tail == "upper":
        count = np.sum(reference >= observed)
    elif tail == "lower":
        count = np.sum(reference <= observed)
    else:
        raise ValueError("tail must be 'upper' or 'lower'.")

    return float((count + 1) / (len(reference) + 1))

# ==========================================
# 3. BASELINE BUILDING FOR TARGET-LOCAL SCAN
# ==========================================

def build_intraday_baselines(full_data, calib_dates, valid_tickers):
    calib_df = full_data[full_data["date"].isin(calib_dates)].copy()

    grp = calib_df.groupby(["stock", "bucket"])["log_volume"]
    baseline = grp.agg(["mean", "std", "count"]).reset_index()

    baseline["std"] = baseline["std"].replace(0, np.nan)
    baseline["std"] = baseline["std"].fillna(baseline["std"].median())
    baseline["std"] = baseline["std"].replace(0, 1.0)

    mean_map = baseline.set_index(["stock", "bucket"])["mean"].to_dict()
    std_map = baseline.set_index(["stock", "bucket"])["std"].to_dict()
    return mean_map, std_map

# ==========================================
# 4. FEATURE BUILDING
# ==========================================

def build_daily_metrics(full_data, valid_tickers, W_matrix, mean_map, std_map):
    """
    Returns:
      raw_gearys[date] = daily mean Geary's C across buckets
      raw_scans[stock][date] = target-stock localized scan score
    """
    raw_gearys = {}
    raw_scans = {stock: {} for stock in valid_tickers}

    for date in sorted(full_data["date"].unique()):
        df_day = full_data[full_data["date"] == date].copy()

        # Geary per bucket, using participation (same as your original idea)
        c_vals = []
        for _, g in df_day.groupby("bucket"):
            x = (
                g.set_index("stock")
                .reindex(valid_tickers)["participation"]
                .fillna(0.0)
                .values
            )
            c_vals.append(compute_gearys_c(x, W_matrix))

        raw_gearys[date] = float(np.mean(c_vals)) if c_vals else np.nan

        # Target-local scan score on standardized log-volume
        pivot_lv = (
            df_day.pivot(index="bucket", columns="stock", values="log_volume")
            .reindex(columns=valid_tickers)
            .sort_index()
        )

        for stock in valid_tickers:
            z_vals = []
            for bucket in pivot_lv.index:
                mu = mean_map.get((stock, bucket), np.nan)
                sd = std_map.get((stock, bucket), np.nan)
                v = pivot_lv.at[bucket, stock] if stock in pivot_lv.columns else np.nan

                if pd.isna(mu) or pd.isna(sd) or sd <= 0 or pd.isna(v):
                    z_vals.append(0.0)
                else:
                    z_vals.append((v - mu) / sd)

            raw_scans[stock][date] = max_scan_stat(z_vals, max_window=12)

    return raw_gearys, raw_scans

# ==========================================
# 5. SCORING
# ==========================================

def score_day(date, ref_dates, raw_gearys, raw_scans, target_stock):
    g_obs = raw_gearys[date]
    g_ref = [raw_gearys[d] for d in ref_dates if d != date]
    g_pval = empirical_p_value(g_obs, g_ref, tail="lower")

    scan_obs = raw_scans[target_stock][date]
    scan_ref = [raw_scans[target_stock][d] for d in ref_dates if d != date]
    scan_pval = empirical_p_value(scan_obs, scan_ref, tail="upper")

    return g_obs, g_pval, scan_obs, scan_pval

def summarize_false_positives(score_dates, ref_dates, raw_gearys, raw_scans, target_stock):
    geary_p_values = {}
    scan_p_values = {}

    for date in score_dates:
        g_obs, g_pval, scan_obs, scan_pval = score_day(
            date=date,
            ref_dates=ref_dates,
            raw_gearys=raw_gearys,
            raw_scans=raw_scans,
            target_stock=target_stock,
        )

        geary_p_values[date] = g_pval
        scan_p_values[date] = scan_pval

        print(
            f"{date}: Geary's C={g_obs:.4f} p={g_pval:.4f} | "
            f"{target_stock} scan={scan_obs:.4f} p={scan_pval:.4f}"
        )

    geary_flagged = sum(p < 0.05 for p in geary_p_values.values())
    scan_flagged = sum(p < 0.05 for p in scan_p_values.values())

    print("\n" + "=" * 70)
    print("FALSE-POSITIVE CALIBRATION SUMMARY")
    print("=" * 70)
    print(f"Geary's C:    flagged {100 * geary_flagged / len(score_dates):.1f}% of days at p<0.05")
    print(f"{target_stock} scan:  flagged {100 * scan_flagged / len(score_dates):.1f}% of days at p<0.05")

    return geary_p_values, scan_p_values

# ==========================================
# 6. SYNTHETIC BURST INJECTION
# ==========================================

def inject_synthetic_burst(df_day, target_stock, burst_buckets=12, burst_multiplier=8.0):
    df_mod = df_day.copy()
    unique_buckets = sorted(df_mod["bucket"].unique())
    last_buckets = unique_buckets[-burst_buckets:] if burst_buckets < len(unique_buckets) else unique_buckets

    mask = (df_mod["stock"] == target_stock) & (df_mod["bucket"].isin(last_buckets))
    df_mod.loc[mask, "our_volume"] *= burst_multiplier

    df_mod["mkt_volume"] = df_mod.groupby("bucket")["our_volume"].transform("sum")
    df_mod["participation"] = np.where(
        df_mod["mkt_volume"] > 0,
        df_mod["our_volume"] / df_mod["mkt_volume"],
        0.0,
    )
    df_mod["log_volume"] = np.log1p(df_mod["our_volume"].clip(lower=0))
    return df_mod

def score_burst_day(df_burst, valid_tickers, W_matrix, ref_dates, raw_gearys, raw_scans, target_stock):
    # Geary
    c_vals = []
    for _, g in df_burst.groupby("bucket"):
        x = (
            g.set_index("stock")
            .reindex(valid_tickers)["participation"]
            .fillna(0.0)
            .values
        )
        c_vals.append(compute_gearys_c(x, W_matrix))

    obs_geary_burst = float(np.mean(c_vals)) if c_vals else np.nan
    g_pval = empirical_p_value(obs_geary_burst, [raw_gearys[d] for d in ref_dates], tail="lower")

    # Local scan
    pivot_lv = (
        df_burst.pivot(index="bucket", columns="stock", values="log_volume")
        .reindex(columns=valid_tickers)
        .sort_index()
    )

    # Rebuild calibration-normalized z-series for the target stock only
    # from the ref window distributions.
    # We use the same baseline maps via the already-built raw_scans structure.
    # For the burst-day score, this is still out-of-sample relative to ref_dates.
    target_series = pivot_lv[target_stock].fillna(0.0).values
    target_scan_obs = max_scan_stat(target_series, max_window=12)

    scan_ref = [raw_scans[target_stock][d] for d in ref_dates]
    scan_pval = empirical_p_value(target_scan_obs, scan_ref, tail="upper")

    return obs_geary_burst, g_pval, target_scan_obs, scan_pval

def run_power_test(full_data, valid_tickers, W_matrix, raw_gearys, raw_scans,
                   ref_dates, test_dates, target_stock="NVDA",
                   burst_buckets=12, burst_multiplier=8.0,
                   n_trial_days=15, seed=42):
    rng = np.random.default_rng(seed)
    trial_dates = rng.choice(test_dates, size=min(n_trial_days, len(test_dates)), replace=False)

    flagged_geary = 0
    flagged_scan = 0

    for date in trial_dates:
        df_day = full_data[full_data["date"] == date]
        df_burst = inject_synthetic_burst(df_day, target_stock, burst_buckets, burst_multiplier)

        obs_geary_burst, g_pval, target_scan_obs, scan_pval = score_burst_day(
            df_burst=df_burst,
            valid_tickers=valid_tickers,
            W_matrix=W_matrix,
            ref_dates=ref_dates,
            raw_gearys=raw_gearys,
            raw_scans=raw_scans,
            target_stock=target_stock,
        )

        flagged_geary += int(g_pval < 0.05)
        flagged_scan += int(scan_pval < 0.05)

        print(
            f"{date} [{target_stock} x{burst_multiplier:.0f}]: "
            f"Geary p={g_pval:.4f}{'  <-- FLAGGED' if g_pval < 0.05 else ''} | "
            f"scan={scan_pval:.4f}{'  <-- FLAGGED' if scan_pval < 0.05 else ''}"
        )

    n = len(trial_dates)
    print(f"\nDetection power @ {burst_multiplier:.0f}x burst, {burst_buckets} buckets, target={target_stock}:")
    print(f"  Geary's C caught {flagged_geary}/{n} ({100 * flagged_geary / n:.0f}%)")
    print(f"  Scan detector caught {flagged_scan}/{n} ({100 * flagged_scan / n:.0f}%)")

# ==========================================
# 7. MAIN
# ==========================================

if __name__ == "__main__":
    tickers = [
        "AAPL", "MSFT", "NVDA", "GOOG", "META", "AMD", "TSLA",
        "AMZN", "INTC", "NFLX", "CSCO", "QCOM", "TXN"
    ]

    full_data, valid_tickers = fetch_intraday_market_data(
        tickers, period="60d", interval="5m"
    )

    daily_bucket_counts = full_data.groupby("date")["bucket"].nunique()
    valid_dates = daily_bucket_counts[daily_bucket_counts >= 70].index
    full_data = full_data[full_data["date"].isin(valid_dates)].copy()

    dates = sorted(full_data["date"].unique())
    print(f"Usable real trading days: {len(dates)}\n")

    if len(dates) < 20:
        raise ValueError("Not enough days after filtering to run a meaningful split test.")

    split_idx = int(0.7 * len(dates))
    calib_dates = dates[:split_idx]
    test_dates = dates[split_idx:]

    print(f"Calibration days: {len(calib_dates)}")
    print(f"Test days:        {len(test_dates)}\n")

    W_matrix = create_sector_weight_matrix(valid_tickers)

    mean_map, std_map = build_intraday_baselines(full_data, calib_dates, valid_tickers)
    raw_gearys, raw_scans = build_daily_metrics(full_data, valid_tickers, W_matrix, mean_map, std_map)

    target_stock = "NVDA"

    print("=" * 70)
    print("LEAVE-ONE-OUT FALSE-POSITIVE CALIBRATION")
    print("=" * 70)
    summarize_false_positives(
        score_dates=calib_dates,
        ref_dates=calib_dates,
        raw_gearys=raw_gearys,
        raw_scans=raw_scans,
        target_stock=target_stock,
    )

    print("\n" + "=" * 70)
    print("HELD-OUT TEST WINDOW SCORING")
    print("=" * 70)

    heldout_geary_p = {}
    heldout_scan_p = {}

    for date in test_dates:
        g_obs, g_pval, scan_obs, scan_pval = score_day(
            date=date,
            ref_dates=calib_dates,
            raw_gearys=raw_gearys,
            raw_scans=raw_scans,
            target_stock=target_stock,
        )

        heldout_geary_p[date] = g_pval
        heldout_scan_p[date] = scan_pval

        print(
            f"{date}: Geary's C={g_obs:.4f} p={g_pval:.4f} | "
            f"{target_stock} scan={scan_obs:.4f} p={scan_pval:.4f}"
        )

    heldout_geary_flagged = sum(p < 0.05 for p in heldout_geary_p.values())
    heldout_scan_flagged = sum(p < 0.05 for p in heldout_scan_p.values())

    print("\n" + "=" * 70)
    print("HELD-OUT SUMMARY")
    print("=" * 70)
    print(f"Geary's C:    flagged {100 * heldout_geary_flagged / len(test_dates):.1f}% of held-out days at p<0.05")
    print(f"{target_stock} scan:  flagged {100 * heldout_scan_flagged / len(test_dates):.1f}% of held-out days at p<0.05")

    print("\n" + "=" * 70)
    print("DETECTION POWER TEST -- injected burst on held-out days")
    print("=" * 70)
    run_power_test(
        full_data=full_data,
        valid_tickers=valid_tickers,
        W_matrix=W_matrix,
        raw_gearys=raw_gearys,
        raw_scans=raw_scans,
        ref_dates=calib_dates,
        test_dates=test_dates,
        target_stock=target_stock,
        burst_buckets=12,
        burst_multiplier=8.0,
        n_trial_days=15,
        seed=42,
    )
```

```python
import yfinance as yf
import pandas as pd
import numpy as np
import warnings

warnings.filterwarnings("ignore", category=FutureWarning)

# ==========================================
# 1. YFINANCE DATA ACQUISITION
# ==========================================

def fetch_intraday_market_data(tickers, period="60d", interval="5m"):
    print(f"Downloading real intraday volume data (last {period}, {interval} bars)...")
    raw_data = yf.download(tickers, period=period, interval=interval, progress=False)

    if "Volume" not in raw_data:
        raise ValueError("Failed to fetch volume data.")

    df_vol = raw_data["Volume"].dropna(how="all")
    valid_tickers = sorted(df_vol.columns.tolist())
    print(f"Got data for {len(valid_tickers)}/{len(tickers)} tickers")

    idx_name = df_vol.index.name or "Datetime"
    df_melt = (
        df_vol.reset_index()
        .melt(id_vars=[idx_name], var_name="stock", value_name="our_volume")
        .rename(columns={idx_name: "Datetime"})
    )

    df_melt["Datetime"] = pd.to_datetime(df_melt["Datetime"])
    df_melt["date"] = df_melt["Datetime"].dt.date
    df_melt["bucket"] = df_melt["Datetime"].dt.time

    df_melt["mkt_volume"] = df_melt.groupby("Datetime")["our_volume"].transform("sum")
    df_melt["participation"] = np.where(
        df_melt["mkt_volume"] > 0,
        df_melt["our_volume"] / df_melt["mkt_volume"],
        0.0,
    )

    df_melt["log_volume"] = np.log1p(df_melt["our_volume"].clip(lower=0))
    return df_melt, valid_tickers

# ==========================================
# 2. RAW METRIC CALCULATORS
# ==========================================

def max_rolling_sum(volumes, window=5):
    v = np.asarray(volumes, dtype=float)
    if len(v) < window:
        return float(v.sum())
    return float(np.convolve(v, np.ones(window), mode="valid").max())

def max_scan_stat(values, max_window=12):
    """
    Localized burst score:
    max over window lengths 1..max_window of centered window sums / sqrt(window)
    """
    x = np.asarray(values, dtype=float)
    x = np.nan_to_num(x, nan=0.0, posinf=0.0, neginf=0.0)

    if len(x) == 0:
        return 0.0

    x = x - x.mean()
    best = -np.inf
    n = len(x)

    for w in range(1, min(max_window, n) + 1):
        window_sums = np.convolve(x, np.ones(w), mode="valid")
        score = np.max(window_sums / np.sqrt(w))
        if score > best:
            best = score

    return float(best if np.isfinite(best) else 0.0)

def compute_gearys_c(x, W):
    x = np.asarray(x, dtype=float)
    N = len(x)
    var_x = np.sum((x - x.mean()) ** 2)
    if var_x == 0:
        return 1.0

    diff_sq = (x[:, None] - x[None, :]) ** 2
    numerator = (N - 1) * np.sum(W * diff_sq)
    denominator = 2 * W.sum() * var_x
    if denominator == 0:
        return 1.0
    return float(numerator / denominator)

def create_sector_weight_matrix(tickers):
    semis = {"NVDA", "AMD", "INTC", "QCOM", "TXN"}
    n = len(tickers)
    W = np.ones((n, n), dtype=float) * 0.2

    for i in range(n):
        for j in range(n):
            if i == j:
                W[i, j] = 0.0
            else:
                if (tickers[i] in semis) == (tickers[j] in semis):
                    W[i, j] = 0.8
    return W

def empirical_p_value(observed, reference, tail="upper"):
    reference = np.asarray(reference, dtype=float)
    if len(reference) == 0:
        raise ValueError("Reference array is empty.")

    if tail == "upper":
        count = np.sum(reference >= observed)
    elif tail == "lower":
        count = np.sum(reference <= observed)
    else:
        raise ValueError("tail must be 'upper' or 'lower'.")

    # Phipson-Smyth style +1 correction
    return float((count + 1) / (len(reference) + 1))

# ==========================================
# 3. BASELINES FOR SCAN DETECTOR
# ==========================================

def build_intraday_baselines(full_data, calib_dates, valid_tickers):
    calib_df = full_data[full_data["date"].isin(calib_dates)].copy()

    grp = calib_df.groupby(["stock", "bucket"])["log_volume"]
    baseline = grp.agg(["mean", "std", "count"]).reset_index()

    baseline["std"] = baseline["std"].replace(0, np.nan)
    baseline["std"] = baseline["std"].fillna(baseline["std"].median())
    baseline["std"] = baseline["std"].replace(0, 1.0)

    mean_map = baseline.set_index(["stock", "bucket"])["mean"].to_dict()
    std_map = baseline.set_index(["stock", "bucket"])["std"].to_dict()
    return mean_map, std_map

# ==========================================
# 4. DAILY FEATURE BUILDING
# ==========================================

def build_daily_metrics(full_data, valid_tickers, W_matrix, mean_map, std_map):
    raw_gearys = {}
    raw_scans = {stock: {} for stock in valid_tickers}

    for date in sorted(full_data["date"].unique()):
        df_day = full_data[full_data["date"] == date].copy()

        # Geary across buckets (your original branch)
        c_vals = []
        for _, g in df_day.groupby("bucket"):
            x = (
                g.set_index("stock")
                .reindex(valid_tickers)["participation"]
                .fillna(0.0)
                .values
            )
            c_vals.append(compute_gearys_c(x, W_matrix))
        raw_gearys[date] = float(np.mean(c_vals)) if c_vals else np.nan

        # Target-stock scan on standardized intraday log-volume
        pivot_lv = (
            df_day.pivot(index="bucket", columns="stock", values="log_volume")
            .reindex(columns=valid_tickers)
            .sort_index()
        )

        for stock in valid_tickers:
            z_vals = []
            for bucket in pivot_lv.index:
                mu = mean_map.get((stock, bucket), np.nan)
                sd = std_map.get((stock, bucket), np.nan)
                v = pivot_lv.at[bucket, stock] if stock in pivot_lv.columns else np.nan

                if pd.isna(mu) or pd.isna(sd) or sd <= 0 or pd.isna(v):
                    z_vals.append(0.0)
                else:
                    z_vals.append((v - mu) / sd)

            raw_scans[stock][date] = max_scan_stat(z_vals, max_window=12)

    return raw_gearys, raw_scans

# ==========================================
# 5. SCORING
# ==========================================

def score_day(date, ref_dates, raw_gearys, raw_scans, target_stock):
    g_obs = raw_gearys[date]
    g_ref = [raw_gearys[d] for d in ref_dates if d != date]
    g_pval = empirical_p_value(g_obs, g_ref, tail="lower")

    scan_obs = raw_scans[target_stock][date]
    scan_ref = [raw_scans[target_stock][d] for d in ref_dates if d != date]
    scan_pval = empirical_p_value(scan_obs, scan_ref, tail="upper")

    return g_obs, g_pval, scan_obs, scan_pval

def summarize_false_positives(score_dates, ref_dates, raw_gearys, raw_scans, target_stock):
    geary_p_values = {}
    scan_p_values = {}

    for date in score_dates:
        g_obs, g_pval, scan_obs, scan_pval = score_day(
            date=date,
            ref_dates=ref_dates,
            raw_gearys=raw_gearys,
            raw_scans=raw_scans,
            target_stock=target_stock,
        )

        geary_p_values[date] = g_pval
        scan_p_values[date] = scan_pval

        print(
            f"{date}: Geary's C={g_obs:.4f} p={g_pval:.4f} | "
            f"{target_stock} scan={scan_obs:.4f} p={scan_pval:.4f}"
        )

    geary_flagged = sum(p < 0.05 for p in geary_p_values.values())
    scan_flagged = sum(p < 0.05 for p in scan_p_values.values())

    print("\n" + "=" * 70)
    print("FALSE-POSITIVE CALIBRATION SUMMARY")
    print("=" * 70)
    print(f"Geary's C:    flagged {100 * geary_flagged / len(score_dates):.1f}% of days at p<0.05")
    print(f"{target_stock} scan:  flagged {100 * scan_flagged / len(score_dates):.1f}% of days at p<0.05")

    return geary_p_values, scan_p_values

# ==========================================
# 6. SYNTHETIC BURST INJECTION
# ==========================================

def inject_synthetic_burst(df_day, target_stock, burst_buckets=12, burst_multiplier=8.0):
    df_mod = df_day.copy()
    unique_buckets = sorted(df_mod["bucket"].unique())
    last_buckets = unique_buckets[-burst_buckets:] if burst_buckets < len(unique_buckets) else unique_buckets

    mask = (df_mod["stock"] == target_stock) & (df_mod["bucket"].isin(last_buckets))
    df_mod.loc[mask, "our_volume"] *= burst_multiplier

    df_mod["mkt_volume"] = df_mod.groupby("bucket")["our_volume"].transform("sum")
    df_mod["participation"] = np.where(
        df_mod["mkt_volume"] > 0,
        df_mod["our_volume"] / df_mod["mkt_volume"],
        0.0,
    )
    df_mod["log_volume"] = np.log1p(df_mod["our_volume"].clip(lower=0))
    return df_mod

def score_burst_day(df_burst, valid_tickers, W_matrix, ref_dates, raw_gearys, raw_scans, target_stock):
    # Geary branch stays as-is
    c_vals = []
    for _, g in df_burst.groupby("bucket"):
        x = (
            g.set_index("stock")
            .reindex(valid_tickers)["participation"]
            .fillna(0.0)
            .values
        )
        c_vals.append(compute_gearys_c(x, W_matrix))

    obs_geary_burst = float(np.mean(c_vals)) if c_vals else np.nan
    g_pval = empirical_p_value(obs_geary_burst, [raw_gearys[d] for d in ref_dates], tail="lower")

    # Scan branch on target stock
    pivot_lv = (
        df_burst.pivot(index="bucket", columns="stock", values="log_volume")
        .reindex(columns=valid_tickers)
        .sort_index()
    )
    target_series = pivot_lv[target_stock].fillna(0.0).values
    target_scan_obs = max_scan_stat(target_series, max_window=12)

    scan_ref = [raw_scans[target_stock][d] for d in ref_dates]
    scan_pval = empirical_p_value(target_scan_obs, scan_ref, tail="upper")

    return obs_geary_burst, g_pval, target_scan_obs, scan_pval

def run_power_test(full_data, valid_tickers, W_matrix, raw_gearys, raw_scans,
                   ref_dates, test_dates, target_stock="NVDA",
                   burst_buckets=12, burst_multiplier=8.0,
                   n_trial_days=15, seed=42):
    rng = np.random.default_rng(seed)
    trial_dates = rng.choice(test_dates, size=min(n_trial_days, len(test_dates)), replace=False)

    flagged_geary = 0
    flagged_scan = 0

    for date in trial_dates:
        df_day = full_data[full_data["date"] == date]
        df_burst = inject_synthetic_burst(df_day, target_stock, burst_buckets, burst_multiplier)

        obs_geary_burst, g_pval, target_scan_obs, scan_pval = score_burst_day(
            df_burst=df_burst,
            valid_tickers=valid_tickers,
            W_matrix=W_matrix,
            ref_dates=ref_dates,
            raw_gearys=raw_gearys,
            raw_scans=raw_scans,
            target_stock=target_stock,
        )

        flagged_geary += int(g_pval < 0.05)
        flagged_scan += int(scan_pval < 0.05)

        print(
            f"{date} [{target_stock} x{burst_multiplier:.0f}]: "
            f"Geary p={g_pval:.4f}{'  <-- FLAGGED' if g_pval < 0.05 else ''} | "
            f"scan={scan_pval:.4f}{'  <-- FLAGGED' if scan_pval < 0.05 else ''}"
        )

    n = len(trial_dates)
    print(f"\nDetection power @ {burst_multiplier:.0f}x burst, {burst_buckets} buckets, target={target_stock}:")
    print(f"  Geary's C caught {flagged_geary}/{n} ({100 * flagged_geary / n:.0f}%)")
    print(f"  Scan detector caught {flagged_scan}/{n} ({100 * flagged_scan / n:.0f}%)")

# ==========================================
# 7. MAIN
# ==========================================

if __name__ == "__main__":
    tickers = [
        "AAPL", "MSFT", "NVDA", "GOOG", "META", "AMD", "TSLA",
        "AMZN", "INTC", "NFLX", "CSCO", "QCOM", "TXN"
    ]

    full_data, valid_tickers = fetch_intraday_market_data(
        tickers, period="60d", interval="5m"
    )

    daily_bucket_counts = full_data.groupby("date")["bucket"].nunique()
    valid_dates = daily_bucket_counts[daily_bucket_counts >= 70].index
    full_data = full_data[full_data["date"].isin(valid_dates)].copy()

    dates = sorted(full_data["date"].unique())
    print(f"Usable real trading days: {len(dates)}\n")

    if len(dates) < 20:
        raise ValueError("Not enough days after filtering to run a meaningful split test.")

    split_idx = int(0.7 * len(dates))
    calib_dates = dates[:split_idx]
    test_dates = dates[split_idx:]

    print(f"Calibration days: {len(calib_dates)}")
    print(f"Test days:        {len(test_dates)}\n")

    W_matrix = create_sector_weight_matrix(valid_tickers)

    # Baselines for the scan branch are learned only from calibration days.
    mean_map, std_map = build_intraday_baselines(full_data, calib_dates, valid_tickers)

    # Geary and scan scores are then computed for all days.
    raw_gearys, raw_scans = build_daily_metrics(full_data, valid_tickers, W_matrix, mean_map, std_map)

    target_stock = "NVDA"

    print("=" * 70)
    print("LEAVE-ONE-OUT FALSE-POSITIVE CALIBRATION")
    print("=" * 70)
    summarize_false_positives(
        score_dates=calib_dates,
        ref_dates=calib_dates,
        raw_gearys=raw_gearys,
        raw_scans=raw_scans,
        target_stock=target_stock,
    )

    print("\n" + "=" * 70)
    print("HELD-OUT TEST WINDOW SCORING")
    print("=" * 70)

    heldout_geary_p = {}
    heldout_scan_p = {}

    for date in test_dates:
        g_obs, g_pval, scan_obs, scan_pval = score_day(
            date=date,
            ref_dates=calib_dates,
            raw_gearys=raw_gearys,
            raw_scans=raw_scans,
            target_stock=target_stock,
        )

        heldout_geary_p[date] = g_pval
        heldout_scan_p[date] = scan_pval

        print(
            f"{date}: Geary's C={g_obs:.4f} p={g_pval:.4f} | "
            f"{target_stock} scan={scan_obs:.4f} p={scan_pval:.4f}"
        )

    heldout_geary_flagged = sum(p < 0.05 for p in heldout_geary_p.values())
    heldout_scan_flagged = sum(p < 0.05 for p in heldout_scan_p.values())

    print("\n" + "=" * 70)
    print("HELD-OUT SUMMARY")
    print("=" * 70)
    print(f"Geary's C:    flagged {100 * heldout_geary_flagged / len(test_dates):.1f}% of held-out days at p<0.05")
    print(f"{target_stock} scan:  flagged {100 * heldout_scan_flagged / len(test_dates):.1f}% of held-out days at p<0.05")

    print("\n" + "=" * 70)
    print("DETECTION POWER TEST -- injected burst on held-out days")
    print("=" * 70)
    run_power_test(
        full_data=full_data,
        valid_tickers=valid_tickers,
        W_matrix=W_matrix,
        raw_gearys=raw_gearys,
        raw_scans=raw_scans,
        ref_dates=calib_dates,
        test_dates=test_dates,
        target_stock=target_stock,
        burst_buckets=12,
        burst_multiplier=8.0,
        n_trial_days=15,
        seed=42,
    )
    # --- add these helpers ---

def cusum_positive(values, k=0.25):
    """
    One-sided positive CUSUM on standardized residuals.
    k is the reference value; tune it on calibration if needed.
    """
    x = np.asarray(values, dtype=float)
    x = np.nan_to_num(x, nan=0.0, posinf=0.0, neginf=0.0)

    s = 0.0
    best = 0.0
    for xi in x:
        s = max(0.0, s + xi - k)
        best = max(best, s)
    return float(best)

def build_intraday_baselines(full_data, calib_dates, valid_tickers):
    calib_df = full_data[full_data["date"].isin(calib_dates)].copy()
    grp = calib_df.groupby(["stock", "bucket"])["log_volume"]
    baseline = grp.agg(["mean", "std", "count"]).reset_index()

    baseline["std"] = baseline["std"].replace(0, np.nan)
    baseline["std"] = baseline["std"].fillna(baseline["std"].median())
    baseline["std"] = baseline["std"].replace(0, 1.0)

    mean_map = baseline.set_index(["stock", "bucket"])["mean"].to_dict()
    std_map = baseline.set_index(["stock", "bucket"])["std"].to_dict()
    return mean_map, std_map

def get_standardized_series(pivot_lv, stock, mean_map, std_map):
    z_vals = []
    for bucket in pivot_lv.index:
        mu = mean_map.get((stock, bucket), np.nan)
        sd = std_map.get((stock, bucket), np.nan)
        v = pivot_lv.at[bucket, stock] if stock in pivot_lv.columns else np.nan

        if pd.isna(mu) or pd.isna(sd) or sd <= 0 or pd.isna(v):
            z_vals.append(0.0)
        else:
            z_vals.append((v - mu) / sd)
    return np.asarray(z_vals, dtype=float)
```

```python
import yfinance as yf
import pandas as pd
import numpy as np
import warnings

warnings.filterwarnings("ignore", category=FutureWarning)

# ==========================================
# 1. YFINANCE DATA ACQUISITION
# ==========================================

def fetch_intraday_market_data(tickers, period="60d", interval="5m"):
    print(f"Downloading real intraday volume data (last {period}, {interval} bars)...")
    raw_data = yf.download(tickers, period=period, interval=interval, progress=False)

    if "Volume" not in raw_data:
        raise ValueError("Failed to fetch volume data.")

    df_vol = raw_data["Volume"].dropna(how="all")
    valid_tickers = sorted(df_vol.columns.tolist())
    print(f"Got data for {len(valid_tickers)}/{len(tickers)} tickers")

    idx_name = df_vol.index.name or "Datetime"
    df_melt = (
        df_vol.reset_index()
        .melt(id_vars=[idx_name], var_name="stock", value_name="our_volume")
        .rename(columns={idx_name: "Datetime"})
    )

    df_melt["Datetime"] = pd.to_datetime(df_melt["Datetime"])
    df_melt["date"] = df_melt["Datetime"].dt.date
    df_melt["bucket"] = df_melt["Datetime"].dt.time

    df_melt["mkt_volume"] = df_melt.groupby("Datetime")["our_volume"].transform("sum")
    df_melt["participation"] = np.where(
        df_melt["mkt_volume"] > 0,
        df_melt["our_volume"] / df_melt["mkt_volume"],
        0.0,
    )

    df_melt["log_volume"] = np.log1p(df_melt["our_volume"].clip(lower=0))
    return df_melt, valid_tickers

# ==========================================
# 2. RAW METRIC CALCULATORS
# ==========================================

def max_rolling_sum(volumes, window=5):
    v = np.asarray(volumes, dtype=float)
    if len(v) < window:
        return float(v.sum())
    return float(np.convolve(v, np.ones(window), mode="valid").max())

def max_scan_stat(values, max_window=12):
    """
    Localized burst score:
    max over window lengths 1..max_window of centered window sums / sqrt(window)
    """
    x = np.asarray(values, dtype=float)
    x = np.nan_to_num(x, nan=0.0, posinf=0.0, neginf=0.0)

    if len(x) == 0:
        return 0.0

    x = x - x.mean()
    best = -np.inf
    n = len(x)

    for w in range(1, min(max_window, n) + 1):
        window_sums = np.convolve(x, np.ones(w), mode="valid")
        score = np.max(window_sums / np.sqrt(w))
        if score > best:
            best = score

    return float(best if np.isfinite(best) else 0.0)

def cusum_positive(values, k=0.25):
    """
    One-sided positive CUSUM on standardized residuals.
    """
    x = np.asarray(values, dtype=float)
    x = np.nan_to_num(x, nan=0.0, posinf=0.0, neginf=0.0)

    s = 0.0
    best = 0.0
    for xi in x:
        s = max(0.0, s + xi - k)
        best = max(best, s)
    return float(best)

def compute_gearys_c(x, W):
    x = np.asarray(x, dtype=float)
    N = len(x)
    var_x = np.sum((x - x.mean()) ** 2)
    if var_x == 0:
        return 1.0

    diff_sq = (x[:, None] - x[None, :]) ** 2
    numerator = (N - 1) * np.sum(W * diff_sq)
    denominator = 2 * W.sum() * var_x
    if denominator == 0:
        return 1.0
    return float(numerator / denominator)

def create_sector_weight_matrix(tickers):
    semis = {"NVDA", "AMD", "INTC", "QCOM", "TXN"}
    n = len(tickers)
    W = np.ones((n, n), dtype=float) * 0.2

    for i in range(n):
        for j in range(n):
            if i == j:
                W[i, j] = 0.0
            else:
                if (tickers[i] in semis) == (tickers[j] in semis):
                    W[i, j] = 0.8
    return W

def empirical_p_value(observed, reference, tail="upper"):
    """
    Empirical p-value with +1 correction.
    tail="upper" -> large observed values are extreme
    tail="lower" -> small observed values are extreme
    """
    reference = np.asarray(reference, dtype=float)
    if len(reference) == 0:
        raise ValueError("Reference array is empty.")

    if tail == "upper":
        count = np.sum(reference >= observed)
    elif tail == "lower":
        count = np.sum(reference <= observed)
    else:
        raise ValueError("tail must be 'upper' or 'lower'.")

    return float((count + 1) / (len(reference) + 1))

# ==========================================
# 3. BASELINES FOR SCAN / CUSUM DETECTORS
# ==========================================

def build_intraday_baselines(full_data, calib_dates, valid_tickers):
    calib_df = full_data[full_data["date"].isin(calib_dates)].copy()
    grp = calib_df.groupby(["stock", "bucket"])["log_volume"]
    baseline = grp.agg(["mean", "std", "count"]).reset_index()

    baseline["std"] = baseline["std"].replace(0, np.nan)
    baseline["std"] = baseline["std"].fillna(baseline["std"].median())
    baseline["std"] = baseline["std"].replace(0, 1.0)

    mean_map = baseline.set_index(["stock", "bucket"])["mean"].to_dict()
    std_map = baseline.set_index(["stock", "bucket"])["std"].to_dict()
    return mean_map, std_map

def get_standardized_series(pivot_lv, stock, mean_map, std_map):
    z_vals = []
    for bucket in pivot_lv.index:
        mu = mean_map.get((stock, bucket), np.nan)
        sd = std_map.get((stock, bucket), np.nan)
        v = pivot_lv.at[bucket, stock] if stock in pivot_lv.columns else np.nan

        if pd.isna(mu) or pd.isna(sd) or sd <= 0 or pd.isna(v):
            z_vals.append(0.0)
        else:
            z_vals.append((v - mu) / sd)
    return np.asarray(z_vals, dtype=float)

# ==========================================
# 4. FEATURE BUILDING
# ==========================================

def build_daily_metrics(full_data, valid_tickers, W_matrix, mean_map, std_map):
    """
    Returns:
      raw_gearys[date] = daily mean Geary's C across buckets
      raw_scans[stock][date] = target-stock localized scan score
      raw_cusums[stock][date] = target-stock positive CUSUM score
    """
    raw_gearys = {}
    raw_scans = {stock: {} for stock in valid_tickers}
    raw_cusums = {stock: {} for stock in valid_tickers}

    for date in sorted(full_data["date"].unique()):
        df_day = full_data[full_data["date"] == date].copy()

        # Geary across buckets (your original branch)
        c_vals = []
        for _, g in df_day.groupby("bucket"):
            x = (
                g.set_index("stock")
                .reindex(valid_tickers)["participation"]
                .fillna(0.0)
                .values
            )
            c_vals.append(compute_gearys_c(x, W_matrix))
        raw_gearys[date] = float(np.mean(c_vals)) if c_vals else np.nan

        # Local detectors on standardized residuals
        pivot_lv = (
            df_day.pivot(index="bucket", columns="stock", values="log_volume")
            .reindex(columns=valid_tickers)
            .sort_index()
        )

        for stock in valid_tickers:
            z_vals = get_standardized_series(pivot_lv, stock, mean_map, std_map)
            raw_scans[stock][date] = max_scan_stat(z_vals, max_window=12)
            raw_cusums[stock][date] = cusum_positive(z_vals, k=0.25)

    return raw_gearys, raw_scans, raw_cusums

# ==========================================
# 5. SCORING
# ==========================================

def score_day(date, ref_dates, raw_gearys, raw_scans, raw_cusums, target_stock):
    g_obs = raw_gearys[date]
    g_ref = [raw_gearys[d] for d in ref_dates if d != date]
    g_pval = empirical_p_value(g_obs, g_ref, tail="lower")

    scan_obs = raw_scans[target_stock][date]
    scan_ref = [raw_scans[target_stock][d] for d in ref_dates if d != date]
    scan_pval = empirical_p_value(scan_obs, scan_ref, tail="upper")

    cusum_obs = raw_cusums[target_stock][date]
    cusum_ref = [raw_cusums[target_stock][d] for d in ref_dates if d != date]
    cusum_pval = empirical_p_value(cusum_obs, cusum_ref, tail="upper")

    return g_obs, g_pval, scan_obs, scan_pval, cusum_obs, cusum_pval

def summarize_false_positives(score_dates, ref_dates, raw_gearys, raw_scans, raw_cusums, target_stock):
    geary_p_values = {}
    scan_p_values = {}
    cusum_p_values = {}

    for date in score_dates:
        g_obs, g_pval, scan_obs, scan_pval, cusum_obs, cusum_pval = score_day(
            date=date,
            ref_dates=ref_dates,
            raw_gearys=raw_gearys,
            raw_scans=raw_scans,
            raw_cusums=raw_cusums,
            target_stock=target_stock,
        )

        geary_p_values[date] = g_pval
        scan_p_values[date] = scan_pval
        cusum_p_values[date] = cusum_pval

        print(
            f"{date}: Geary={g_obs:.4f} p={g_pval:.4f} | "
            f"{target_stock} scan={scan_obs:.4f} p={scan_pval:.4f} | "
            f"CUSUM={cusum_obs:.4f} p={cusum_pval:.4f}"
        )

    print("\n" + "=" * 70)
    print("FALSE-POSITIVE CALIBRATION SUMMARY")
    print("=" * 70)
    print(f"Geary's C:    {100 * sum(p < 0.05 for p in geary_p_values.values()) / len(score_dates):.1f}%")
    print(f"{target_stock} scan:  {100 * sum(p < 0.05 for p in scan_p_values.values()) / len(score_dates):.1f}%")
    print(f"{target_stock} CUSUM:  {100 * sum(p < 0.05 for p in cusum_p_values.values()) / len(score_dates):.1f}%")

    return geary_p_values, scan_p_values, cusum_p_values

# ==========================================
# 6. SYNTHETIC BURST INJECTION
# ==========================================

def inject_synthetic_burst(df_day, target_stock, burst_buckets=12, burst_multiplier=8.0):
    df_mod = df_day.copy()
    unique_buckets = sorted(df_mod["bucket"].unique())
    last_buckets = unique_buckets[-burst_buckets:] if burst_buckets < len(unique_buckets) else unique_buckets

    mask = (df_mod["stock"] == target_stock) & (df_mod["bucket"].isin(last_buckets))
    df_mod.loc[mask, "our_volume"] *= burst_multiplier

    df_mod["mkt_volume"] = df_mod.groupby("bucket")["our_volume"].transform("sum")
    df_mod["participation"] = np.where(
        df_mod["mkt_volume"] > 0,
        df_mod["our_volume"] / df_mod["mkt_volume"],
        0.0,
    )
    df_mod["log_volume"] = np.log1p(df_mod["our_volume"].clip(lower=0))
    return df_mod

def score_burst_day(
    df_burst,
    valid_tickers,
    W_matrix,
    ref_dates,
    raw_gearys,
    raw_scans,
    raw_cusums,
    mean_map,
    std_map,
    target_stock
):
    # Geary branch
    c_vals = []
    for _, g in df_burst.groupby("bucket"):
        x = (
            g.set_index("stock")
            .reindex(valid_tickers)["participation"]
            .fillna(0.0)
            .values
        )
        c_vals.append(compute_gearys_c(x, W_matrix))

    obs_geary_burst = float(np.mean(c_vals)) if c_vals else np.nan
    g_pval = empirical_p_value(obs_geary_burst, [raw_gearys[d] for d in ref_dates], tail="lower")

    # Scan / CUSUM on standardized residuals
    pivot_lv = (
        df_burst.pivot(index="bucket", columns="stock", values="log_volume")
        .reindex(columns=valid_tickers)
        .sort_index()
    )

    z_target = get_standardized_series(pivot_lv, target_stock, mean_map, std_map)

    scan_obs = max_scan_stat(z_target, max_window=12)
    scan_ref = [raw_scans[target_stock][d] for d in ref_dates]
    scan_pval = empirical_p_value(scan_obs, scan_ref, tail="upper")

    cusum_obs = cusum_positive(z_target, k=0.25)
    cusum_ref = [raw_cusums[target_stock][d] for d in ref_dates]
    cusum_pval = empirical_p_value(cusum_obs, cusum_ref, tail="upper")

    return obs_geary_burst, g_pval, scan_obs, scan_pval, cusum_obs, cusum_pval

def run_power_test(
    full_data,
    valid_tickers,
    W_matrix,
    raw_gearys,
    raw_scans,
    raw_cusums,
    ref_dates,
    test_dates,
    mean_map,
    std_map,
    target_stock="NVDA",
    burst_buckets=12,
    burst_multiplier=8.0,
    n_trial_days=15,
    seed=42
):
    rng = np.random.default_rng(seed)
    trial_dates = rng.choice(test_dates, size=min(n_trial_days, len(test_dates)), replace=False)

    flagged_geary = 0
    flagged_scan = 0
    flagged_cusum = 0

    for date in trial_dates:
        df_day = full_data[full_data["date"] == date]
        df_burst = inject_synthetic_burst(df_day, target_stock, burst_buckets, burst_multiplier)

        obs_geary_burst, g_pval, scan_obs, scan_pval, cusum_obs, cusum_pval = score_burst_day(
            df_burst=df_burst,
            valid_tickers=valid_tickers,
            W_matrix=W_matrix,
            ref_dates=ref_dates,
            raw_gearys=raw_gearys,
            raw_scans=raw_scans,
            raw_cusums=raw_cusums,
            mean_map=mean_map,
            std_map=std_map,
            target_stock=target_stock,
        )

        flagged_geary += int(g_pval < 0.05)
        flagged_scan += int(scan_pval < 0.05)
        flagged_cusum += int(cusum_pval < 0.05)

        print(
            f"{date} [{target_stock} x{burst_multiplier:.0f}]: "
            f"Geary p={g_pval:.4f} | "
            f"scan={scan_pval:.4f}{'  <-- FLAGGED' if scan_pval < 0.05 else ''} | "
            f"CUSUM={cusum_pval:.4f}{'  <-- FLAGGED' if cusum_pval < 0.05 else ''}"
        )

    n = len(trial_dates)
    print(f"\nDetection power @ {burst_multiplier:.0f}x burst, {burst_buckets} buckets, target={target_stock}:")
    print(f"  Geary's C caught {flagged_geary}/{n} ({100 * flagged_geary / n:.0f}%)")
    print(f"  Scan detector caught {flagged_scan}/{n} ({100 * flagged_scan / n:.0f}%)")
    print(f"  CUSUM detector caught {flagged_cusum}/{n} ({100 * flagged_cusum / n:.0f}%)")

# ==========================================
# 7. MAIN
# ==========================================

if __name__ == "__main__":
    tickers = [
        "AAPL", "MSFT", "NVDA", "GOOG", "META", "AMD", "TSLA",
        "AMZN", "INTC", "NFLX", "CSCO", "QCOM", "TXN"
    ]

    full_data, valid_tickers = fetch_intraday_market_data(
        tickers, period="60d", interval="5m"
    )

    daily_bucket_counts = full_data.groupby("date")["bucket"].nunique()
    valid_dates = daily_bucket_counts[daily_bucket_counts >= 70].index
    full_data = full_data[full_data["date"].isin(valid_dates)].copy()

    dates = sorted(full_data["date"].unique())
    print(f"Usable real trading days: {len(dates)}\n")

    if len(dates) < 20:
        raise ValueError("Not enough days after filtering to run a meaningful split test.")

    split_idx = int(0.7 * len(dates))
    calib_dates = dates[:split_idx]
    test_dates = dates[split_idx:]

    print(f"Calibration days: {len(calib_dates)}")
    print(f"Test days:        {len(test_dates)}\n")

    W_matrix = create_sector_weight_matrix(valid_tickers)

    # Baselines for the scan/CUSUM branches are learned only from calibration days.
    mean_map, std_map = build_intraday_baselines(full_data, calib_dates, valid_tickers)

    # Geary and local detectors are computed for all days.
    raw_gearys, raw_scans, raw_cusums = build_daily_metrics(
        full_data, valid_tickers, W_matrix, mean_map, std_map
    )

    target_stock = "NVDA"

    print("=" * 70)
    print("LEAVE-ONE-OUT FALSE-POSITIVE CALIBRATION")
    print("=" * 70)
    summarize_false_positives(
        score_dates=calib_dates,
        ref_dates=calib_dates,
        raw_gearys=raw_gearys,
        raw_scans=raw_scans,
        raw_cusums=raw_cusums,
        target_stock=target_stock,
    )

    print("\n" + "=" * 70)
    print("HELD-OUT TEST WINDOW SCORING")
    print("=" * 70)

    heldout_geary_p = {}
    heldout_scan_p = {}
    heldout_cusum_p = {}

    for date in test_dates:
        g_obs, g_pval, scan_obs, scan_pval, cusum_obs, cusum_pval = score_day(
            date=date,
            ref_dates=calib_dates,
            raw_gearys=raw_gearys,
            raw_scans=raw_scans,
            raw_cusums=raw_cusums,
            target_stock=target_stock,
        )

        heldout_geary_p[date] = g_pval
        heldout_scan_p[date] = scan_pval
        heldout_cusum_p[date] = cusum_pval

        print(
            f"{date}: Geary={g_obs:.4f} p={g_pval:.4f} | "
            f"{target_stock} scan={scan_obs:.4f} p={scan_pval:.4f} | "
            f"CUSUM={cusum_obs:.4f} p={cusum_pval:.4f}"
        )

    heldout_geary_flagged = sum(p < 0.05 for p in heldout_geary_p.values())
    heldout_scan_flagged = sum(p < 0.05 for p in heldout_scan_p.values())
    heldout_cusum_flagged = sum(p < 0.05 for p in heldout_cusum_p.values())

    print("\n" + "=" * 70)
    print("HELD-OUT SUMMARY")
    print("=" * 70)
    print(f"Geary's C:    flagged {100 * heldout_geary_flagged / len(test_dates):.1f}% of held-out days at p<0.05")
    print(f"{target_stock} scan:  flagged {100 * heldout_scan_flagged / len(test_dates):.1f}% of held-out days at p<0.05")
    print(f"{target_stock} CUSUM:  flagged {100 * heldout_cusum_flagged / len(test_dates):.1f}% of held-out days at p<0.05")

    print("\n" + "=" * 70)
    print("DETECTION POWER TEST -- injected burst on held-out days")
    print("=" * 70)
    run_power_test(
        full_data=full_data,
        valid_tickers=valid_tickers,
        W_matrix=W_matrix,
        raw_gearys=raw_gearys,
        raw_scans=raw_scans,
        raw_cusums=raw_cusums,
        ref_dates=calib_dates,
        test_dates=test_dates,
        mean_map=mean_map,
        std_map=std_map,
        target_stock=target_stock,
        burst_buckets=12,
        burst_multiplier=8.0,
        n_trial_days=15,
        seed=42,
    )
```

```python
import yfinance as yf
import pandas as pd
import numpy as np
import warnings

warnings.filterwarnings("ignore", category=FutureWarning)

# ==========================================
# 1. YFINANCE DATA ACQUISITION
# ==========================================

def fetch_intraday_market_data(tickers, period="60d", interval="5m"):
    print(f"Downloading real intraday volume data (last {period}, {interval} bars)...")
    raw_data = yf.download(tickers, period=period, interval=interval, progress=False)

    if "Volume" not in raw_data:
        raise ValueError("Failed to fetch volume data.")

    df_vol = raw_data["Volume"].dropna(how="all")
    valid_tickers = sorted(df_vol.columns.tolist())
    print(f"Got data for {len(valid_tickers)}/{len(tickers)} tickers")

    idx_name = df_vol.index.name or "Datetime"
    df_melt = (
        df_vol.reset_index()
        .melt(id_vars=[idx_name], var_name="stock", value_name="our_volume")
        .rename(columns={idx_name: "Datetime"})
    )

    df_melt["Datetime"] = pd.to_datetime(df_melt["Datetime"])
    df_melt["date"] = df_melt["Datetime"].dt.date
    df_melt["bucket"] = df_melt["Datetime"].dt.time

    df_melt["mkt_volume"] = df_melt.groupby("Datetime")["our_volume"].transform("sum")
    df_melt["participation"] = np.where(
        df_melt["mkt_volume"] > 0,
        df_melt["our_volume"] / df_melt["mkt_volume"],
        0.0,
    )

    df_melt["log_volume"] = np.log1p(df_melt["our_volume"].clip(lower=0))
    return df_melt, valid_tickers

# ==========================================
# 2. RAW METRIC CALCULATORS
# ==========================================

def max_rolling_sum(volumes, window=5):
    v = np.asarray(volumes, dtype=float)
    if len(v) < window:
        return float(v.sum())
    return float(np.convolve(v, np.ones(window), mode="valid").max())

def max_scan_stat(values, max_window=12):
    """
    Localized burst score:
    max over window lengths 1..max_window of centered window sums / sqrt(window)
    """
    x = np.asarray(values, dtype=float)
    x = np.nan_to_num(x, nan=0.0, posinf=0.0, neginf=0.0)

    if len(x) == 0:
        return 0.0

    x = x - x.mean()
    best = -np.inf
    n = len(x)

    for w in range(1, min(max_window, n) + 1):
        window_sums = np.convolve(x, np.ones(w), mode="valid")
        score = np.max(window_sums / np.sqrt(w))
        if score > best:
            best = score

    return float(best if np.isfinite(best) else 0.0)

def cusum_positive(values, k=0.25):
    """
    One-sided positive CUSUM on standardized residuals.

    Design note (SPC convention, e.g. Montgomery / NIST handbook): the
    reference value k is typically set to HALF the standardized shift you
    want to detect quickly. k=0.25 here means this detector is tuned to
    pick up a persistent ~0.5-sigma drift in standardized volume, not just
    single-bucket spikes -- that's why it complements the scan stat rather
    than duplicating it.
    """
    x = np.asarray(values, dtype=float)
    x = np.nan_to_num(x, nan=0.0, posinf=0.0, neginf=0.0)

    s = 0.0
    best = 0.0
    for xi in x:
        s = max(0.0, s + xi - k)
        best = max(best, s)
    return float(best)

def compute_gearys_c(x, W):
    x = np.asarray(x, dtype=float)
    N = len(x)
    var_x = np.sum((x - x.mean()) ** 2)
    if var_x == 0:
        return 1.0

    diff_sq = (x[:, None] - x[None, :]) ** 2
    numerator = (N - 1) * np.sum(W * diff_sq)
    denominator = 2 * W.sum() * var_x
    if denominator == 0:
        return 1.0
    return float(numerator / denominator)

def create_sector_weight_matrix(tickers):
    """
    Hand-built binary block weight matrix: 0.8 within the semis cluster,
    0.2 otherwise. Kept for comparison against the data-driven alternative
    below -- see create_correlation_weight_matrix.
    """
    semis = {"NVDA", "AMD", "INTC", "QCOM", "TXN"}
    n = len(tickers)
    W = np.ones((n, n), dtype=float) * 0.2

    for i in range(n):
        for j in range(n):
            if i == j:
                W[i, j] = 0.0
            else:
                if (tickers[i] in semis) == (tickers[j] in semis):
                    W[i, j] = 0.8
    return W

def create_correlation_weight_matrix(full_data, calib_dates, valid_tickers, value_col="participation"):
    """
    Data-driven alternative to the hand-built sector matrix. Built from the
    realized cross-sectional correlation of `value_col` during the
    calibration window only (no leakage into test/held-out days).

    Rationale: spatial-econometrics-in-finance work has repeatedly found
    that weight matrices built from realized co-movement / sector
    similarity outperform geography-style or arbitrary block matrices for
    explaining how stocks actually move together. This gives you a
    citeable, non-arbitrary alternative to defend in the paper instead of
    "we picked 0.8/0.2 because semis felt related."
    """
    calib_df = full_data[full_data["date"].isin(calib_dates)]
    pivot = calib_df.pivot_table(index=["date", "bucket"], columns="stock", values=value_col)
    pivot = pivot.reindex(columns=valid_tickers)
    corr = pivot.corr().values
    corr = np.nan_to_num(corr, nan=0.0)
    W = np.clip(corr, 0.0, None)
    np.fill_diagonal(W, 0.0)
    return W

def empirical_p_value(observed, reference, tail="upper"):
    """
    Empirical p-value with +1 correction (standard Monte-Carlo p-value:
    p = (r + 1) / (R + 1)).

    IMPORTANT LIMITATION: with R reference days, the smallest p-value this
    can EVER report is 1/(R+1). With 42 calibration days that floor is
    ~0.0233. Any observed statistic that beats every single reference day
    gets the same floored p-value regardless of how extreme it actually is
    -- see gumbel_refine_pvalue below for the fix used in the power sweep.
    """
    reference = np.asarray(reference, dtype=float)
    if len(reference) == 0:
        raise ValueError("Reference array is empty.")

    if tail == "upper":
        count = np.sum(reference >= observed)
    elif tail == "lower":
        count = np.sum(reference <= observed)
    else:
        raise ValueError("tail must be 'upper' or 'lower'.")

    return float((count + 1) / (len(reference) + 1))

def gumbel_refine_pvalue(observed, reference, tail="upper"):
    """
    EVT-refined p-value, following Abrams, Kleinman & Kulldorff (2010)
    "Gumbel based p-value approximations for spatial scan statistics."
    They showed that for scan-type statistics (which are themselves maxima
    over many candidate windows), the Gumbel extreme-value distribution
    fits the null distribution far better than normal/gamma/lognormal, and
    lets you estimate p-values well past the resolution floor of a plain
    rank-based Monte-Carlo p-value.

    Method-of-moments Gumbel fit on the reference distribution:
        beta = std(reference) * sqrt(6) / pi
        mu   = mean(reference) - euler_mascheroni * beta
    CDF (max-Gumbel): F(x) = exp(-exp(-(x-mu)/beta))

    For the lower tail we fit a Gumbel to the NEGATED reference (i.e. treat
    small values as the tail of interest, same trick mirrored).

    Use this ALONGSIDE empirical_p_value, not instead of it -- the
    empirical one is exact when not at the floor; this one is the
    extrapolation you need when it is.
    """
    EULER_MASCHERONI = 0.5772156649015329
    reference = np.asarray(reference, dtype=float)
    reference = reference[np.isfinite(reference)]
    if len(reference) < 5:
        return np.nan

    if tail == "upper":
        std_ref = reference.std(ddof=1)
        if std_ref == 0:
            return np.nan
        beta = std_ref * np.sqrt(6.0) / np.pi
        mu = reference.mean() - EULER_MASCHERONI * beta
        z = (observed - mu) / beta
        p = 1.0 - np.exp(-np.exp(-z))
    elif tail == "lower":
        neg_ref = -reference
        std_ref = neg_ref.std(ddof=1)
        if std_ref == 0:
            return np.nan
        beta = std_ref * np.sqrt(6.0) / np.pi
        mu = neg_ref.mean() - EULER_MASCHERONI * beta
        z = (-observed - mu) / beta
        p = 1.0 - np.exp(-np.exp(-z))
    else:
        raise ValueError("tail must be 'upper' or 'lower'.")

    return float(np.clip(p, 1e-12, 1.0))

# ==========================================
# 3. BASELINES FOR SCAN / CUSUM DETECTORS
# ==========================================

def build_intraday_baselines(full_data, calib_dates, valid_tickers):
    calib_df = full_data[full_data["date"].isin(calib_dates)].copy()
    grp = calib_df.groupby(["stock", "bucket"])["log_volume"]
    baseline = grp.agg(["mean", "std", "count"]).reset_index()

    baseline["std"] = baseline["std"].replace(0, np.nan)
    baseline["std"] = baseline["std"].fillna(baseline["std"].median())
    baseline["std"] = baseline["std"].replace(0, 1.0)

    mean_map = baseline.set_index(["stock", "bucket"])["mean"].to_dict()
    std_map = baseline.set_index(["stock", "bucket"])["std"].to_dict()
    return mean_map, std_map

def get_standardized_series(pivot_lv, stock, mean_map, std_map):
    z_vals = []
    for bucket in pivot_lv.index:
        mu = mean_map.get((stock, bucket), np.nan)
        sd = std_map.get((stock, bucket), np.nan)
        v = pivot_lv.at[bucket, stock] if stock in pivot_lv.columns else np.nan

        if pd.isna(mu) or pd.isna(sd) or sd <= 0 or pd.isna(v):
            z_vals.append(0.0)
        else:
            z_vals.append((v - mu) / sd)
    return np.asarray(z_vals, dtype=float)

# ==========================================
# 4. FEATURE BUILDING
# ==========================================

def build_daily_metrics(full_data, valid_tickers, W_matrix, mean_map, std_map):
    """
    Returns:
      raw_gearys[date] = daily mean Geary's C across buckets
      raw_scans[stock][date] = target-stock localized scan score
      raw_cusums[stock][date] = target-stock positive CUSUM score
    """
    raw_gearys = {}
    raw_scans = {stock: {} for stock in valid_tickers}
    raw_cusums = {stock: {} for stock in valid_tickers}

    for date in sorted(full_data["date"].unique()):
        df_day = full_data[full_data["date"] == date].copy()

        c_vals = []
        for _, g in df_day.groupby("bucket"):
            x = (
                g.set_index("stock")
                .reindex(valid_tickers)["participation"]
                .fillna(0.0)
                .values
            )
            c_vals.append(compute_gearys_c(x, W_matrix))
        raw_gearys[date] = float(np.mean(c_vals)) if c_vals else np.nan

        pivot_lv = (
            df_day.pivot(index="bucket", columns="stock", values="log_volume")
            .reindex(columns=valid_tickers)
            .sort_index()
        )

        for stock in valid_tickers:
            z_vals = get_standardized_series(pivot_lv, stock, mean_map, std_map)
            raw_scans[stock][date] = max_scan_stat(z_vals, max_window=12)
            raw_cusums[stock][date] = cusum_positive(z_vals, k=0.25)

    return raw_gearys, raw_scans, raw_cusums

def compute_geary_series(full_data, dates, valid_tickers, W_matrix):
    """
    Lighter-weight helper that recomputes ONLY Geary's C for a given W
    matrix, so we can compare the sector-block matrix against the
    correlation-based matrix without re-running the (expensive) scan/CUSUM
    baseline computation twice.
    """
    out = {}
    for date in dates:
        df_day = full_data[full_data["date"] == date]
        c_vals = []
        for _, g in df_day.groupby("bucket"):
            x = (
                g.set_index("stock")
                .reindex(valid_tickers)["participation"]
                .fillna(0.0)
                .values
            )
            c_vals.append(compute_gearys_c(x, W_matrix))
        out[date] = float(np.mean(c_vals)) if c_vals else np.nan
    return out

# ==========================================
# 5. SCORING
# ==========================================

def score_day(date, ref_dates, raw_gearys, raw_scans, raw_cusums, target_stock):
    g_obs = raw_gearys[date]
    g_ref = [raw_gearys[d] for d in ref_dates if d != date]
    g_pval = empirical_p_value(g_obs, g_ref, tail="lower")

    scan_obs = raw_scans[target_stock][date]
    scan_ref = [raw_scans[target_stock][d] for d in ref_dates if d != date]
    scan_pval = empirical_p_value(scan_obs, scan_ref, tail="upper")

    cusum_obs = raw_cusums[target_stock][date]
    cusum_ref = [raw_cusums[target_stock][d] for d in ref_dates if d != date]
    cusum_pval = empirical_p_value(cusum_obs, cusum_ref, tail="upper")

    return g_obs, g_pval, scan_obs, scan_pval, cusum_obs, cusum_pval

def summarize_false_positives(score_dates, ref_dates, raw_gearys, raw_scans, raw_cusums, target_stock):
    geary_p_values = {}
    scan_p_values = {}
    cusum_p_values = {}

    for date in score_dates:
        g_obs, g_pval, scan_obs, scan_pval, cusum_obs, cusum_pval = score_day(
            date=date,
            ref_dates=ref_dates,
            raw_gearys=raw_gearys,
            raw_scans=raw_scans,
            raw_cusums=raw_cusums,
            target_stock=target_stock,
        )

        geary_p_values[date] = g_pval
        scan_p_values[date] = scan_pval
        cusum_p_values[date] = cusum_pval

        print(
            f"{date}: Geary={g_obs:.4f} p={g_pval:.4f} | "
            f"{target_stock} scan={scan_obs:.4f} p={scan_pval:.4f} | "
            f"CUSUM={cusum_obs:.4f} p={cusum_pval:.4f}"
        )

    print("\n" + "=" * 70)
    print("FALSE-POSITIVE CALIBRATION SUMMARY")
    print("=" * 70)
    print(f"Geary's C:    {100 * sum(p < 0.05 for p in geary_p_values.values()) / len(score_dates):.1f}%")
    print(f"{target_stock} scan:  {100 * sum(p < 0.05 for p in scan_p_values.values()) / len(score_dates):.1f}%")
    print(f"{target_stock} CUSUM:  {100 * sum(p < 0.05 for p in cusum_p_values.values()) / len(score_dates):.1f}%")

    return geary_p_values, scan_p_values, cusum_p_values

# ==========================================
# 6. SYNTHETIC BURST INJECTION
# ==========================================

def inject_synthetic_burst(df_day, target_stock, burst_buckets=12, burst_multiplier=8.0):
    """
    Single-stock idiosyncratic burst. This is the alternative hypothesis
    the scan statistic and CUSUM are built for: ONE name suddenly trading
    far more than its own history at that time of day.
    """
    df_mod = df_day.copy()
    unique_buckets = sorted(df_mod["bucket"].unique())
    last_buckets = unique_buckets[-burst_buckets:] if burst_buckets < len(unique_buckets) else unique_buckets

    mask = (df_mod["stock"] == target_stock) & (df_mod["bucket"].isin(last_buckets))
    df_mod.loc[mask, "our_volume"] *= burst_multiplier

    df_mod["mkt_volume"] = df_mod.groupby("bucket")["our_volume"].transform("sum")
    df_mod["participation"] = np.where(
        df_mod["mkt_volume"] > 0,
        df_mod["our_volume"] / df_mod["mkt_volume"],
        0.0,
    )
    df_mod["log_volume"] = np.log1p(df_mod["our_volume"].clip(lower=0))
    return df_mod

def inject_correlated_burst(df_day, target_stocks, burst_buckets=12, burst_multiplier=3.0):
    """
    Coordinated burst across a BASKET of related names simultaneously --
    e.g. a sector-wide spoofing ring, relative-value/basket manipulation,
    or a marking-the-close scheme spanning several linked instruments.

    This is the alternative Geary's C is actually built to catch: low
    Geary's C signals positive spatial autocorrelation, i.e. neighbouring
    (highly-weighted) assets moving together. A single-stock burst does
    the opposite -- it makes that stock MORE different from its neighbours,
    which pushes Geary's C up, not down. That mismatch is why the original
    single-stock 8x test caught 0/15 on Geary's C: it wasn't testing the
    right alternative for that detector.
    """
    df_mod = df_day.copy()
    unique_buckets = sorted(df_mod["bucket"].unique())
    last_buckets = unique_buckets[-burst_buckets:] if burst_buckets < len(unique_buckets) else unique_buckets

    mask = (df_mod["stock"].isin(target_stocks)) & (df_mod["bucket"].isin(last_buckets))
    df_mod.loc[mask, "our_volume"] *= burst_multiplier

    df_mod["mkt_volume"] = df_mod.groupby("bucket")["our_volume"].transform("sum")
    df_mod["participation"] = np.where(
        df_mod["mkt_volume"] > 0,
        df_mod["our_volume"] / df_mod["mkt_volume"],
        0.0,
    )
    df_mod["log_volume"] = np.log1p(df_mod["our_volume"].clip(lower=0))
    return df_mod

def score_burst_day(
    df_burst,
    valid_tickers,
    W_matrix,
    ref_dates,
    raw_gearys,
    raw_scans,
    raw_cusums,
    mean_map,
    std_map,
    target_stock
):
    c_vals = []
    for _, g in df_burst.groupby("bucket"):
        x = (
            g.set_index("stock")
            .reindex(valid_tickers)["participation"]
            .fillna(0.0)
            .values
        )
        c_vals.append(compute_gearys_c(x, W_matrix))

    obs_geary_burst = float(np.mean(c_vals)) if c_vals else np.nan
    g_pval = empirical_p_value(obs_geary_burst, [raw_gearys[d] for d in ref_dates], tail="lower")

    pivot_lv = (
        df_burst.pivot(index="bucket", columns="stock", values="log_volume")
        .reindex(columns=valid_tickers)
        .sort_index()
    )

    z_target = get_standardized_series(pivot_lv, target_stock, mean_map, std_map)

    scan_obs = max_scan_stat(z_target, max_window=12)
    scan_ref = [raw_scans[target_stock][d] for d in ref_dates]
    scan_pval = empirical_p_value(scan_obs, scan_ref, tail="upper")

    cusum_obs = cusum_positive(z_target, k=0.25)
    cusum_ref = [raw_cusums[target_stock][d] for d in ref_dates]
    cusum_pval = empirical_p_value(cusum_obs, cusum_ref, tail="upper")

    return obs_geary_burst, g_pval, scan_obs, scan_pval, cusum_obs, cusum_pval

# ==========================================
# 7. POWER SWEEPS (replaces the single-multiplier power test)
# ==========================================

def run_power_sweep(
    full_data, valid_tickers, W_matrix,
    raw_gearys, raw_scans, raw_cusums,
    ref_dates, test_dates, mean_map, std_map,
    target_stock="NVDA",
    multipliers=(1.5, 2.0, 3.0, 4.0, 6.0, 8.0),
    burst_buckets=12,
    n_trial_days=15,
    seed=42,
):
    """
    The original test only tried 8x and reported 15/15 with an IDENTICAL
    p-value (the empirical floor 1/(len(calib_dates)+1)) for both scan and
    CUSUM on nearly every trial day. That's a ceiling effect, not a power
    measurement -- 8x is so far outside the reference distribution that
    every trial maxes out the statistic. Sweeping smaller multipliers
    traces where detection power actually drops off, and the Gumbel-
    refined p-value shows the trend continuing even where the raw
    empirical p-value is stuck at its floor.
    """
    rng = np.random.default_rng(seed)
    trial_dates = rng.choice(test_dates, size=min(n_trial_days, len(test_dates)), replace=False)

    results = []
    for mult in multipliers:
        flagged_geary = flagged_scan = flagged_cusum = 0
        scan_gumbel_ps, cusum_gumbel_ps = [], []

        for date in trial_dates:
            df_day = full_data[full_data["date"] == date]
            df_burst = inject_synthetic_burst(df_day, target_stock, burst_buckets, mult)

            obs_geary, g_p, scan_obs, scan_p, cusum_obs, cusum_p = score_burst_day(
                df_burst=df_burst, valid_tickers=valid_tickers, W_matrix=W_matrix,
                ref_dates=ref_dates, raw_gearys=raw_gearys, raw_scans=raw_scans,
                raw_cusums=raw_cusums, mean_map=mean_map, std_map=std_map,
                target_stock=target_stock,
            )

            scan_ref = [raw_scans[target_stock][d] for d in ref_dates]
            cusum_ref = [raw_cusums[target_stock][d] for d in ref_dates]
            scan_gumbel_ps.append(gumbel_refine_pvalue(scan_obs, scan_ref, tail="upper"))
            cusum_gumbel_ps.append(gumbel_refine_pvalue(cusum_obs, cusum_ref, tail="upper"))

            flagged_geary += int(g_p < 0.05)
            flagged_scan += int(scan_p < 0.05)
            flagged_cusum += int(cusum_p < 0.05)

        n = len(trial_dates)
        results.append({
            "multiplier": mult,
            "geary_power": flagged_geary / n,
            "scan_power": flagged_scan / n,
            "cusum_power": flagged_cusum / n,
            "scan_gumbel_p_mean": np.nanmean(scan_gumbel_ps),
            "cusum_gumbel_p_mean": np.nanmean(cusum_gumbel_ps),
        })

    print(f"\nPower curve for {target_stock}, burst_buckets={burst_buckets}, "
          f"{len(trial_dates)} trial days:")
    print(f"{'mult':>6} | {'Geary':>7} | {'Scan':>7} | {'CUSUM':>7} | "
          f"{'Scan~p(EVT)':>12} | {'CUSUM~p(EVT)':>13}")
    for r in results:
        print(
            f"{r['multiplier']:6.1f}x | {100*r['geary_power']:6.1f}% | "
            f"{100*r['scan_power']:6.1f}% | {100*r['cusum_power']:6.1f}% | "
            f"{r['scan_gumbel_p_mean']:12.6f} | {r['cusum_gumbel_p_mean']:13.6f}"
        )
    print("(Scan/CUSUM power saturates at 100% well before 8x -- the EVT-refined")
    print(" p-value columns are the ones that keep moving past that point, and")
    print(" are the more honest number to report for an obviously-extreme burst.)")
    return results

def run_duration_sweep(
    full_data, valid_tickers, W_matrix,
    raw_gearys, raw_scans, raw_cusums,
    ref_dates, test_dates, mean_map, std_map,
    target_stock="NVDA",
    burst_multiplier=4.0,
    durations=(3, 6, 12, 24, 48),
    n_trial_days=15,
    seed=42,
):
    """
    Sweeps how LONG the manipulation episode lasts, at a moderate (non-
    saturating) 4x multiplier. max_scan_stat already scans window lengths
    1..max_window, so it's explicitly designed to be duration-agnostic --
    this checks whether that design goal holds up against CUSUM, which has
    no such built-in flexibility and is expected to degrade on very short
    bursts.
    """
    rng = np.random.default_rng(seed)
    trial_dates = rng.choice(test_dates, size=min(n_trial_days, len(test_dates)), replace=False)

    results = []
    for dur in durations:
        flagged_geary = flagged_scan = flagged_cusum = 0
        for date in trial_dates:
            df_day = full_data[full_data["date"] == date]
            df_burst = inject_synthetic_burst(df_day, target_stock, dur, burst_multiplier)

            obs_geary, g_p, scan_obs, scan_p, cusum_obs, cusum_p = score_burst_day(
                df_burst=df_burst, valid_tickers=valid_tickers, W_matrix=W_matrix,
                ref_dates=ref_dates, raw_gearys=raw_gearys, raw_scans=raw_scans,
                raw_cusums=raw_cusums, mean_map=mean_map, std_map=std_map,
                target_stock=target_stock,
            )
            flagged_geary += int(g_p < 0.05)
            flagged_scan += int(scan_p < 0.05)
            flagged_cusum += int(cusum_p < 0.05)

        n = len(trial_dates)
        results.append({
            "duration": dur,
            "geary_power": flagged_geary / n,
            "scan_power": flagged_scan / n,
            "cusum_power": flagged_cusum / n,
        })

    print(f"\nDuration sensitivity for {target_stock}, burst_multiplier={burst_multiplier}x:")
    print(f"{'buckets':>8} | {'Geary':>7} | {'Scan':>7} | {'CUSUM':>7}")
    for r in results:
        print(
            f"{r['duration']:8d} | {100*r['geary_power']:6.1f}% | "
            f"{100*r['scan_power']:6.1f}% | {100*r['cusum_power']:6.1f}%"
        )
    return results

def run_correlated_burst_test(
    full_data, valid_tickers, W_matrix, raw_gearys, ref_dates, test_dates,
    group_stocks=("NVDA", "AMD", "INTC", "QCOM", "TXN"),
    multipliers=(1.5, 2.0, 3.0, 4.0),
    burst_buckets=12,
    n_trial_days=15,
    seed=42,
):
    """
    Tests Geary's C against the alternative it's actually built for: a
    coordinated move across a basket of related names, not one stock alone.
    """
    rng = np.random.default_rng(seed)
    trial_dates = rng.choice(test_dates, size=min(n_trial_days, len(test_dates)), replace=False)
    group_stocks = [s for s in group_stocks if s in valid_tickers]

    results = []
    for mult in multipliers:
        flagged = 0
        for date in trial_dates:
            df_day = full_data[full_data["date"] == date]
            df_burst = inject_correlated_burst(df_day, group_stocks, burst_buckets, mult)

            c_vals = []
            for _, g in df_burst.groupby("bucket"):
                x = (
                    g.set_index("stock")
                    .reindex(valid_tickers)["participation"]
                    .fillna(0.0)
                    .values
                )
                c_vals.append(compute_gearys_c(x, W_matrix))
            obs_geary = float(np.mean(c_vals)) if c_vals else np.nan
            g_p = empirical_p_value(obs_geary, [raw_gearys[d] for d in ref_dates], tail="lower")
            flagged += int(g_p < 0.05)

        n = len(trial_dates)
        results.append({"multiplier": mult, "geary_power": flagged / n})

    print(f"\nGeary's C power against a COORDINATED burst across {group_stocks},")
    print(f"burst_buckets={burst_buckets}, {len(trial_dates)} trial days:")
    print(f"{'mult':>6} | {'Geary power':>12}")
    for r in results:
        print(f"{r['multiplier']:6.1f}x | {100*r['geary_power']:11.1f}%")
    return results

# ==========================================
# 8. MULTI-STOCK / MULTIPLE-TESTING CORRECTION
# ==========================================

def benjamini_hochberg(pvalues, alpha=0.05):
    """
    Standard BH (1995) step-up FDR procedure. Watching N stocks at once and
    flagging each at raw p<0.05 produces ~0.05*N expected false alarms per
    day from chance alone (with 13 tickers, ~0.65/day). BH controls the
    expected proportion of false discoveries among whatever actually gets
    flagged, which is the right correction once you stop testing one
    pre-chosen target stock and start scanning a whole basket.
    """
    pvalues = np.asarray(pvalues, dtype=float)
    n = len(pvalues)
    order = np.argsort(pvalues)
    ranked = pvalues[order]
    thresholds = (np.arange(1, n + 1) / n) * alpha
    below = ranked <= thresholds
    reject = np.zeros(n, dtype=bool)
    if below.any():
        max_idx = np.max(np.where(below)[0])
        reject[order[:max_idx + 1]] = True
    return reject

def run_multistock_heldout_scoring(test_dates, calib_dates, raw_scans, valid_tickers, alpha=0.05):
    """
    Repeats the held-out scan-detector scoring for EVERY stock, not just
    NVDA, then applies BH-FDR correction across stocks within each day.
    NVDA is a fine single target to debug a pipeline on, but a real
    surveillance system watches the whole basket simultaneously, and that
    changes the multiple-testing math.
    """
    raw_flag_count = 0
    fdr_flag_count = 0
    total_tests = 0

    for date in test_dates:
        day_pvals = {}
        for stock in valid_tickers:
            scan_obs = raw_scans[stock][date]
            scan_ref = [raw_scans[stock][d] for d in calib_dates]
            day_pvals[stock] = empirical_p_value(scan_obs, scan_ref, tail="upper")

        pvals_array = np.array(list(day_pvals.values()))
        raw_flag_count += int((pvals_array < alpha).sum())

        rejected = benjamini_hochberg(pvals_array, alpha=alpha)
        fdr_flag_count += int(rejected.sum())
        total_tests += len(pvals_array)

    print(f"\nMulti-stock scan detector, held-out window, "
          f"{len(valid_tickers)} tickers x {len(test_dates)} days:")
    print(f"  Raw p<0.05 flags:      {raw_flag_count}/{total_tests} "
          f"({100*raw_flag_count/total_tests:.1f}%)")
    print(f"  BH-FDR-adjusted flags: {fdr_flag_count}/{total_tests} "
          f"({100*fdr_flag_count/total_tests:.1f}%)")

# ==========================================
# 9. MAIN
# ==========================================

if __name__ == "__main__":
    tickers = [
        "AAPL", "MSFT", "NVDA", "GOOG", "META", "AMD", "TSLA",
        "AMZN", "INTC", "NFLX", "CSCO", "QCOM", "TXN"
    ]

    full_data, valid_tickers = fetch_intraday_market_data(
        tickers, period="60d", interval="5m"
    )

    daily_bucket_counts = full_data.groupby("date")["bucket"].nunique()
    valid_dates = daily_bucket_counts[daily_bucket_counts >= 70].index
    full_data = full_data[full_data["date"].isin(valid_dates)].copy()

    dates = sorted(full_data["date"].unique())
    print(f"Usable real trading days: {len(dates)}\n")

    if len(dates) < 20:
        raise ValueError("Not enough days after filtering to run a meaningful split test.")

    split_idx = int(0.7 * len(dates))
    calib_dates = dates[:split_idx]
    test_dates = dates[split_idx:]

    print(f"Calibration days: {len(calib_dates)}")
    print(f"Test days:        {len(test_dates)}\n")

    W_sector = create_sector_weight_matrix(valid_tickers)
    W_corr = create_correlation_weight_matrix(full_data, calib_dates, valid_tickers)

    mean_map, std_map = build_intraday_baselines(full_data, calib_dates, valid_tickers)

    # Main pipeline keeps the original sector-block W for continuity.
    raw_gearys, raw_scans, raw_cusums = build_daily_metrics(
        full_data, valid_tickers, W_sector, mean_map, std_map
    )

    target_stock = "NVDA"

    print("=" * 70)
    print("LEAVE-ONE-OUT FALSE-POSITIVE CALIBRATION")
    print("=" * 70)
    summarize_false_positives(
        score_dates=calib_dates, ref_dates=calib_dates,
        raw_gearys=raw_gearys, raw_scans=raw_scans, raw_cusums=raw_cusums,
        target_stock=target_stock,
    )

    print("\n" + "=" * 70)
    print("HELD-OUT TEST WINDOW SCORING")
    print("=" * 70)
    heldout_geary_p, heldout_scan_p, heldout_cusum_p = {}, {}, {}
    for date in test_dates:
        g_obs, g_pval, scan_obs, scan_pval, cusum_obs, cusum_pval = score_day(
            date=date, ref_dates=calib_dates, raw_gearys=raw_gearys,
            raw_scans=raw_scans, raw_cusums=raw_cusums, target_stock=target_stock,
        )
        heldout_geary_p[date] = g_pval
        heldout_scan_p[date] = scan_pval
        heldout_cusum_p[date] = cusum_pval
        print(
            f"{date}: Geary={g_obs:.4f} p={g_pval:.4f} | "
            f"{target_stock} scan={scan_obs:.4f} p={scan_pval:.4f} | "
            f"CUSUM={cusum_obs:.4f} p={cusum_pval:.4f}"
        )

    n_test = len(test_dates)
    print("\n" + "=" * 70)
    print("HELD-OUT SUMMARY")
    print("=" * 70)
    print(f"Geary's C:    flagged {100*sum(p<0.05 for p in heldout_geary_p.values())/n_test:.1f}% of held-out days")
    print(f"{target_stock} scan:  flagged {100*sum(p<0.05 for p in heldout_scan_p.values())/n_test:.1f}% of held-out days")
    print(f"{target_stock} CUSUM:  flagged {100*sum(p<0.05 for p in heldout_cusum_p.values())/n_test:.1f}% of held-out days")

    # --- NEW: weight matrix diagnostic ---
    print("\n" + "=" * 70)
    print("WEIGHT MATRIX DIAGNOSTIC: hand-built sector blocks vs realized correlation")
    print("=" * 70)
    geary_sector_calib = compute_geary_series(full_data, calib_dates, valid_tickers, W_sector)
    geary_corr_calib = compute_geary_series(full_data, calib_dates, valid_tickers, W_corr)

    fp_sector = sum(
        empirical_p_value(geary_sector_calib[d], [geary_sector_calib[x] for x in calib_dates if x != d], tail="lower") < 0.05
        for d in calib_dates
    ) / len(calib_dates)
    fp_corr = sum(
        empirical_p_value(geary_corr_calib[d], [geary_corr_calib[x] for x in calib_dates if x != d], tail="lower") < 0.05
        for d in calib_dates
    ) / len(calib_dates)
    print(f"  Sector-block W:      {100*fp_sector:.1f}% LOO false-positive rate")
    print(f"  Correlation-based W: {100*fp_corr:.1f}% LOO false-positive rate")
    print("  (Both should sit near 5% if well-calibrated -- large divergence")
    print("   tells you the hand-built blocks are mis-specified.)")

    # --- NEW: power sweep (replaces the single 8x test) ---
    print("\n" + "=" * 70)
    print("DETECTION POWER SWEEP -- burst STRENGTH, held-out days")
    print("=" * 70)
    run_power_sweep(
        full_data=full_data, valid_tickers=valid_tickers, W_matrix=W_sector,
        raw_gearys=raw_gearys, raw_scans=raw_scans, raw_cusums=raw_cusums,
        ref_dates=calib_dates, test_dates=test_dates, mean_map=mean_map, std_map=std_map,
        target_stock=target_stock, multipliers=(1.5, 2.0, 3.0, 4.0, 6.0, 8.0),
        burst_buckets=12, n_trial_days=15, seed=42,
    )

    # --- NEW: duration sweep ---
    print("\n" + "=" * 70)
    print("DETECTION POWER SWEEP -- burst DURATION, held-out days")
    print("=" * 70)
    run_duration_sweep(
        full_data=full_data, valid_tickers=valid_tickers, W_matrix=W_sector,
        raw_gearys=raw_gearys, raw_scans=raw_scans, raw_cusums=raw_cusums,
        ref_dates=calib_dates, test_dates=test_dates, mean_map=mean_map, std_map=std_map,
        target_stock=target_stock, burst_multiplier=4.0,
        durations=(3, 6, 12, 24, 48), n_trial_days=15, seed=42,
    )

    # --- NEW: coordinated burst test for Geary's C ---
    print("\n" + "=" * 70)
    print("DETECTION POWER -- COORDINATED basket burst (Geary's C alternative)")
    print("=" * 70)
    run_correlated_burst_test(
        full_data=full_data, valid_tickers=valid_tickers, W_matrix=W_sector,
        raw_gearys=raw_gearys, ref_dates=calib_dates, test_dates=test_dates,
        group_stocks=("NVDA", "AMD", "INTC", "QCOM", "TXN"),
        multipliers=(1.5, 2.0, 3.0, 4.0), burst_buckets=12, n_trial_days=15, seed=42,
    )

    # --- NEW: multi-stock FDR-adjusted held-out scoring ---
    print("\n" + "=" * 70)
    print("MULTI-STOCK SCORING WITH FDR CORRECTION")
    print("=" * 70)
    run_multistock_heldout_scoring(
        test_dates=test_dates, calib_dates=calib_dates,
        raw_scans=raw_scans, valid_tickers=valid_tickers, alpha=0.05,
    )
```

```python
import yfinance as yf
import pandas as pd
import numpy as np
import warnings

warnings.filterwarnings("ignore", category=FutureWarning)

# ==========================================
# 1. YFINANCE DATA ACQUISITION
# ==========================================

def fetch_intraday_market_data(tickers, period="60d", interval="5m"):
    print(f"Downloading real intraday volume data (last {period}, {interval} bars)...")
    raw_data = yf.download(tickers, period=period, interval=interval, progress=False)

    if "Volume" not in raw_data:
        raise ValueError("Failed to fetch volume data.")

    df_vol = raw_data["Volume"].dropna(how="all")
    valid_tickers = sorted(df_vol.columns.tolist())
    print(f"Got data for {len(valid_tickers)}/{len(tickers)} tickers")

    idx_name = df_vol.index.name or "Datetime"
    df_melt = (
        df_vol.reset_index()
        .melt(id_vars=[idx_name], var_name="stock", value_name="our_volume")
        .rename(columns={idx_name: "Datetime"})
    )

    df_melt["Datetime"] = pd.to_datetime(df_melt["Datetime"])
    df_melt["date"] = df_melt["Datetime"].dt.date
    df_melt["bucket"] = df_melt["Datetime"].dt.time

    df_melt["mkt_volume"] = df_melt.groupby("Datetime")["our_volume"].transform("sum")
    df_melt["participation"] = np.where(
        df_melt["mkt_volume"] > 0,
        df_melt["our_volume"] / df_melt["mkt_volume"],
        0.0,
    )

    df_melt["log_volume"] = np.log1p(df_melt["our_volume"].clip(lower=0))
    return df_melt, valid_tickers

# ==========================================
# 2. RAW METRIC CALCULATORS
# ==========================================

def max_rolling_sum(volumes, window=5):
    v = np.asarray(volumes, dtype=float)
    if len(v) < window:
        return float(v.sum())
    return float(np.convolve(v, np.ones(window), mode="valid").max())

def max_scan_stat(values, max_window=12):
    """
    Localized burst score:
    max over window lengths 1..max_window of centered window sums / sqrt(window)

    Note: for a fixed per-bucket excess signal, this score scales roughly
    with sqrt(window length) -- so a brief, intense burst that only lasts
    a few buckets is intrinsically harder for THIS detector to catch than
    a longer one at the same intensity. CUSUM (below) doesn't have that
    same penalty, which is why the two detectors fail at different
    durations rather than redundantly.
    """
    x = np.asarray(values, dtype=float)
    x = np.nan_to_num(x, nan=0.0, posinf=0.0, neginf=0.0)

    if len(x) == 0:
        return 0.0

    x = x - x.mean()
    best = -np.inf
    n = len(x)

    for w in range(1, min(max_window, n) + 1):
        window_sums = np.convolve(x, np.ones(w), mode="valid")
        score = np.max(window_sums / np.sqrt(w))
        if score > best:
            best = score

    return float(best if np.isfinite(best) else 0.0)

def cusum_positive(values, k=0.25):
    """
    One-sided positive CUSUM on standardized residuals.

    Design note (SPC convention, e.g. Montgomery / NIST handbook): k is
    typically set to HALF the standardized shift you want to detect
    quickly. k=0.25 here tunes this detector for a persistent ~0.5-sigma
    drift, which is why it complements the scan stat instead of
    duplicating it.
    """
    x = np.asarray(values, dtype=float)
    x = np.nan_to_num(x, nan=0.0, posinf=0.0, neginf=0.0)

    s = 0.0
    best = 0.0
    for xi in x:
        s = max(0.0, s + xi - k)
        best = max(best, s)
    return float(best)

def compute_gearys_c(x, W):
    x = np.asarray(x, dtype=float)
    N = len(x)
    var_x = np.sum((x - x.mean()) ** 2)
    if var_x == 0:
        return 1.0

    diff_sq = (x[:, None] - x[None, :]) ** 2
    numerator = (N - 1) * np.sum(W * diff_sq)
    denominator = 2 * W.sum() * var_x
    if denominator == 0:
        return 1.0
    return float(numerator / denominator)

def create_sector_weight_matrix(tickers):
    """
    Hand-built binary block weight matrix: 0.8 within the semis cluster,
    0.2 otherwise. Kept for comparison against the data-driven alternative
    below.
    """
    semis = {"NVDA", "AMD", "INTC", "QCOM", "TXN"}
    n = len(tickers)
    W = np.ones((n, n), dtype=float) * 0.2

    for i in range(n):
        for j in range(n):
            if i == j:
                W[i, j] = 0.0
            else:
                if (tickers[i] in semis) == (tickers[j] in semis):
                    W[i, j] = 0.8
    return W

def create_correlation_weight_matrix(full_data, calib_dates, valid_tickers, value_col="log_volume"):
    """
    Data-driven alternative to the hand-built sector matrix, built from
    realized correlation during the calibration window only (no leakage).

    Uses log_volume rather than participation deliberately: participation
    values across stocks sum to 1 at every timestamp (compositional data),
    which induces a built-in negative-correlation bias between unrelated
    stocks purely from that sum-to-one constraint, not genuine economic
    relationship. log_volume isn't constrained that way.
    """
    calib_df = full_data[full_data["date"].isin(calib_dates)]
    pivot = calib_df.pivot_table(index=["date", "bucket"], columns="stock", values=value_col)
    pivot = pivot.reindex(columns=valid_tickers)
    corr = pivot.corr().values
    corr = np.nan_to_num(corr, nan=0.0)
    W = np.clip(corr, 0.0, None)
    np.fill_diagonal(W, 0.0)
    return W

def empirical_p_value(observed, reference, tail="upper"):
    """
    Empirical p-value with +1 correction: p = (r + 1) / (R + 1).

    LIMITATION: with R reference days, the smallest p-value this can EVER
    report is 1/(R+1). Any observed statistic beating every reference day
    gets the same floored p-value no matter how extreme it actually is.
    See gumbel_refine_pvalue below.
    """
    reference = np.asarray(reference, dtype=float)
    if len(reference) == 0:
        raise ValueError("Reference array is empty.")

    if tail == "upper":
        count = np.sum(reference >= observed)
    elif tail == "lower":
        count = np.sum(reference <= observed)
    else:
        raise ValueError("tail must be 'upper' or 'lower'.")

    return float((count + 1) / (len(reference) + 1))

def gumbel_refine_pvalue(observed, reference, tail="upper"):
    """
    EVT-refined p-value, following Abrams, Kleinman & Kulldorff (2010)
    "Gumbel based p-value approximations for spatial scan statistics."
    Scan-type statistics are maxima over many candidate windows, and the
    Gumbel extreme-value distribution fits that null distribution far
    better than normal/gamma/lognormal -- letting you estimate p-values
    well past the floor of a plain rank-based Monte-Carlo p-value.

    This is no longer just a "nicer power curve" tool -- it's required to
    make FDR correction across multiple stocks meaningful at all when the
    calibration window is short (see run_multistock_heldout_scoring).

    Method-of-moments Gumbel fit on the reference distribution:
        beta = std(reference) * sqrt(6) / pi
        mu   = mean(reference) - euler_mascheroni * beta
    CDF (max-Gumbel): F(x) = exp(-exp(-(x-mu)/beta))
    """
    EULER_MASCHERONI = 0.5772156649015329
    reference = np.asarray(reference, dtype=float)
    reference = reference[np.isfinite(reference)]
    if len(reference) < 5:
        return np.nan

    if tail == "upper":
        std_ref = reference.std(ddof=1)
        if std_ref == 0:
            return np.nan
        beta = std_ref * np.sqrt(6.0) / np.pi
        mu = reference.mean() - EULER_MASCHERONI * beta
        z = (observed - mu) / beta
        p = 1.0 - np.exp(-np.exp(-z))
    elif tail == "lower":
        neg_ref = -reference
        std_ref = neg_ref.std(ddof=1)
        if std_ref == 0:
            return np.nan
        beta = std_ref * np.sqrt(6.0) / np.pi
        mu = neg_ref.mean() - EULER_MASCHERONI * beta
        z = (-observed - mu) / beta
        p = 1.0 - np.exp(-np.exp(-z))
    else:
        raise ValueError("tail must be 'upper' or 'lower'.")

    return float(np.clip(p, 1e-12, 1.0))

# ==========================================
# 3. BASELINES FOR SCAN / CUSUM / GEARY DETECTORS
# ==========================================

def build_intraday_baselines(full_data, calib_dates, valid_tickers):
    calib_df = full_data[full_data["date"].isin(calib_dates)].copy()
    grp = calib_df.groupby(["stock", "bucket"])["log_volume"]
    baseline = grp.agg(["mean", "std", "count"]).reset_index()

    baseline["std"] = baseline["std"].replace(0, np.nan)
    baseline["std"] = baseline["std"].fillna(baseline["std"].median())
    baseline["std"] = baseline["std"].replace(0, 1.0)

    mean_map = baseline.set_index(["stock", "bucket"])["mean"].to_dict()
    std_map = baseline.set_index(["stock", "bucket"])["std"].to_dict()
    return mean_map, std_map

def get_standardized_series(pivot_lv, stock, mean_map, std_map):
    z_vals = []
    for bucket in pivot_lv.index:
        mu = mean_map.get((stock, bucket), np.nan)
        sd = std_map.get((stock, bucket), np.nan)
        v = pivot_lv.at[bucket, stock] if stock in pivot_lv.columns else np.nan

        if pd.isna(mu) or pd.isna(sd) or sd <= 0 or pd.isna(v):
            z_vals.append(0.0)
        else:
            z_vals.append((v - mu) / sd)
    return np.asarray(z_vals, dtype=float)

def geary_for_day(df_day, valid_tickers, W_matrix, mean_map, std_map):
    """
    Mean Geary's C for one day, computed on STANDARDIZED (z-scored) volume
    -- NOT raw participation share.

    Why this matters: Geary's C pools squared differences (x_i - x_j)
    across ALL units on one shared scale. NVDA naturally trades far more
    volume than TXN, so raw participation differences are dominated by
    that baseline gap, not by whether the two are moving together. This is
    exactly the heteroscedasticity problem spatial epidemiology hit
    running Moran's I / Geary's C on raw disease counts across regions of
    different population size -- the standard fix there is to standardize
    (rates/SMRs) before computing the autocorrelation statistic. Same fix
    here: standardize each stock against its OWN per-time-of-day baseline
    (the same mean_map/std_map the scan and CUSUM detectors already use)
    before computing Geary's C, so "moving together" means "deviating from
    normal together," not "having a similar absolute share of volume."
    """
    pivot_lv = (
        df_day.pivot(index="bucket", columns="stock", values="log_volume")
        .reindex(columns=valid_tickers)
        .sort_index()
    )
    z_matrix = np.column_stack([
        get_standardized_series(pivot_lv, stock, mean_map, std_map)
        for stock in valid_tickers
    ])
    c_vals = [compute_gearys_c(row, W_matrix) for row in z_matrix]
    return float(np.mean(c_vals)) if c_vals else np.nan

# ==========================================
# 4. FEATURE BUILDING
# ==========================================

def build_daily_metrics(full_data, valid_tickers, W_matrix, mean_map, std_map):
    """
    Returns:
      raw_gearys[date] = daily mean Geary's C (on standardized volume)
      raw_scans[stock][date] = target-stock localized scan score
      raw_cusums[stock][date] = target-stock positive CUSUM score
    """
    raw_gearys = {}
    raw_scans = {stock: {} for stock in valid_tickers}
    raw_cusums = {stock: {} for stock in valid_tickers}

    for date in sorted(full_data["date"].unique()):
        df_day = full_data[full_data["date"] == date].copy()

        raw_gearys[date] = geary_for_day(df_day, valid_tickers, W_matrix, mean_map, std_map)

        pivot_lv = (
            df_day.pivot(index="bucket", columns="stock", values="log_volume")
            .reindex(columns=valid_tickers)
            .sort_index()
        )
        for stock in valid_tickers:
            z_vals = get_standardized_series(pivot_lv, stock, mean_map, std_map)
            raw_scans[stock][date] = max_scan_stat(z_vals, max_window=12)
            raw_cusums[stock][date] = cusum_positive(z_vals, k=0.25)

    return raw_gearys, raw_scans, raw_cusums

# ==========================================
# 5. SCORING
# ==========================================

def score_day(date, ref_dates, raw_gearys, raw_scans, raw_cusums, target_stock):
    g_obs = raw_gearys[date]
    g_ref = [raw_gearys[d] for d in ref_dates if d != date]
    g_pval = empirical_p_value(g_obs, g_ref, tail="lower")

    scan_obs = raw_scans[target_stock][date]
    scan_ref = [raw_scans[target_stock][d] for d in ref_dates if d != date]
    scan_pval = empirical_p_value(scan_obs, scan_ref, tail="upper")

    cusum_obs = raw_cusums[target_stock][date]
    cusum_ref = [raw_cusums[target_stock][d] for d in ref_dates if d != date]
    cusum_pval = empirical_p_value(cusum_obs, cusum_ref, tail="upper")

    return g_obs, g_pval, scan_obs, scan_pval, cusum_obs, cusum_pval

def summarize_false_positives(score_dates, ref_dates, raw_gearys, raw_scans, raw_cusums, target_stock):
    geary_p_values, scan_p_values, cusum_p_values = {}, {}, {}

    for date in score_dates:
        g_obs, g_pval, scan_obs, scan_pval, cusum_obs, cusum_pval = score_day(
            date=date, ref_dates=ref_dates, raw_gearys=raw_gearys,
            raw_scans=raw_scans, raw_cusums=raw_cusums, target_stock=target_stock,
        )
        geary_p_values[date] = g_pval
        scan_p_values[date] = scan_pval
        cusum_p_values[date] = cusum_pval

        print(
            f"{date}: Geary={g_obs:.4f} p={g_pval:.4f} | "
            f"{target_stock} scan={scan_obs:.4f} p={scan_pval:.4f} | "
            f"CUSUM={cusum_obs:.4f} p={cusum_pval:.4f}"
        )

    print("\n" + "=" * 70)
    print("FALSE-POSITIVE CALIBRATION SUMMARY")
    print("=" * 70)
    print(f"Geary's C:    {100 * sum(p < 0.05 for p in geary_p_values.values()) / len(score_dates):.1f}%")
    print(f"{target_stock} scan:  {100 * sum(p < 0.05 for p in scan_p_values.values()) / len(score_dates):.1f}%")
    print(f"{target_stock} CUSUM:  {100 * sum(p < 0.05 for p in cusum_p_values.values()) / len(score_dates):.1f}%")

    return geary_p_values, scan_p_values, cusum_p_values

# ==========================================
# 6. SYNTHETIC BURST INJECTION
# ==========================================

def inject_synthetic_burst(df_day, target_stock, burst_buckets=12, burst_multiplier=8.0):
    """ Single-stock idiosyncratic burst. """
    df_mod = df_day.copy()
    unique_buckets = sorted(df_mod["bucket"].unique())
    last_buckets = unique_buckets[-burst_buckets:] if burst_buckets < len(unique_buckets) else unique_buckets

    mask = (df_mod["stock"] == target_stock) & (df_mod["bucket"].isin(last_buckets))
    df_mod.loc[mask, "our_volume"] *= burst_multiplier

    df_mod["mkt_volume"] = df_mod.groupby("bucket")["our_volume"].transform("sum")
    df_mod["participation"] = np.where(
        df_mod["mkt_volume"] > 0,
        df_mod["our_volume"] / df_mod["mkt_volume"],
        0.0,
    )
    df_mod["log_volume"] = np.log1p(df_mod["our_volume"].clip(lower=0))
    return df_mod

def inject_correlated_burst(df_day, target_stocks, burst_buckets=12, burst_multiplier=3.0):
    """
    Coordinated burst across a BASKET of related names simultaneously --
    e.g. a sector-wide spoofing ring or basket/relative-value manipulation.
    Combined with geary_for_day's standardization fix, all members
    jumping by a similar number of STANDARD DEVIATIONS (not the same raw
    multiplier on different baseline sizes) is what should now register
    as clustering / low Geary's C.
    """
    df_mod = df_day.copy()
    unique_buckets = sorted(df_mod["bucket"].unique())
    last_buckets = unique_buckets[-burst_buckets:] if burst_buckets < len(unique_buckets) else unique_buckets

    mask = (df_mod["stock"].isin(target_stocks)) & (df_mod["bucket"].isin(last_buckets))
    df_mod.loc[mask, "our_volume"] *= burst_multiplier

    df_mod["mkt_volume"] = df_mod.groupby("bucket")["our_volume"].transform("sum")
    df_mod["participation"] = np.where(
        df_mod["mkt_volume"] > 0,
        df_mod["our_volume"] / df_mod["mkt_volume"],
        0.0,
    )
    df_mod["log_volume"] = np.log1p(df_mod["our_volume"].clip(lower=0))
    return df_mod

def score_burst_day(
    df_burst, valid_tickers, W_matrix, ref_dates, raw_gearys, raw_scans,
    raw_cusums, mean_map, std_map, target_stock
):
    obs_geary_burst = geary_for_day(df_burst, valid_tickers, W_matrix, mean_map, std_map)
    g_pval = empirical_p_value(obs_geary_burst, [raw_gearys[d] for d in ref_dates], tail="lower")

    pivot_lv = (
        df_burst.pivot(index="bucket", columns="stock", values="log_volume")
        .reindex(columns=valid_tickers)
        .sort_index()
    )
    z_target = get_standardized_series(pivot_lv, target_stock, mean_map, std_map)

    scan_obs = max_scan_stat(z_target, max_window=12)
    scan_ref = [raw_scans[target_stock][d] for d in ref_dates]
    scan_pval = empirical_p_value(scan_obs, scan_ref, tail="upper")

    cusum_obs = cusum_positive(z_target, k=0.25)
    cusum_ref = [raw_cusums[target_stock][d] for d in ref_dates]
    cusum_pval = empirical_p_value(cusum_obs, cusum_ref, tail="upper")

    return obs_geary_burst, g_pval, scan_obs, scan_pval, cusum_obs, cusum_pval

# ==========================================
# 7. POWER SWEEPS
# ==========================================

def run_power_sweep(
    full_data, valid_tickers, W_matrix, raw_gearys, raw_scans, raw_cusums,
    ref_dates, test_dates, mean_map, std_map, target_stock="NVDA",
    multipliers=(1.5, 2.0, 3.0, 4.0, 6.0, 8.0), burst_buckets=12,
    n_trial_days=15, seed=42,
):
    rng = np.random.default_rng(seed)
    trial_dates = rng.choice(test_dates, size=min(n_trial_days, len(test_dates)), replace=False)

    results = []
    for mult in multipliers:
        flagged_geary = flagged_scan = flagged_cusum = 0
        scan_gumbel_ps, cusum_gumbel_ps = [], []

        for date in trial_dates:
            df_day = full_data[full_data["date"] == date]
            df_burst = inject_synthetic_burst(df_day, target_stock, burst_buckets, mult)

            obs_geary, g_p, scan_obs, scan_p, cusum_obs, cusum_p = score_burst_day(
                df_burst=df_burst, valid_tickers=valid_tickers, W_matrix=W_matrix,
                ref_dates=ref_dates, raw_gearys=raw_gearys, raw_scans=raw_scans,
                raw_cusums=raw_cusums, mean_map=mean_map, std_map=std_map,
                target_stock=target_stock,
            )

            scan_ref = [raw_scans[target_stock][d] for d in ref_dates]
            cusum_ref = [raw_cusums[target_stock][d] for d in ref_dates]
            scan_gumbel_ps.append(gumbel_refine_pvalue(scan_obs, scan_ref, tail="upper"))
            cusum_gumbel_ps.append(gumbel_refine_pvalue(cusum_obs, cusum_ref, tail="upper"))

            flagged_geary += int(g_p < 0.05)
            flagged_scan += int(scan_p < 0.05)
            flagged_cusum += int(cusum_p < 0.05)

        n = len(trial_dates)
        results.append({
            "multiplier": mult,
            "geary_power": flagged_geary / n,
            "scan_power": flagged_scan / n,
            "cusum_power": flagged_cusum / n,
            "scan_gumbel_p_mean": np.nanmean(scan_gumbel_ps),
            "cusum_gumbel_p_mean": np.nanmean(cusum_gumbel_ps),
        })

    print(f"\nPower curve for {target_stock}, burst_buckets={burst_buckets}, "
          f"{len(trial_dates)} trial days:")
    print(f"{'mult':>6} | {'Geary':>7} | {'Scan':>7} | {'CUSUM':>7} | "
          f"{'Scan~p(EVT)':>12} | {'CUSUM~p(EVT)':>13}")
    for r in results:
        print(
            f"{r['multiplier']:6.1f}x | {100*r['geary_power']:6.1f}% | "
            f"{100*r['scan_power']:6.1f}% | {100*r['cusum_power']:6.1f}% | "
            f"{r['scan_gumbel_p_mean']:12.6f} | {r['cusum_gumbel_p_mean']:13.6f}"
        )
    return results

def run_duration_sweep(
    full_data, valid_tickers, W_matrix, raw_gearys, raw_scans, raw_cusums,
    ref_dates, test_dates, mean_map, std_map, target_stock="NVDA",
    burst_multiplier=4.0, durations=(3, 6, 12, 24, 48),
    n_trial_days=15, seed=42,
):
    rng = np.random.default_rng(seed)
    trial_dates = rng.choice(test_dates, size=min(n_trial_days, len(test_dates)), replace=False)

    results = []
    for dur in durations:
        flagged_geary = flagged_scan = flagged_cusum = 0
        for date in trial_dates:
            df_day = full_data[full_data["date"] == date]
            df_burst = inject_synthetic_burst(df_day, target_stock, dur, burst_multiplier)

            obs_geary, g_p, scan_obs, scan_p, cusum_obs, cusum_p = score_burst_day(
                df_burst=df_burst, valid_tickers=valid_tickers, W_matrix=W_matrix,
                ref_dates=ref_dates, raw_gearys=raw_gearys, raw_scans=raw_scans,
                raw_cusums=raw_cusums, mean_map=mean_map, std_map=std_map,
                target_stock=target_stock,
            )
            flagged_geary += int(g_p < 0.05)
            flagged_scan += int(scan_p < 0.05)
            flagged_cusum += int(cusum_p < 0.05)

        n = len(trial_dates)
        results.append({
            "duration": dur,
            "geary_power": flagged_geary / n,
            "scan_power": flagged_scan / n,
            "cusum_power": flagged_cusum / n,
        })

    print(f"\nDuration sensitivity for {target_stock}, burst_multiplier={burst_multiplier}x:")
    print(f"{'buckets':>8} | {'Geary':>7} | {'Scan':>7} | {'CUSUM':>7}")
    for r in results:
        print(
            f"{r['duration']:8d} | {100*r['geary_power']:6.1f}% | "
            f"{100*r['scan_power']:6.1f}% | {100*r['cusum_power']:6.1f}%"
        )
    print("(Scan score scales ~sqrt(duration) for fixed intensity, so it can")
    print(" lag CUSUM at very short durations -- that's expected, not a bug.)")
    return results

def run_correlated_burst_test(
    full_data, valid_tickers, W_matrix, raw_gearys, mean_map, std_map,
    ref_dates, test_dates, group_stocks=("NVDA", "AMD", "INTC", "QCOM", "TXN"),
    multipliers=(1.5, 2.0, 3.0, 4.0), burst_buckets=12,
    n_trial_days=15, seed=42, label="",
):
    rng = np.random.default_rng(seed)
    trial_dates = rng.choice(test_dates, size=min(n_trial_days, len(test_dates)), replace=False)
    group_stocks = [s for s in group_stocks if s in valid_tickers]

    results = []
    for mult in multipliers:
        flagged = 0
        for date in trial_dates:
            df_day = full_data[full_data["date"] == date]
            df_burst = inject_correlated_burst(df_day, group_stocks, burst_buckets, mult)
            obs_geary = geary_for_day(df_burst, valid_tickers, W_matrix, mean_map, std_map)
            g_p = empirical_p_value(obs_geary, [raw_gearys[d] for d in ref_dates], tail="lower")
            flagged += int(g_p < 0.05)
        n = len(trial_dates)
        results.append({"multiplier": mult, "geary_power": flagged / n})

    print(f"\n  [{label}] group={group_stocks}, burst_buckets={burst_buckets}, "
          f"{len(trial_dates)} trial days:")
    print(f"  {'mult':>6} | {'Geary power':>12}")
    for r in results:
        print(f"  {r['multiplier']:6.1f}x | {100*r['geary_power']:11.1f}%")
    return results

# ==========================================
# 8. MULTI-STOCK / MULTIPLE-TESTING CORRECTION
# ==========================================

def benjamini_hochberg(pvalues, alpha=0.05):
    """ Standard BH (1995) step-up FDR procedure. """
    pvalues = np.asarray(pvalues, dtype=float)
    n = len(pvalues)
    order = np.argsort(pvalues)
    ranked = pvalues[order]
    thresholds = (np.arange(1, n + 1) / n) * alpha
    below = ranked <= thresholds
    reject = np.zeros(n, dtype=bool)
    if below.any():
        max_idx = np.max(np.where(below)[0])
        reject[order[:max_idx + 1]] = True
    return reject

def run_multistock_heldout_scoring(test_dates, calib_dates, raw_scans, valid_tickers, alpha=0.05):
    """
    False-positive side: scores every stock on every held-out day, then
    applies BH-FDR using BOTH the floored empirical p-value and the
    Gumbel-refined one, so the resolution-floor problem is visible
    directly in the output instead of silently making "0 flags" look like
    a clean result.
    """
    raw_flag_count = fdr_emp_flag_count = fdr_gumbel_flag_count = total_tests = 0

    for date in test_dates:
        emp_pvals, gumbel_pvals = [], []
        for stock in valid_tickers:
            scan_obs = raw_scans[stock][date]
            scan_ref = [raw_scans[stock][d] for d in calib_dates]
            emp_pvals.append(empirical_p_value(scan_obs, scan_ref, tail="upper"))
            gumbel_pvals.append(gumbel_refine_pvalue(scan_obs, scan_ref, tail="upper"))

        emp_pvals = np.array(emp_pvals)
        gumbel_pvals = np.nan_to_num(np.array(gumbel_pvals), nan=1.0)

        raw_flag_count += int((emp_pvals < alpha).sum())
        fdr_emp_flag_count += int(benjamini_hochberg(emp_pvals, alpha=alpha).sum())
        fdr_gumbel_flag_count += int(benjamini_hochberg(gumbel_pvals, alpha=alpha).sum())
        total_tests += len(emp_pvals)

    print(f"\nMulti-stock scan detector, held-out window, "
          f"{len(valid_tickers)} tickers x {len(test_dates)} days:")
    print(f"  Raw p<0.05 flags:           {raw_flag_count}/{total_tests} ({100*raw_flag_count/total_tests:.1f}%)")
    print(f"  BH-FDR on empirical p:      {fdr_emp_flag_count}/{total_tests} ({100*fdr_emp_flag_count/total_tests:.1f}%)")
    print(f"  BH-FDR on Gumbel-refined p: {fdr_gumbel_flag_count}/{total_tests} ({100*fdr_gumbel_flag_count/total_tests:.1f}%)")

    floor_p = 1.0 / (len(calib_dates) + 1)
    rank1_thresh = (1.0 / len(valid_tickers)) * alpha
    print(f"  (Floored empirical p can never go below {floor_p:.4f}; BH's own")
    print(f"   rank-1 threshold across {len(valid_tickers)} tickers is {rank1_thresh:.4f}.")
    if floor_p > rank1_thresh:
        print(f"   {floor_p:.4f} > {rank1_thresh:.4f} -- the empirical p-value can structurally")
        print(f"   NEVER clear correction for a single flagged stock here, real or not.")
        print(f"   Only the Gumbel-refined column means anything for this basket size.)")

def run_multistock_fdr_power_test(
    full_data, valid_tickers, calib_dates, test_dates, raw_scans,
    mean_map, std_map, alpha=0.05, burst_multiplier=4.0, burst_buckets=12,
    n_trial_days=10, seed=42,
):
    """
    True-positive side of the same question: inject a real single-stock
    burst into ONE randomly chosen stock within the basket on each trial
    day, score ALL stocks that day, and check whether BH-FDR (on the
    Gumbel-refined p-value) still flags the manipulated one. "0 false
    alarms" on clean data only means something if paired with this.
    """
    rng = np.random.default_rng(seed)
    trial_dates = rng.choice(test_dates, size=min(n_trial_days, len(test_dates)), replace=False)
    caught = 0

    for date in trial_dates:
        burst_stock = rng.choice(valid_tickers)
        df_day = full_data[full_data["date"] == date]
        df_burst = inject_synthetic_burst(df_day, burst_stock, burst_buckets, burst_multiplier)

        pivot_lv = (
            df_burst.pivot(index="bucket", columns="stock", values="log_volume")
            .reindex(columns=valid_tickers)
            .sort_index()
        )

        gumbel_pvals = []
        for stock in valid_tickers:
            if stock == burst_stock:
                z_vals = get_standardized_series(pivot_lv, stock, mean_map, std_map)
                scan_obs = max_scan_stat(z_vals, max_window=12)
            else:
                scan_obs = raw_scans[stock][date]
            scan_ref = [raw_scans[stock][d] for d in calib_dates]
            gumbel_pvals.append(gumbel_refine_pvalue(scan_obs, scan_ref, tail="upper"))

        gumbel_pvals = np.nan_to_num(np.array(gumbel_pvals), nan=1.0)
        rejected = benjamini_hochberg(gumbel_pvals, alpha=alpha)
        burst_idx = valid_tickers.index(burst_stock)
        caught += int(rejected[burst_idx])
        print(f"{date}: injected into {burst_stock} -> "
              f"{'CAUGHT' if rejected[burst_idx] else 'missed'} after BH-FDR correction")

    n = len(trial_dates)
    print(f"\nTrue-positive rate under BH-FDR correction ({burst_multiplier}x burst, "
          f"{burst_buckets} buckets, 1 manipulated stock among {len(valid_tickers)}):")
    print(f"  {caught}/{n} ({100*caught/n:.1f}%) survived correction")

# ==========================================
# 9. MAIN
# ==========================================

if __name__ == "__main__":
    tickers = [
        "AAPL", "MSFT", "NVDA", "GOOG", "META", "AMD", "TSLA",
        "AMZN", "INTC", "NFLX", "CSCO", "QCOM", "TXN"
    ]

    full_data, valid_tickers = fetch_intraday_market_data(tickers, period="60d", interval="5m")

    daily_bucket_counts = full_data.groupby("date")["bucket"].nunique()
    valid_dates = daily_bucket_counts[daily_bucket_counts >= 70].index
    full_data = full_data[full_data["date"].isin(valid_dates)].copy()

    dates = sorted(full_data["date"].unique())
    print(f"Usable real trading days: {len(dates)}\n")

    if len(dates) < 20:
        raise ValueError("Not enough days after filtering to run a meaningful split test.")

    split_idx = int(0.7 * len(dates))
    calib_dates = dates[:split_idx]
    test_dates = dates[split_idx:]
    print(f"Calibration days: {len(calib_dates)}")
    print(f"Test days:        {len(test_dates)}\n")

    W_sector = create_sector_weight_matrix(valid_tickers)
    W_corr = create_correlation_weight_matrix(full_data, calib_dates, valid_tickers)

    mean_map, std_map = build_intraday_baselines(full_data, calib_dates, valid_tickers)

    # Main pipeline keeps W_sector for continuity with the original setup.
    raw_gearys, raw_scans, raw_cusums = build_daily_metrics(
        full_data, valid_tickers, W_sector, mean_map, std_map
    )
    # Separate Geary series for W_corr, just for the comparison sections below.
    raw_gearys_corr = {d: geary_for_day(full_data[full_data["date"] == d], valid_tickers, W_corr, mean_map, std_map) for d in dates}

    target_stock = "NVDA"

    print("=" * 70)
    print("LEAVE-ONE-OUT FALSE-POSITIVE CALIBRATION")
    print("=" * 70)
    summarize_false_positives(
        score_dates=calib_dates, ref_dates=calib_dates,
        raw_gearys=raw_gearys, raw_scans=raw_scans, raw_cusums=raw_cusums,
        target_stock=target_stock,
    )

    print("\n" + "=" * 70)
    print("HELD-OUT TEST WINDOW SCORING")
    print("=" * 70)
    heldout_geary_p, heldout_scan_p, heldout_cusum_p = {}, {}, {}
    for date in test_dates:
        g_obs, g_pval, scan_obs, scan_pval, cusum_obs, cusum_pval = score_day(
            date=date, ref_dates=calib_dates, raw_gearys=raw_gearys,
            raw_scans=raw_scans, raw_cusums=raw_cusums, target_stock=target_stock,
        )
        heldout_geary_p[date] = g_pval
        heldout_scan_p[date] = scan_pval
        heldout_cusum_p[date] = cusum_pval
        print(
            f"{date}: Geary={g_obs:.4f} p={g_pval:.4f} | "
            f"{target_stock} scan={scan_obs:.4f} p={scan_pval:.4f} | "
            f"CUSUM={cusum_obs:.4f} p={cusum_pval:.4f}"
        )

    n_test = len(test_dates)
    print("\n" + "=" * 70)
    print("HELD-OUT SUMMARY")
    print("=" * 70)
    print(f"Geary's C:    flagged {100*sum(p<0.05 for p in heldout_geary_p.values())/n_test:.1f}% of held-out days")
    print(f"{target_stock} scan:  flagged {100*sum(p<0.05 for p in heldout_scan_p.values())/n_test:.1f}% of held-out days")
    print(f"{target_stock} CUSUM:  flagged {100*sum(p<0.05 for p in heldout_cusum_p.values())/n_test:.1f}% of held-out days")

    print("\n" + "=" * 70)
    print("WEIGHT MATRIX DIAGNOSTIC: hand-built sector blocks vs realized correlation")
    print("=" * 70)
    fp_sector = sum(
        empirical_p_value(raw_gearys[d], [raw_gearys[x] for x in calib_dates if x != d], tail="lower") < 0.05
        for d in calib_dates
    ) / len(calib_dates)
    fp_corr = sum(
        empirical_p_value(raw_gearys_corr[d], [raw_gearys_corr[x] for x in calib_dates if x != d], tail="lower") < 0.05
        for d in calib_dates
    ) / len(calib_dates)
    print(f"  Sector-block W:      {100*fp_sector:.1f}% LOO false-positive rate")
    print(f"  Correlation-based W: {100*fp_corr:.1f}% LOO false-positive rate")

    print("\n" + "=" * 70)
    print("DETECTION POWER SWEEP -- burst STRENGTH, held-out days")
    print("=" * 70)
    run_power_sweep(
        full_data=full_data, valid_tickers=valid_tickers, W_matrix=W_sector,
        raw_gearys=raw_gearys, raw_scans=raw_scans, raw_cusums=raw_cusums,
        ref_dates=calib_dates, test_dates=test_dates, mean_map=mean_map, std_map=std_map,
        target_stock=target_stock, multipliers=(1.5, 2.0, 3.0, 4.0, 6.0, 8.0),
        burst_buckets=12, n_trial_days=15, seed=42,
    )

    print("\n" + "=" * 70)
    print("DETECTION POWER SWEEP -- burst DURATION, held-out days")
    print("=" * 70)
    run_duration_sweep(
        full_data=full_data, valid_tickers=valid_tickers, W_matrix=W_sector,
        raw_gearys=raw_gearys, raw_scans=raw_scans, raw_cusums=raw_cusums,
        ref_dates=calib_dates, test_dates=test_dates, mean_map=mean_map, std_map=std_map,
        target_stock=target_stock, burst_multiplier=4.0,
        durations=(3, 6, 12, 24, 48), n_trial_days=15, seed=42,
    )

    print("\n" + "=" * 70)
    print("DETECTION POWER -- COORDINATED basket burst (Geary's C alternative), now on")
    print("standardized volume -- sector-block W vs correlation-based W")
    print("=" * 70)
    run_correlated_burst_test(
        full_data=full_data, valid_tickers=valid_tickers, W_matrix=W_sector,
        raw_gearys=raw_gearys, mean_map=mean_map, std_map=std_map,
        ref_dates=calib_dates, test_dates=test_dates,
        group_stocks=("NVDA", "AMD", "INTC", "QCOM", "TXN"),
        multipliers=(1.5, 2.0, 3.0, 4.0), burst_buckets=12, n_trial_days=15, seed=42,
        label="sector-block W",
    )
    run_correlated_burst_test(
        full_data=full_data, valid_tickers=valid_tickers, W_matrix=W_corr,
        raw_gearys=raw_gearys_corr, mean_map=mean_map, std_map=std_map,
        ref_dates=calib_dates, test_dates=test_dates,
        group_stocks=("NVDA", "AMD", "INTC", "QCOM", "TXN"),
        multipliers=(1.5, 2.0, 3.0, 4.0), burst_buckets=12, n_trial_days=15, seed=42,
        label="correlation-based W",
    )

    print("\n" + "=" * 70)
    print("MULTI-STOCK SCORING WITH FDR CORRECTION")
    print("=" * 70)
    run_multistock_heldout_scoring(
        test_dates=test_dates, calib_dates=calib_dates,
        raw_scans=raw_scans, valid_tickers=valid_tickers, alpha=0.05,
    )
    run_multistock_fdr_power_test(
        full_data=full_data, valid_tickers=valid_tickers, calib_dates=calib_dates,
        test_dates=test_dates, raw_scans=raw_scans, mean_map=mean_map, std_map=std_map,
        alpha=0.05, burst_multiplier=4.0, burst_buckets=12, n_trial_days=10, seed=42,
    )
```
